```python
# Task 2.1

from pymongo import MongoClient
import json

client = MongoClient("127.0.0.1", 27017)

if "Vaccine" in client.list_database_names():
    client.drop_database("Vaccine")
    
db = client["Vaccine"]
coll = db["Staff"]

json_file = open("STAFF.json", 'r')
json_data = json.load(json_file)

result = coll.insert_many(json_data)
print(result)
```

```python
# Task 2.2

from pprint import pprint

query1 = {"Acceptance": {"$eq": "No"}}
proj1 = {"_id": 0, "Name": 1, "Age": 1, "Reason": 1}

print(f"{"Name":<12}{"Age":<6}Reason")
print("-"*50)

for staff in coll.find(query1, proj1):
    print(f"{staff["Name"]:<12}{staff["Age"]:<6}{staff["Reason"]}")
```

```python
query2 = {"Acceptance": {"$eq": "Yes"}, "VacDates": {"$exists": True}}
print("Number of staff with at least one dose:", coll.count_documents(query2))
```

```python
query3 = {"Name": {"$eq": "Charlie Lee"}}

original_date = coll.find_one(query3)["VacDates"]
new_dates = [original_date, "5 Apr 2021"]
update = {"$set": {"VacDates": new_dates}}

result = coll.update_one(query3, update)
```

```python
query4 = {"Age": {"$gte": 50}, "VacDates": {"$exists": True}}

print("Staff of at least 50 years old and completed both doses:")
for staff in coll.find(query4):
    if type(staff["VacDates"]) == list and len(staff["VacDates"]) == 2:
        print(staff["Name"])
```
