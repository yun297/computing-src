```python
from socket import socket

listener = socket()
listener.bind(("127.0.0.1", 12345))
listener.listen()

server, addr = listener.accept()
print("Connected to:", addr)

nugget_count = int(input("How many nuggets? "))

while True:
    if nugget_count > 0:
        server.sendall(str(nugget_count).encode() + b'\n')
    else:
        server.sendall(b'end\n')
        print("You lose.")
        break

    message = ""

    while '\n' not in message:
        message += server.recv(1024).decode()

    if message.strip() == "end":
        print("You win.")
        break
    
    nugget_count = int(message.strip())

    print(f"There are {nugget_count} nuggets left.")
    take_nuggets = int(input("How many to take? "))

    while take_nuggets > nugget_count or take_nuggets not in [1, 2, 3]:
        print(f"There are {nugget_count} nuggets left.")
        take_nuggets = int(input("How many to take? "))

    nugget_count -= take_nuggets
    
    if nugget_count > 0:
        print(f"{take_nuggets} taken, {nugget_count} left.")
```
