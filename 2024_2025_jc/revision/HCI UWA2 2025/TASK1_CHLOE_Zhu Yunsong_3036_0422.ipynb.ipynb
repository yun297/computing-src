```python
from socket import socket

client = socket()
client.connect(("127.0.0.1", 12345))

while True:
    message = ""

    while '\n' not in message:
        message += client.recv(1024).decode()

    if message.strip() == "end":
        print("You win.")
        break
    
    nugget_count = int(message.strip())

    print(f"There are {nugget_count} nuggets left.")
    take_nuggets = int(input("How many to take? "))

    while take_nuggets > nugget_count or take_nuggets not in [1, 2, 3]:
        print(f"There are {nugget_count} nuggets left.")
        take_nuggets = int(input("How many to take? "))

    nugget_count -= take_nuggets
    
    if nugget_count > 0:
        print(f"{take_nuggets} taken, {nugget_count} left.")

    if nugget_count > 0:
        client.sendall(str(nugget_count).encode() + b'\n')
    else:
        client.sendall(b'end\n')
        print("You lose.")
        break
```
