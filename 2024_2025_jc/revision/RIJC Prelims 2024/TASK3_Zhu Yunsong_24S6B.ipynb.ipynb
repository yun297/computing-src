```python
# Task 3.1

class Person:
    def __init__(self, name: str, personID: str):
        self.name = name
        self.personID = personID

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}"


class Student(Person):
    def __init__(self, name: str, personID: str, course: str, year: int):
        super().__init__(name, personID)
        self.course = course
        self.year = year

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}, Course: {self.course}, Year: {self.year}"


class Staff(Person):
    def __init__(self, name: str, personID: str, department: str, position: str, salary: int):
        super().__init__(name, personID)
        self.department = department
        self.position = position
        self.salary = salary

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}, Department: {self.department}, Position: {self.position}, Salary: {self.salary}"
```
