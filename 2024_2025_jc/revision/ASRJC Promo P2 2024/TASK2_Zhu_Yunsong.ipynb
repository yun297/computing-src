```python
# Task 2.1

class Soldier:
    def __init__(self, name: str, x: int, y: int):
        self.name = name
        self.x_pos = x
        self.y_pos = y

    def get_x(self):
        return self.x_pos

    def get_y(self):
        return self.y_pos

    def increment_x(self, value: int):
        new_x = self.x_pos + value

        if new_x > 5000:
            self.x_pos = 5000
        elif new_x < 0:
            self.x_pos = 0
        else:
            self.x_pos = new_x

    def increment_y(self, value: int):
        new_y = self.y_pos + value

        if new_y > 5000:
            self.y_pos = 5000
        elif new_y < 0:
            self.y_pos = 0
        else:
            self.y_pos = new_y

    def move(self, direction: str):
        if direction == "up":
            self.increment_y(5)
        elif direction == "down":
            self.increment_y(-5)
        elif direction == "left":
            self.increment_x(-5)
        elif direction == "right":
            self.increment_x(5)
        else:
            print("Invalid direction")
```

```python
# Task 2.2

class Spartan(Soldier):
    def __init__(self, name: str, x: int, y: int, skills: list):
        super().__init__(name, x, y)
        self.skills = skills

    def display_skills(self):
        print("Skills:", self.skills)

    def move(self, direction: str):
        if direction == "up":
            self.increment_y(9)
        elif direction == "down":
            self.increment_y(-9)
        elif direction == "left":
            self.increment_x(-8)
        elif direction == "right":
            self.increment_x(8)
        else:
            print("Invalid direction")
```

```python
# Task 2.3

jeremy = Spartan("Jeremy", 888, 999, ["swordsmanship", "archery", "spear throwing"])

jeremy.display_skills()

direction = input('Enter a direction to move in ("up", "down", "left", "right"): ')
jeremy.move(direction)

print(f"{jeremy.name}'s position is ({jeremy.get_x()}, {jeremy.get_y()})")
```
