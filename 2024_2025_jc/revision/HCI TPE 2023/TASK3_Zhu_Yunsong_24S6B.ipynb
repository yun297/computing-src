```python
# Task 3

class WordRank:
    def __init__(self, data: tuple):
        self.left = None
        self.right = None
        self.data = data # (word, rank)

class Tree:
    def __init__(self):
        self.root = None

    def insert(self, new: WordRank):
        if not self.root:
            self.root = new
            return
            
        ptr = self.root
        inserted = False

        while not inserted:
            if new.data[1] > ptr.data[1]:
                if ptr.right is None:
                    ptr.right = new
                    inserted = True
                else:
                    ptr = ptr.right
            else:
                if ptr.left is None:
                    ptr.left = new
                    inserted = True
                else:
                    ptr = ptr.left

    def inorder(self, root):
        if root:
            self.inorder(root.left)
            print(root.data)
            self.inorder(root.right)

    def preorder(self, root):
        if root:
            print(root.data[0])
            self.preorder(root.left)
            self.preorder(root.right)
```

```python
from csv import reader

words_file = open("WORDS.csv", "r")
words_file_array = []
words = []

for line in reader(words_file):
    words_file_array.append(line)

for item in words_file_array[1:]:
    word = item[0]
    rank = int(item[1])

    words.append((word, rank))

print(words)
```

```python
word_tree = Tree()

for word in words:
    word_rank = WordRank(word)

    word_tree.insert(word_rank)
```

```python
word_tree.inorder(word_tree.root)
```

```python
word_tree.preorder(word_tree.root)
```
