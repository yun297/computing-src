```python
# Task 1.1

f = open("votes.txt", 'r')

votes = {}
results = []

for line in f:
    _, artist = line.strip().split(' ')
    if artist not in votes:
        votes[artist] = 1
    else:
        votes[artist] += 1

for artist, vote in votes.items():
    results.append([artist, vote])
    
f.close()

print(results)
```

```python
# Task 1.2

def insertion_sort(arr):
    for i in range(1, len(arr)):
        space = i
        temp = arr[space]

        while space > 0 and arr[space - 1][1] < temp[1]:
            arr[space] = arr[space - 1]
            space -= 1

        arr[space] = temp

insertion_sort(results)
print(results)
```

```python
highest_vote = results[0][1]

print("Artist(s) with highest number of votes:")

for artist in results:
    if artist[1] == highest_vote:
        print(artist[0])
```

```python
# Task 1.3

def task1_3(arr, t):
    def binary(arr, t):
        low, high = 0, len(arr) - 1
        
        while low <= high:
            mid = (low + high) // 2
            if arr[mid][1] == t:
                return mid
            elif arr[mid][1] < t: # array sorted descending, so go left
                high = mid - 1
            else: # arr[mid][1] > t
                low = mid + 1
        return -1 # not found
    
    index = binary(arr, t)
    if index == -1:
        return [] # no artist found with t votes
    
    # If found, expand left and right to find all artists with same vote count
    result = []
    
    # Expand left
    i = index
    while i >= 0 and arr[i][1] == t:
        result.append(arr[i][0])
        i -= 1
    
    # Expand right
    i = index + 1
    while i < len(arr) and arr[i][1] == t:
        result.append(arr[i][0])
        i += 1
    
    return result

print(task1_3(results, 16))
print(task1_3(results, 2))
```
