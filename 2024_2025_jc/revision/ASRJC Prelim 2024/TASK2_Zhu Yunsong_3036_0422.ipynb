```python
# Task 2.1

class LinkedListNode:
    def __init__(self, player_id):
        self.player_id = player_id
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, player_id):
        """Insert at head (stack behaviour)."""
        new_node = LinkedListNode(player_id)
        new_node.next = self.head
        self.head = new_node

class Stack:
    def __init__(self, max_size=10):
        self.top = None
        self.size = 0
        self.max_size = max_size

    def push(self, player_id):
        if self.size >= self.max_size:
            print("Stack overflow – cannot add more players.")
            return
        new_node = LinkedListNode(player_id)
        new_node.next = self.top
        self.top = new_node
        self.size += 1

    def pop(self):
        if self.top is None:
            print("Stack underflow – empty stack.")
            return None
        popped = self.top.player_id
        self.top = self.top.next
        self.size -= 1
        return popped

    def display(self):
        temp = self.top
        items = []
        while temp:
            items.append(temp.player_id)
            temp = temp.next
        print("Stack (top → bottom):", items)

class Queue:
    def __init__(self, max_size=10):
        self.front = 0
        self.rear = -1
        self.size = 0
        self.max_size = max_size

    def enqueue(self, player_id):
        if self.size >= self.max_size:
            print("Queue overflow – cannot add more players.")
            return
            
        self.data[self.rear] = player_id
        self.rear = (self.rear + 1) % self.max_size
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            print("Queue underflow – empty queue.")
            return None
        player_id = self.data[self.front]
        self.front = (self.front + 1) % self.max_size
        self.size -= 1
        return player_id

    def display(self):
        if self.size == 0:
            print("Queue is empty.")
            return
        items = []
        idx = self.front
        for _ in range(self.size):
            items.append(self.data[idx])
            idx = (idx + 1) % self.max_size
        print("Queue (front → rear):", items)

class BSTNode:
    def __init__(self, ds, score):
        self.ds = ds              # 'S' for stack, 'Q' for queue
        self.score = score        # game score (BST key)
        self.data = Stack() if ds == 'S' else Queue()
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    def insert(self, ds, score, player_id):
        """Insert player into BST by score. 
        If score exists, add to its stack/queue."""
        def _insert(root, ds, score, player_id):
            if root is None:
                node = BSTNode(ds, score)
                node.data.push(player_id) if ds == 'S' else node.data.enqueue(player_id)
                return node

            if score < root.score:
                root.left = _insert(root.left, ds, score, player_id)
            elif score > root.score:
                root.right = _insert(root.right, ds, score, player_id)
            else:  # score already exists
                if root.ds == 'S':
                    root.data.push(player_id)
                else:
                    root.data.enqueue(player_id)
            return root

        self.root = _insert(self.root, ds, score, player_id)

    def search(self, score):
        """Search node by score."""
        current = self.root
        while current:
            if current.score == score:
                return current
            elif score < current.score:
                current = current.left
            else:
                current = current.right
        return None

    def inorder(self):
        """Inorder traversal (ascending by score)."""
        def _inorder(node):
            if node:
                _inorder(node.left)
                print(f"Score {node.score} ({'Stack' if node.ds == 'S' else 'Queue'}) → ", end="")
                node.data.display()
                _inorder(node.right)
        _inorder(self.root)
```
