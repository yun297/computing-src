```python
# Task 3.1

class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def insert(self, word, p):
        new_node = Node(word)

        # Clamp p to range [1, size + 1]
        if p > self.size:
            p = self.size + 1

        if p == 1:
            new_node.next = self.head
            self.head = new_node
        else:
            count = 1
            ptr = self.head
            while count < p - 1:
                ptr = ptr.next
                count += 1
            new_node.next = ptr.next
            ptr.next = new_node

        self.size += 1

    def delete(self, p):
        if self.head is None or self.size == 0:
            return

        if p > self.size:
            p = self.size

        if p == 1:
            self.head = self.head.next
        else:
            count = 1
            ptr = self.head
            while count < p - 1:
                ptr = ptr.next
                count += 1
            ptr.next = ptr.next.next

        self.size -= 1

    def search(self, word):
        ptr = self.head
        while ptr is not None:
            if ptr.data == word:
                return True
            ptr = ptr.next
        return False

    def to_String(self):
        elements = []
        ptr = self.head
        while ptr is not None:
            elements.append(ptr.data)
            ptr = ptr.next
        return ", ".join(elements)
```

```python
linked_list = LinkedList()

# inserting at the start
linked_list.insert("apple", 1)
print(linked_list.to_String())

# inserting at the end where p > size of linked list
linked_list.insert("banana", 3)
print(linked_list.to_String())

# inserting in the middle
linked_list.insert("orange", 2)
print(linked_list.to_String())

# inserting at the end where p > size of linked list
linked_list.insert("grape", 5)
print(linked_list.to_String())

# deleting a node
linked_list.delete(2)
print(linked_list.to_String())

# searching for item
print(linked_list.search("grape"))
```

```python
# Task 3.2

class Stack(LinkedList):
    def __init__(self):
        super().__init__()

    def push(self, word):
        self.insert(word, 1)

    def pop(self):
        self.delete(1)
```

```python
stack = Stack()

stack.push("apple")
print(stack.to_String())

stack.push("pear")
print(stack.to_String())

stack.push("carrot")
print(stack.to_String())

stack.pop()
print(stack.to_String())
```

```python
# Task 3.3

class Queue(LinkedList):
    def __init__(self):
        super().__init__()

    def enqueue(self, word):
        self.insert(word, self.size + 1)

    def dequeue(self):
        self.delete(1)
```

```python
queue = Queue()

queue.enqueue("apple")
print(queue.to_String())

queue.enqueue("pear")
print(queue.to_String())

queue.enqueue("carrot")
print(queue.to_String())

queue.dequeue()
print(queue.to_String())
```
