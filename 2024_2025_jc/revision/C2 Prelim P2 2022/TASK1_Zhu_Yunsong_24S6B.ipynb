```python
# Task 1.1

#file handling
minefield_txt = open('minefield.txt', 'r')
minefield_list = []

for line in minefield_txt:
    minefield_list.append(line.strip())

# print(minefield_list)

n = int(minefield_list[0])

# create minefield
minefield = []

for i in range(n):
    row = []
    for j in range(n):
        row.append('.')
    minefield.append(row)

# store positions
minefield[n//2][n//2] = 'S'

for coord in minefield_list[1:]:
    y = int(coord[0])
    x = int(coord[2])
    minefield[y][x] = 'M'

# display minefield
def display():
    for row in minefield:
        for coloumn in row:
            print(f'{coloumn:^3}', end = '')
        print()

display()
```

```python
# Task 1.2

from random import choice

solder_coords = (n//2, n//2)
mine = False
won = False

print('STEPS: ', end = '')

while not mine and not won:
    next_coords = (0, 0)
    direction = choice(['U', 'D', 'L', 'R']) # U is up, D is down, L is left, R is right
    
    if direction == 'U':
        print('UP ', end = '')
        next_coords = (solder_coords[0] - 1, solder_coords[1])
    elif direction == 'D':
        print('DOWN ', end = '')
        next_coords = (solder_coords[0] + 1, solder_coords[1])
    elif direction == 'L':
        print('LEFT ', end = '')
        next_coords = (solder_coords[0], solder_coords[1] - 1)
    elif direction == 'R':
        print('RIGHT ', end = '')
        next_coords = (solder_coords[0], solder_coords[1] + 1)

    if minefield[next_coords[0]][next_coords[1]] == 'M':
        mine = True
        
    minefield[next_coords[0]][next_coords[1]] = 'S'
    minefield[solder_coords[0]][solder_coords[1]] = 'P'
    
    solder_coords = next_coords

    if (solder_coords[0] == 0 or solder_coords[1] == 0) or (solder_coords[0] == n - 1 or solder_coords[1] == n - 1):
        won = True

print()
display()

if mine:
    print('LOST! You stepped in a mine!')
elif won:
    print('WIN! You walked to the boundary!')
    
```
