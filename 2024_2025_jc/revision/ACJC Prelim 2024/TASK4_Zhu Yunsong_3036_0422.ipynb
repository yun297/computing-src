```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("SUPERMARKET.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS Item (
    ItemID INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Price INTEGER NOT NULL
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Customer (
    CustomerNumber INTEGER PRIMARY KEY,
    Name TEXT NOT NULL
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Purchase (
    PurchaseID INTEGER PRIMARY KEY,
    CustomerNumber INTEGER REFERENCES Customer(CustomerNumber),
    ItemID INTEGER REFERENCES Item(ItemID),
    Quantity INTEGER NOT NULL,
    DateTime TEXT NOT NULL
)
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

item_f = open("ITEM.txt", 'r')
customer_f = open("CUSTOMER.txt", 'r')
purchase_f = open("PURCHASE.txt", 'r')

item_data = []
customer_data = []
purchase_data = []

for line in item_f:
    item_data.append(line.strip().split(','))

for line in customer_f:
    customer_data.append(line.strip().split(','))

for line in purchase_f:
    purchase_data.append(line.strip().split(','))

item_f.close()
customer_f.close()
purchase_f.close()

conn = sqlite3.connect("SUPERMARKET.db")
cur = conn.cursor()

cur.executemany("""
INSERT INTO Item (ItemID, Name, Price) VALUES (?, ?, ?)
""", item_data)

cur.executemany("""
INSERT INTO Customer (CustomerNumber, Name) VALUES (?, ?)
""", customer_data)

cur.executemany("""
INSERT INTO Purchase (PurchaseID, CustomerNumber, ItemID, Quantity, DateTime) VALUES (?, ?, ?, ?, ?)
""", purchase_data)

conn.commit()
conn.close()
```

```python
# Task 4.3

conn = sqlite3.connect("SUPERMARKET.db")
cur = conn.cursor()

customer_number = int(input("Enter a customer number: "))
date = input("Enter a date: ")

cur.execute("""
SELECT i.Name, p.Quantity FROM Purchase AS p
INNER JOIN Item as i
ON i.ItemID = p.ItemID
WHERE p.DateTime >= ? AND p.CustomerNumber = ?
""", (date, customer_number))

data = cur.fetchall()

for r in data:
    print(r)
```
