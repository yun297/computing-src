```python
# Task 3.1

class Node:
    def __init__(self, data: tuple, pointer = None):
        self.data = data
        self.pointer = pointer

class LinkedList:
    def __init__(self):
        self.headpointer = None

    def insert_head(self, data: tuple):
        new = Node(data)

        new.pointer = self.headpointer
        self.headpointer = new

    def display(self):
        if self.headpointer:
            ptr = self.headpointer
    
            while ptr:
                print(ptr.data)
                ptr = ptr.pointer

    def search(self, name: str):
        ptr = self.headpointer

        while ptr:
            if ptr.data[0] == name:
                return True
            ptr = ptr.pointer

        return False

    def remove(self, name: str):
        if self.headpointer:
            if self.headpointer.data[0] == name:
                self.headpointer = self.headpointer.pointer
                return True

            ptr = self.headpointer

            while ptr and ptr.pointer.data[0] != name:
                ptr = ptr.pointer

            if ptr.pointer:
                ptr.pointer = ptr.pointer.pointer
                return True
                
            return False

    def reverse(self):
        prev = None
        ptr = self.headpointer

        while ptr:
            next_node = ptr.pointer
            ptr.pointer = prev
            prev = ptr
            ptr = next_node

        self.headpointer = prev

    def get(self, name: str):
        ptr = self.headpointer
        
        while ptr:
            if ptr.data[0] == name:
                return ptr.data[1]
            ptr = ptr.pointer
            
        return False
```

```python
# Task 3.2

l = LinkedList()

f = open("names.txt", 'r')

names = []
for line in f:
    name, score = line.strip().split(',')
    names.append((name, int(score)))
print(names)

f.close()

for n in names:
    l.insert_head(n)

l.display()
l.reverse()
print()
l.display()
```

```python
# Task 3.3

class HashTable:
    def __init__(self):
        self.max_size = 97
        self.table = []

        for i in range(self.max_size):
            self.table.append(LinkedList())

    def hash_name(self, name: str):
        ascii_sum = 0

        for c in name:
            ascii_sum += ord(c)

        return ascii_sum % 97

    def add(self, tup: tuple):
        self.table[self.hash_name(tup[0])].insert_head(tup)

    def retrieve(self, name: str):
        index = self.hash_name(name)
        return self.table[index].get(name)

    def show(self):
        for i in range(len(self.table)):
            print(i)
            self.table[i].display()

h = HashTable()
f = open("names.txt", 'r')

names = []
for line in f:
    name, score = line.strip().split(',')
    names.append((name, int(score)))

f.close()

for n in names:
    h.add(n)

h.show()
```
