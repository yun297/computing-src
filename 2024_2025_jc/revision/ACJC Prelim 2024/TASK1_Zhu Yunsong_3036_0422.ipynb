```python
# Task 1.1

def string_value(s1):
    vowel_count = 0

    for c in s1.lower():
        if c in ['a', 'e', 'i', 'o', 'u']:
            vowel_count += 1

    return vowel_count * len(s1)
```

```python
# Task 1.2

def StalinSort(lst):
    def compare(a, b):
        va, vb = string_value(a), string_value(b)

        if va != vb:
            return va <= vb
        return a <= b

    moved = True
    while moved:
        moved = False
        i = 0
        while i < len(lst) - 1:
            if not compare(lst[i], lst[i + 1]):
                popped = lst.pop(i + 1)
                lst.insert(0, popped)
                moved = True
            else:
                i += 1

    return lst

print(StalinSort(["Peter", "Josh", "See", "Hi"]))
print(StalinSort(["PetEr", "JOsh", "Hi", "See"]))
print(StalinSort(["Peter", "Hello", "BYE", "Hi", "ah", "bbbb"]))
```

```python
# Task 1.3

f = open("words.txt", 'r')

lst = []

for line in f:
    lst.append(line.strip())

print(StalinSort(lst))
```

```python
# Task 1.4

def QuickSort(lst):
    if len(lst) <= 1:
        return lst

    pivot = lst[0]
    pivot_val = string_value(pivot)
    left = []
    right = []

    for item in lst[1:]:
        val = string_value(item)
        if val < pivot_val:
            left.append(item)
        elif val > pivot_val:
            right.append(item)
        else:
            if item.lower() < pivot.lower():
                left.append(item)
            else:
                right.append(item)

    return QuickSort(left) + [pivot] + QuickSort(right)

QuickSort(lst)
```

```python
# Task 1.5

def BinarySearch(lst, value):
    low = 0
    high = len(lst) - 1

    while high >= low:
        mid = (low + high) // 2
        mid_val = string_value(lst[mid])

        if mid_val == value:
            return mid
        elif mid_val > value:
            high = mid - 1
        elif mid_val < value:
            low = mid + 1

    return False
```
