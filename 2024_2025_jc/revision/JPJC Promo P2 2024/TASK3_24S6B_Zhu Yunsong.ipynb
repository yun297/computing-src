```python
# Task 3.1

class Node:
    def __init__(self):
        self.title = ""
        self.artiste = ""
        self.next = None
```

```python
# Task 3.2

class LinkedList:
    def __init__(self):
        self.head = None

    def addNode(self, new_node: Node):
        if not self.head:
            self.head = new_node
            return 

        ptr = self.head

        while ptr.next:
            ptr = ptr.next

        ptr.next = new_node

    def removeNode(self, title: str):
        if self.head:
            if self.head.title == title:
                self.head = self.head.next
                return

            ptr = self.head

            while ptr.next and ptr.next.title != title:
                ptr = ptr.next

            if ptr.next.title == title:
                ptr.next = ptr.next.next

    def display(self):
        ptr = self.head

        while ptr:
            print(ptr.title, ptr.artiste)
            ptr = ptr.next
```

```python
# Task 3.3

linkedlist = LinkedList()

node1 = Node()
node1.title = "24K Magic"
node1.artiste = "Bruno Mars"

node2 = Node()
node2.title = "Set Fire to the Rain"
node2.artiste = "Adele"

node3 = Node()
node3.title = "Just the Way You Are"
node3.artiste = "Bruno Mars"

node4 = Node()
node4.title = "Treasure"
node4.artiste = "Bruno Mars"

linkedlist.addNode(node1)
linkedlist.addNode(node2)
linkedlist.addNode(node3)
linkedlist.removeNode("Just the Way You Are")
linkedlist.addNode(node4)
linkedlist.removeNode("24K Magic")
linkedlist.display()
```
