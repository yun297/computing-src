```python
# Task 1.1

def read_books_data():
    books_txt = open("TASK1.txt", 'r')
    books = []

    for line in books_txt:
        book = line.strip().split(',')
        books.append([book[0], book[1], float(book[2])])

    books_txt.close()
    
    return books

read_books_data()
```

```python
# Task 1.2

def get_books_by_genre(books, genre):
    result = []

    for book in books:
        if book[1] == genre:
            result.append(book[0])

    return result

get_books_by_genre(read_books_data(), "Fiction")
```

```python
# Task 1.3

def get_expensive_books(books, price_threshold):
    result = []

    for book in books:
        if book[2] >= price_threshold:
            result.append(book[0])

    return result

get_expensive_books(read_books_data(), 15.99)
```

```python
# Task 1.4

print("""
1) Display Books by Genre
2) Display Books by Price
3) Quit
""")

option = input("Enter your option (1, 2 or 3): ")

while option != "3":
    while option not in ['1', '2', '3']:
        option = input("Invalid option! Enter your option (1, 2 or 3): ")
    
    if option == '1':
        genre = input("Enter the genre: ")
        print(get_books_by_genre(read_books_data(), genre))
    elif option == '2':
        threshold = float(input("Enter the price: "))
        print(get_expensive_books(read_books_data(), threshold))

    option = input("Enter your option (1, 2 or 3): ")
```
