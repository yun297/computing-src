```python
# Task 4.1

def up_to_two_digit(num: int):
    ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]
    tens = ["ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen", "seventeen", "eighteen", "nineteen"]
    twenties = ["twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]

    if num <= 9:
        return ones[num]

    if num >= 10 and num <= 19:
        return tens[num%10]

    if num >= 20 and num <= 99:
        return twenties[num//10 - 2] + " " + ones[num%10]

write = ""
for i in range(1, 100):
    write += up_to_two_digit(i)
    write += "\n"

file = open("TASK4_NUMBERS_Zhu_Yunsong_24S6B.txt", "w")
file.write(write)
file.close()
```

```python
# Task 4.2

def up_to_three_digit(num: int):
    if num <= 99:
        return up_to_two_digit(num)
        
    hundreds_string = up_to_two_digit(num//100)
    return hundreds_string + " hundred " + up_to_two_digit(num%100)

print(up_to_three_digit(60))
print(up_to_three_digit(700))
print(up_to_three_digit(890)) 
```

```python
# Task 4.3

def convert(money: float):
    money_string = str(money)
    dollars, cents = money_string.split(".")
    dollars = int(dollars)
    cents = int(cents)
    
    billions = dollars//1000000000
    millions = (dollars%1000000000)//1000000
    thousands = ((dollars%1000000000)%1000000)//1000
    ones = dollars%1000

    # number form
    number = f"{billions},{millions:03d},{thousands:03d},{ones:03d}"
    number_array = []

    flag = False
    
    for item in number.split(","):
       if flag:
           number_array.append(item)
       elif int(item) != 0:
           number_array.append(item)
           flag = True

    number_array[0] = number_array[0].lstrip("0")

    number_form = ",".join(number_array)
    number_form += f".{cents:<2}".replace(" ", "0")
    
    print(number_form)

    # word form
    word_form = "Word Form: "

    if billions != 0: word_form += f"{up_to_three_digit(billions)} billion "
    if millions != 0: word_form += f"{up_to_three_digit(millions)} million "
    if thousands != 0: word_form += f"{up_to_three_digit(thousands)} thousand "
    if ones != 0: word_form += f"{up_to_three_digit(ones)} "

    word_form += "dollars"
    
    if cents != 0: word_form += f" {up_to_three_digit(cents)} cents "

    print(word_form)

convert(1.00)
convert(1234000567.89)
```
