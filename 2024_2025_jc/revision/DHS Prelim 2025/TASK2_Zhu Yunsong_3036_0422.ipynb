```python
# Task 2.1

from random import sample, randint

class GuessingPairGame:
    def __init__(self):
        self.board = self.create_board()
        self.revealed = self.create_revealed_board()
        self.player_score = 0
        self.computer_score = 0
    
    def create_board(self):
        board = []
        for row_index in range(2):
            row = []
            for column_index in range(4):
                row.append(-1)
            board.append(row)
        return board
    
    def create_revealed_board(self):
        revealed = []
        for row_index in range(2):
            row = []
            for column_index in range(4):
                row.append(False)
            revealed.append(row)
        return revealed
    
    #code the below function
    def populate_board(self):
        numbers = sample(range(1, 10), k = 4) * 2
        
        for n in numbers:
            placed = False

            while not placed:
                row = randint(0, 1)
                col = randint(0, 3)
                
                if self.board[row][col] == -1:
                    self.board[row][col] = n
                    placed = True
    
    def set_board(self, values):
        values = values.split(",") 
        receivedboard = []
        index = 0
        for i in range(2):
            row = []
            for j in range(4):
                row.append(int(values[index])) 
                index = index + 1
            receivedboard.append(row)
        self.board = receivedboard
    
    def print_board(self):
        print("\nBoard:")
        for row_index in range(2):
            row_display = []
            for column_index in range(4):
                if self.revealed[row_index][column_index]:
                    row_display.append(" " + str(self.board[row_index][column_index]) + " ")
                else:
                    row_display.append(" X ")
            print("|".join(row_display))
        print()
        
    def board_in_string(self):
        result_string = ""
        for row in self.board:
            for value in row:
                result_string = result_string + str(value) + ","
        result_string = result_string[:-1]
        return result_string
    
    def valid_choice(self, row, column):
        valid = False
        if row >= 0 and row < 2 and column >= 0 and column < 4:
            if self.revealed[row][column] == False:
                valid = True
        return valid
    
    #code the below function
    def play_move(self, row1, column1, row2, column2):
        if self.board[row1][column1] == self.board[row2][column2]:
            self.revealed[row1][column1] = True
            self.revealed[row2][column2] = True
            
            return self.board[row1][column1]
            
        else:
            return 0
    
    #code the below function  
    def is_game_over(self):
        over = True

        for row in self.board:
            for col in self.board:
                if not col:
                    over = False

        return over
```

```python
# Task 2.2

import socket 
# from GuessingPairGame import GuessingPairGame

print("-------------------")
print("GUESSING PAIR GAME")
print("-------------------")
print()

game = GuessingPairGame()

#type the rest of your code below

client = socket.socket()
client.connect(("127.0.0.1", 9876))

received_board = client.recv(1024).decode().strip()
game.set_board(received_board)

while True:
    # First cell input
    cell_1_valid = False
    while not cell_1_valid:
        row_1 = int(input("INPUT first coordinates row number: "))
        col_1 = int(input("INPUT first coordinates column number: "))
        if game.valid_choice(row_1, col_1):
            cell_1_valid = True

    print()
    
    # Second cell input
    cell_2_valid = False
    while not cell_2_valid:
        row_2 = int(input("INPUT second coordinates row number: "))
        col_2 = int(input("INPUT second coordinates column number: "))
        if game.valid_choice(row_2, col_2) and not (row_2 == row_1 and col_2 == col_1):
            cell_2_valid = True

    print()
    
    score = game.play_move(row_1, col_1, row_2, col_2)

    if score > 0:
        print(f"Its a match! You scored {score} points")
    else:
        print("No match!")

    game.player_score += score

    print(f"Player Score: {game.player_score}")
    print(f"Computer Score: {game.computer_score}")

    game.print_board()

    coordinates = f"{str(row_1)},{str(col_1)},{str(row_2)},{str(col_2)}"
    client.sendall(coordinates.encode())
    
    if game.is_game_over():
        break

    # lower cycle
    
    coordinates = client.recv(1024).decode()

    row_1, col_1, row_2, col_2 = coordinates.split(",")
    print(f"Computer chose: row 1: {row_1}, col 1: {col_1}, row 2: {row_2}, col 2: {col_2}")
    score = game.play_move(int(row_1), int(col_1), int(row_2), int(col_2))

    score = game.play_move(row_1, col_1, row_2, col_2)

    if score > 0:
        print(f"Its a match! Computer scored {score} points")
    else:
        print("No match!")

    game.computer_score += score

    print(f"Player Score: {game.player_score}")
    print(f"Computer Score: {game.computer_score}")
    
    game.print_board()
    
    if game.is_game_over():
        break

if game.player_score > game.computer_score:
    print("Player win!")
else:
    print("Player lose!")
    
client.close()

print("-------------------")
print("GAME ENDED")
print("-------------------")
```
