```python
# Task 1.1

from random import randint

def generate_hand():
    result = []

    for _ in range(10):
        result.append((
            ["red", "green", "blue"][randint(0, 2)],
            randint(1, 4)
        ))

    return result

player1 = generate_hand()
player2 = generate_hand()

print("Player 1's hand:", player1)
print("Player 2's hand:", player2)
```

```python
# Task 1.2

def compare_colour(colour1, colour2):
    if colour1 == colour2:
        return '3'
    elif colour1 == "red" and colour2 == "green":
        return '1'
    elif colour1 == "green" and colour2 == "blue":
        return '1'
    elif colour1 == "blue" and colour2 == "red":
        return '1'
    else:
        return '2'
```

```python
# Task 1.3

def sort_hand(arr):
    colour_order = ["red", "green", "blue"]
    
    n = len(arr)

    while n > 1:
        last_swap = 1 
        
        for i in range(1, n):
            if arr[i - 1][1] == arr[i][1]:
                if colour_order.index(arr[i - 1][0]) > colour_order.index(arr[i][0]):
                    arr[i - 1], arr[i] = arr[i], arr[i - 1]
                    last_swap = i
            elif arr[i - 1][1] > arr[i][1]:
                arr[i - 1], arr[i] = arr[i], arr[i - 1]
                last_swap = i

        n = last_swap

    return arr

print('Player 1 cards')
print(player1)
print('Player 1 cards after sorting')
sort_hand(player1)
print(player1)
print()

print('Player 2 cards')
print(player2)
print('Player 2 cards after sorting')
sort_hand(player2)
print(player2)
print()
```

```python
# Task 1.4

def calculate_score(player1, player2):
    score1 = 0
    score2 = 0

    for i in range(10):
        card1 = player1[i]
        card2 = player2[i]

        if compare_colour(card1[0], card2[0]) == '1':
            score1 += card1[1] * 2
            score2 += card2[1]
        elif compare_colour(card1[0], card2[0]) == '2':
            score1 += card1[1]
            score2 += card2[1] * 2
        else:
            score1 += card1[1]
            score2 += card2[1]

    return score1, score2

score1, score2 = calculate_score(player1, player2)

print(f'Player 1 sorted hand: {player1}, score: {score1}')
print(f'Player 2 sorted hand: {player2}, score: {score2}')
```

```python
# Task 1.5

def declare_winner(score1, score2):
    if score1 == score2:
        print('Draw!')
    elif score1 > score2:
        print('Player 1 wins and Player 2 comes second!')
    else:
        print('Player 2 wins and Player 1 comes second!')

declare_winner(score1, score2)
```
