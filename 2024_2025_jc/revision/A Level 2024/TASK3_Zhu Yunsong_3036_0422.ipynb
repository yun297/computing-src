```python
# Task 3.1

def format_heading(text):
    return f"<h1>{text}</h1>"

def format_subheading(text):
    return f"<h2>{text}</h2>"

def format_paragraph(text):
    return f"<p>{text}</p>"

print(format_heading("My element"))
print(format_subheading("My element"))
print(format_paragraph("My element"))
```

```python
# Task 3.2

def format_hyperlink(text):
    link, content = text.split(',')
    return f'<a href="{link}">{content}</a>'

print(format_hyperlink("https://www.cambridgeinternational.org,Click here"))
```

```python
# Task 3.3

def body(list_of_elements):
    result = ""

    for element in list_of_elements:
        element_type, content = element.split(": ")

        if element_type == "heading":
            result += format_heading(content)
        elif element_type == "subheading":
            result += format_subheading(content)
        elif element_type == "paragraph":
            result += format_paragraph(content)
        elif element_type == "hyperlink":
            result += format_hyperlink(content)
        else:
            return False

    return "<html><body>" + result + "</body></html>"
```

```python
# Task 3.4

def read_data(filename):
    file = open(filename, 'r')

    list_of_elements = []

    for line in file:
        list_of_elements.append(line.strip())
    
    file.close()

    return body(list_of_elements)

print(read_data("firstPage.txt"))
```

```python
# Task 3.5

def write_data(filename, formatted_html):
    write_data = "<!doctype html>\n"
    write_data += formatted_html

    file = open(filename + ".html", 'w')
    file.write(write_data)
    file.close()

formatted_html_1 = read_data("firstPage.txt")
if formatted_html_1:
    write_data("test", formatted_html_1)
else:
    print("Unable to convert to HTML due to unrecognised elements!")

formatted_html_2 = read_data("website.txt")
if formatted_html_2:
    write_data("theWebsite", formatted_html_2)
else:
    print("Unable to convert to HTML due to unrecognised elements!")
```
