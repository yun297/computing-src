```python
# Task 2.1

class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1

    def set_data(self, data):
        self.data_value = data
        return

    def set_left(self, left):
        self.left_node = left
        return

    def set_right(self, right):
        self.right_node = right
        return

    def get_data(self):
        return self.data_value

    def get_left(self):
        return self.left_node

    def get_right(self):
        return self.right_node
```

```python
# Task 2.2

def calculate_nodes(levels):
    total_nodes = (2 ** levels) - 1
    print(f"A {levels} level number tree has {total_nodes} nodes")
    return total_nodes
```

```python
# Task 2.3

calculate_nodes(1)
calculate_nodes(3)
calculate_nodes(5)
calculate_nodes(10)
```

```python
# Task 2.4

def create_empty_tree(list_of_nodes, levels):
    for _ in range(calculate_nodes(levels)):
        list_of_nodes.append(Node())

    return list_of_nodes
```

```python
# Task 2.5

def assign_leaf_nodes(list_of_nodes, levels):
    total_leaf_nodes = 2 ** (levels - 1)

    for i in range(total_leaf_nodes):
        list_of_nodes[i].set_data(i + 1)

    return list_of_nodes
```

```python
# Task 2.6

def create_number_tree(list_of_nodes, levels):
    parent = 2 ** (levels - 1)
    
    for i in range(0, len(list_of_nodes) - 1, 2):
        # print(parent)
        list_of_nodes[parent].set_left(i)
        list_of_nodes[parent].set_right(i + 1)
        list_of_nodes[parent].set_data(list_of_nodes[i].get_data() + list_of_nodes[i + 1].get_data()) 

        parent += 1

    return list_of_nodes
```

```python
# Task 2.7

def tree(levels):
    list_of_nodes = create_empty_tree([], levels)
    list_of_nodes = assign_leaf_nodes(list_of_nodes, levels)
    list_of_nodes = create_number_tree(list_of_nodes, levels)

    print(f"{"left pointer":^12}{"data":^8}{"right pointer":^13}")
    for node in list_of_nodes:
        print(f"{node.get_left():^12}{node.get_data():^8}{node.get_right():^13}")

tree(1)
tree(3)
tree(4)
```
