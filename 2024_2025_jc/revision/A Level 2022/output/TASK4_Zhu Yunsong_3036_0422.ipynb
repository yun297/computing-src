```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("LIBRARY.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE Book(
    BookID INTEGER PRIMARY KEY,
    Title TEXT NOT NULL,
    Genre TEXT NOT NULL
)
""")

cur.execute("""
CREATE TABLE Member(
    MemberNumber INTEGER PRIMARY KEY,
    FamilyName TEXT NOT NULL,
    GivenName TEXT NOT NULL
)
""")

cur.execute("""
CREATE TABLE Loan(
    LoanID INTEGER PRIMARY KEY,
    MemberNumber INTEGER NOT NULL REFERENCES Member(MemberNumber),
    BookID INTEGER NOT NULL REFERENCES Book(BookID),
    DateLoaned TEXT NOT NULL,
    Returned BOOLEAN NOT NULL
)
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

book_file = open("BOOK.txt", 'r')
member_file = open("MEMBER.txt", 'r')
loan_file = open("LOAN.txt", 'r')

book_data = []
member_data = []
loan_data = []

for line in book_file:
    book_data.append(line.strip().split(','))

for line in member_file:
    member_data.append(line.strip().split(','))

for line in loan_file:
    loan_data.append(line.strip().split(','))

print(book_data)
print(member_data)
print(loan_data)

book_file.close()
member_file.close()
loan_file.close() 

conn = sqlite3.connect("LIBRARY.db")
cur = conn.cursor()

cur.executemany("""
INSERT INTO Book(BookID, Title, Genre) VALUES (?, ?, ?)
""", book_data)

cur.executemany("""
INSERT INTO Member(MemberNumber, FamilyName, GivenName) VALUES (?, ?, ?)
""", member_data)

cur.executemany("""
INSERT INTO Loan(LoanID, MemberNumber, BookID, DateLoaned, Returned) VALUES (?, ?, ?, ?, ?)
""", loan_data)

conn.commit()
conn.close()
```

```python
# Task 4.3

member_number = input("Enter member number: ")

conn = sqlite3.connect("LIBRARY.db")
cur = conn.cursor()

cur.execute("""
SELECT b.Title, l.Returned FROM Loan AS l
INNER JOIN Member AS m ON l.MemberNumber = m.MemberNumber
INNER JOIN Book AS b ON l.BookID = b.BookID
WHERE l.MemberNumber = ?
""", (member_number, ))

rows = cur.fetchall()

print(rows)
```
