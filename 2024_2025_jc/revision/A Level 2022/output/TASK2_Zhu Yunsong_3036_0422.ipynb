```python
# Task 2.2

task2_1()

def bubble_sort(arr_input):
    arr = arr_input.copy()

    n = len(arr)

    while n > 1:
        last_swap = 1

        for i in range(1, n):
            if arr[i - 1] > arr[i]:
                arr[i - 1], arr[i] = arr[i], arr[i - 1]
                last_swap = i

        n = last_swap

    return arr

print(bubble_sort(arr))
```

```python
# Task 2.3

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    l = r = 0
    result = []

    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            result.append(left[l])
            l += 1
        else:
            result.append(right[r])
            r += 1
            

    while l < len(left):
        result.append(left[l])
        l += 1

    while r < len(right):
        result.append(right[r])
        r += 1

    return result

algo_choice = input("What algorithm do you want to use? (bubble, merge): ")

while algo_choice not in ["bubble", "merge"]:
    algo_choice = input("What algorithm do you want to use? (bubble, merge): ")

task2_1()

if algo_choice == "bubble":
    sorted_arr = bubble_sort(arr)
else:
    sorted_arr = merge_sort(arr)

print(sorted_arr)
```

```python
# Task 2.4

def binary_search(arr, target, low, high):
    if low > high:
        return -1
        
    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search(arr, target, low, mid - 1)
    else:
        return binary_search(arr, target, mid + 1, high)

arr = merge_sort(arr)
print(arr)

target = int(input("Enter a integer value to search for: "))

output = binary_search(arr, target, 0, len(arr) - 1)

if output != -1:
    print(output)
else:
    print("Integer value not found!")

target = int(input("Enter a integer value to search for: "))

output = binary_search(arr, target, 0, len(arr) - 1)

if output != -1:
    print(output)
else:
    print("Integer value not found!")
```
