```python
# Task 3.1

from random import randint

class Tree:
    def __init__(self, data):
        self.data = data
        self.left_pointer = None
        self.right_pointer = None

    def insert(self, data):
        if data < self.data:
            if not self.left_pointer:
                self.left_pointer = Tree(data)
            else:
                self.left_pointer.insert(data)
        else:
            if not self.right_pointer:
                self.right_pointer = Tree(data)
            else:
                self.right_pointer.insert(data)
                
    def inorder(self):
        result = []
        
        if self.left_pointer:
            result += self.left_pointer.inorder()
            
        result.append(self.data)
        
        if self.right_pointer:
            result += self.right_pointer.inorder()

        return result

    def postorder(self):
        result = []
        
        if self.left_pointer:
            result += self.left_pointer.postorder()
        
        if self.right_pointer:
            result += self.right_pointer.postorder()

        result.append(self.data)

        return result
        
tree = Tree(randint(0, 999))

for i in range(9):
    tree.insert(randint(0, 999))

print(tree.inorder())
print(tree.postorder())
```
