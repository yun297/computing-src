```python
# Task 1.1

def encrypt_char(c):
    if c == ' ':
        return '!'
    elif c.isalpha():
        if c.isupper():
            # print(ord(c))
            ascii_val = ord(c) - 65 + 10
            ascii_val_result = ascii_val % 26 + 65
        else:
            # print(ord(c))
            ascii_val = ord(c) - 97 + 10
            ascii_val_result = ascii_val % 26 + 97

        return chr(ascii_val_result)
    else:
        return -1
```

```python
# Task 1.2

def task1_2():
    user_c = input("Enter a character: ")
    
    result = encrypt_char(user_c)
    
    if result == -1:
        print("Invalid character!")
    else:
        print(result)

task1_2()
task1_2()
task1_2()
task1_2()
```

```python
# Task 1.3

to_encrypt_file = open("DATATOENCRYPT.txt", 'r')

to_encrypt = to_encrypt_file.readline()

to_encrypt_file.close()

encrypted = ""

for c in to_encrypt:
    encrypted_c = encrypt_char(c)

    if encrypted_c != -1:
        encrypted += encrypted_c

encrypted_file = open("ENCRYPTEDMESSAGE.txt", 'w')

encrypted_file.write(encrypted)

encrypted_file.close()

encrypted_file = open("ENCRYPTEDMESSAGE.txt", 'r')

print(encrypted_file.readline())

encrypted_file.close()
```
