```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("LIBRARY.db")
cur = conn.cursor()

# Create Student table
cur.execute("""
CREATE TABLE IF NOT EXISTS Student(
    stu_id INTEGER PRIMARY KEY,
    name TEXT,
    cg TEXT,
    handphone TEXT
);
""")

# Create Latecoming table
cur.execute("""
CREATE TABLE IF NOT EXISTS Latecoming(
    date TEXT,
    stu_id INTEGER,
    reason TEXT,
    FOREIGN KEY (stu_id) REFERENCES Student(stu_id)
);
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

import csv
import sqlite3

student_f = open("STUDENT.csv", 'r')
student = []

for line in csv.reader(student_f):
    student.append(line)

student.pop(0)

student_f.close()

late_f = open("LATE.csv", 'r')
late = []

for line in csv.reader(late_f):
    late.append(line)

late.pop(0)

late_f.close()

conn = sqlite3.connect("LIBRARY.db")
cur = conn.cursor()

cur.executemany("""
INSERT INTO Student (stu_id, name, cg, handphone) VALUES (?, ?, ?, ?)
""", student)

cur.executemany("""
INSERT INTO Latecoming (date, stu_id, reason) VALUES (?, ?, ?)
""", late)

conn.commit()
conn.close()
```

```python
# Task 4.3

import sqlite3

def filter_latecomers(cg, min_latecomings, start_date, end_date):
    conn = sqlite3.connect("LIBRARY.db")
    cur = conn.cursor()

    cur.execute("""
SELECT name FROM Student AS s
INNER JOIN Latecoming As l
ON s.stu_id = l.stu_id
WHERE s.cg = ?
AND l.date BETWEEN ? AND ?
GROUP BY s.stu_id
HAVING COUNT (*) >= ?
ORDER BY s.name ASC
    """, (cg, start_date, end_date, min_latecomings))

    fetched = cur.fetchall()

    conn.close()

    result = []

    for item in fetched:
        result.append(item[0])

    return result

latecomers_august = filter_latecomers("25/13", 2, "2024-08-01", "2024-08-31")
print(latecomers_august)
```
