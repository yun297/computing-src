```python
# Task 3.1

f = open("PARTICIPANTS.txt", 'r')

participants = []

for line in f:
    row = line.split('\t')
    row_stripped = []

    for item in row:
        row_stripped.append(item.strip())

    row_stripped[2] = sum(int(ch) for ch in row_stripped[2])

    participants.append(row_stripped)

for row in participants[-5:]:
    print(row)

f.close()
```

```python
# Task 3.2

def merge_arr(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_arr(arr[:mid])
    right = merge_arr(arr[mid:])

    result = []
    l = r = 0

    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            result.append(left[l])
            l += 1
        else:
            result.append(right[r])
            r += 1

    while l < len(left):
        result.append(left[l])
        l += 1

    while r < len(right):
        result.append(right[r])
        r += 1

    return result

hash = []

for item in participants:
    hash.append(item[1])

hash_sorted = merge_arr(hash)

print(hash_sorted)
```

```python
# Task 3.4

list1 = []

for item in participants:
    if item[2] >= 8:
        list1.append(item)

records = {}

for item in list1:
    date = item[0][:10]
    nric_hash = item[1]
    score = item[2]

    key = (nric_hash, date)

    if key not in records or score > records[key][1]:
        records[key] = [item[0], score]

# print(records)

processed = []

for (nric_hash, date), (timestamp, score) in records.items():
    processed.append([timestamp, nric_hash, score])

# print(processed)

f = open("PROCESSED.txt", 'w')

write = ''

for item in processed:
    write += f"{item[0]}\t{item[1]}\t{str(item[2])}\n"

f.write(write)
f.close()
```
