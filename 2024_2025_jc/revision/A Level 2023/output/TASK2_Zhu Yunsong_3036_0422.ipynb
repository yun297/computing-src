```python
# Task 2.1

class QueueItem:
    def __init__(self, ID, Quantity):
        self.ID = ID
        self.Quantity = Quantity

    def getID(self):
        return self.ID

    def getQuantity(self):
        return self.Quantity
```

```python
# Task 2.2

class Queue:
    def __init__(self):
        self.queueList = [None, None, None, None]
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0

    def enqueue(self, item: QueueItem):
        if self.queueList[self.tailPointer] or self.numberItems == 4:
            return False # queue is full

        self.queueList[self.tailPointer] = item
        self.tailPointer = (self.tailPointer + 1) % 4
        self.numberItems += 1
        return True

    def dequeue(self):
        if not self.queueList[self.headPointer] or self.numberItems == 0:
            return False # queue is empty

        removed = self.queueList[self.headPointer]
        
        self.queueList[self.headPointer] = None
        self.headPointer = (self.headPointer + 1) % 4
        
        self.numberItems -= 1
        return removed

    def printQueue(self):
        if not self.queueList[self.headPointer] or self.numberItems == 0:
            print("Queue is empty!")
            return

        ptr = self.headPointer
        count = 0

        while self.queueList[ptr] and count < 4:
            print(f"ID: {self.queueList[ptr].getID()}, Quantity: {self.queueList[ptr].getQuantity()}")
            count += 1
            ptr = (ptr + 1) % 4
        
```

```python
# Task 2.3

q = Queue()

queue_item_list = []

for i in range(5):
    ID, Quantity = input("Enter id and quantity, comma seperated. (Eg. (3,33)): ").split(",")
    q_item = QueueItem(int(ID), int(Quantity))
    result = q.enqueue(q_item)

    if result:
        print(f"Item with id: {ID} and quantity: {Quantity} has been successfully enqueued!")
    else:
        print(f"Item with id: {ID} and quantity: {Quantity} failed to be enqueued!")
```

```python
# Task 2.4

for i in range(5):
    result = q.dequeue()

    if result:
        print(f"Item with id: {result.getID()} and quantity: {result.getQuantity()} has been successfully dequeued!")
    else:
        print("Empty queue!")
```
