```python
# Task 4.1

def read_file(filename):
    file = open(filename, 'r')

    data = []

    for line in file:
        data.append(line.strip())

    file.close()

    return data

def compress(data):
    result = ""

    last_colour = None
    count = 0
    
    for colour in data:
        if colour != last_colour:
            if last_colour is not None:
                result += last_colour + str(count)
            last_colour = colour
            count = 1
        else:
            count += 1

    result += last_colour + str(count)

    return result

def write_file(data):
    file = open("compressedimage.txt", 'w')

    file.write(data)

    file.close()

# write_file(compress(read_file("imagefile.txt")))
```

```python
# Task 4.2

def read_file_decomp(filename):
    file = open(filename, 'r')

    return file.readline().strip()

def decompress(data):
    bits = []
    result = []
    
    for i in range(0, len(data), 4):
        bits.append(data[i:i + 4])

    for colour in bits:
        for i in range(int(colour[3])):
            result.append(colour[:-1])

    return result

def write_file_decomp(data):
    file = open("decompressedimage.txt", 'w')

    write = ""

    for colour in data:
        write += colour + "\n"

    write.strip()

    file.write(write)

def menu():
    print("Hi user, this program will help compress and decompress image files")
    
    filename = input("Please enter filename: ")
    mode = input("Please enter mode of operation. (Enter '0' for compression, '1' for decompression): ")
    
    if mode == "0": # compression
        write_file(compress(read_file(filename)))
        print("Your compressed file has been written to `compressedimage.txt`")
        
    elif mode == "1": # decompression
        write_file_decomp(decompress(read_file_decomp(filename)))
        print("Your decompressed file has been written to `decompressedimage.txt`")
        
    else:
        print("Invalid mode!")
        
menu()
```
