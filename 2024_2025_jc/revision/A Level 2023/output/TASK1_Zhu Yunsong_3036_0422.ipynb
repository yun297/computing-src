```python
# Task 1.1

import sqlite3

conn = sqlite3.connect("TESTDATABASE")
cur = conn.cursor()

# clear database
conn.execute("DROP TABLE IF EXISTS Student")
conn.execute("DROP TABLE IF EXISTS Test")
conn.execute("DROP TABLE IF EXISTS Class")
conn.execute("DROP TABLE IF EXISTS Student_Test")
conn.commit()

cur.execute("""
CREATE TABLE Class(
    ClassName TEXT PRIMARY KEY,
    YearGroup INTEGER NOT NULL
)
""")

cur.execute("""
CREATE TABLE Student(
    StudentID INTEGER PRIMARY KEY,
    GivenName TEXT NOT NULL,
    FamilyName TEXT NOT NULL,
    ClassName TEXT NOT NULL REFERENCES Class(ClassName)
)
""")

cur.execute("""
CREATE TABLE Test(
    TestID TEXT PRIMARY KEY,
    MaxMarks INTEGER NOT NULL
)
""")

cur.execute("""
CREATE TABLE Student_Test(
    StudentTestID INTEGER PRIMARY KEY,
    StudentID INTEGER NOT NULL REFERENCES Student(StudentID),
    Mark INTEGER NOT NULL,
    DateTest TEXT NOT NULL
)
""")

conn.commit()
conn.close()
```
