```python
# Task 3.1

# Conversion factors in bytes
SI_UNITS = {
    "KB": 10**3,
    "MB": 10**6,
    "GB": 10**9,
    "TB": 10**12
}

BINARY_UNITS = {
    "KiB": 2**10,
    "MiB": 2**20,
    "GiB": 2**30,
    "TiB": 2**40
}

# Combine all units
ALL_UNITS = {**SI_UNITS, **BINARY_UNITS}

def task3_1(quantity_of_data):
    unit = quantity_of_data[-2:]
    if unit in SI_UNITS and quantity_of_data[:-2].isdigit():
        bytes_value = int(quantity_of_data[:-2]) * SI_UNITS[unit]
        print(bytes_value)
        return bytes_value
    print("invalid data")
    return "invalid data"

task3_1("8KB")
task3_1("8MB")
task3_1("8GB")
task3_1("8TB")
task3_1("8G")
```

```python
# Task 3.2

def task3_2(quantity_of_data):
    # Find if last 2 or 3 chars are a valid unit
    for length in (2, 3):
        unit = quantity_of_data[-length:]
        if unit in ALL_UNITS and quantity_of_data[:-length].isdigit():
            bytes_value = int(quantity_of_data[:-length]) * ALL_UNITS[unit]
            print(bytes_value)
            return bytes_value
    print("invalid data")
    return "invalid data"

task3_2("8KB")
task3_2("8MB")
task3_2("8GB")
task3_2("8TB")
task3_2("8KiB")
task3_2("8MiB")
task3_2("8GiB")
task3_2("8TiB")
task3_2("8G")
```

```python
# Task 3.3

def task3_3(quantity_of_data, target_unit):
    if target_unit not in ALL_UNITS:
        print("invalid data")
        return "invalid data"

    bytes_value = task3_2(quantity_of_data)
    if bytes_value == "invalid data":
        return "invalid data"

    converted_value = bytes_value / ALL_UNITS[target_unit]
    print(converted_value)
    return converted_value
        
task3_3("512MiB", "GiB")
```
