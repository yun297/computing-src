```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("school.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE People (
	PersonID INTEGER PRIMARY KEY AUTOINCREMENT,
	FullName TEXT,
	DateOfBirth TEXT,
	ScreenName TEXT,
	IsAdult INTEGER
);
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

class Person:
    def __init__(self, full_name: str, date_of_birth: str):
        self.full_name = full_name
        self.date_of_birth = date_of_birth

    def is_adult(self):
        return (2020 - int(self.date_of_birth[:4])) > 18

    def screen_name(self):
        screen = ""
        for c in self.full_name:
            if c.isalpha():
                screen += c

        screen += self.date_of_birth[5:7]
        screen += self.date_of_birth[8:10]
        
        return screen

class Staff(Person):
    def __init__(self, full_name: str, date_of_birth: str):
        super().__init__(full_name, date_of_birth)

    def screen_name(self):
        screen = ""
        for c in self.full_name:
            if c.isalpha():
                screen += c

        screen += self.date_of_birth[5:7]
        screen += self.date_of_birth[8:10]
        screen += "Staff"
        
        return screen

    def is_adult(self):
        return True

class Student(Person):
    def __init__(self, full_name: str, date_of_birth: str):
        super().__init__(full_name, date_of_birth)
        
    def is_adult(self):
        return False
```

```python
f = open("people.txt", 'r')

conn = sqlite3.connect("school.db")
cur = conn.cursor()

for line in f:
    data = line.strip().split(',')
    name = data[0]
    dob = data[1]
    person_type = data[2]
    
    if person_type == "Person":
        person = Person(name, dob)
    elif person_type == "Staff":
        person = Staff(name, dob)
    elif person_type == "Student":
        person = Student(name, dob)

    cur.execute("""
INSERT INTO People (FullName, DateOfBirth, ScreenName, IsAdult) VALUES (?, ?, ?, ?)
    """, (name, dob, person.screen_name(), person.is_adult()))

    conn.commit()

conn.close()
```
