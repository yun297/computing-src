```python
# Task 2.1 

def read_from_file(filename):
    file = open(filename, "r")

    list_of_int = []

    for line in file:
        list_of_int.append(int(line))

    file.close()

    return list_of_int

list1 = read_from_file("TASK2_DATA.txt")
print(list1)
print(len(list1)) 
```

```python
# Task 2.2

def quick_sort(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers
        
    pivot = list_of_integers[0]
    left = []
    right = []

    for integer in list_of_integers[1:]:
        if integer <= pivot:
            left.append(integer)
        else:
            right.append(integer)

    left = quick_sort(left)
    right = quick_sort(right)

    return left + [pivot] + right

list1 = read_from_file("TASK2_DATA.txt")
list1_sorted = quick_sort(list1)
print(list1_sorted) 
```

```python
# Task 2.3

def linear_search(list_of_integers, target):
    for integer in list_of_integers:
        if integer == target:
            return True

    return False

list1 = read_from_file("TASK2_DATA.txt")
list1_sorted = quick_sort(list1)
print(linear_search(list1_sorted, 2022)) 
```

```python
# Task 2.4

def binary_search(sorted_list, target):
    def binary(arr, target, high = None, low = 0):
        if high is None:
            high = len(sorted_list) - 1

        if low > high:
            return False

        mid = (high + low) // 2

        if arr[mid] == target:
            return True
        elif arr[mid] > target:
            return binary(arr, target, mid - 1, low)
        elif arr[mid] < target:
            return binary(arr, target, high, mid + 1)

    return binary(sorted_list, target)

list1 = read_from_file("TASK2_DATA.txt")
list1_sorted = quick_sort(list1)
print(binary_search(list1_sorted, 2022)) 
```

```python
# Task 2.4 (iterative)

def binary_search(sorted_list, target):
    low = 0
    high = len(sorted_list) - 1

    while low <= high:
        mid = (low + high) // 2
        
        if sorted_list[mid] == target:
            return True
        elif sorted_list[mid] > target:
            high = mid - 1
        elif sorted_list[mid] < target:
            low = mid + 1

    return False

list1 = read_from_file("TASK2_DATA.txt")
list1_sorted = quick_sort(list1)
print(binary_search(list1_sorted, 2022)) 
```
