```python
# Task 1.1

def veh_num_check(string):
    veh_num_list = string.split()
    
    # Must have exactly 3 parts
    if len(veh_num_list) != 3:
        return 2  # Fails general format, treat as invalid front section

    front, middle, end = veh_num_list
    
    # Rule 1: First character must be 'S' or 'E'
    if front[0] not in ['S', 'E']:
        return 1

    # Rule 2: Front should be 1 to 3 uppercase letters
    if len(front) < 1 or len(front) > 3 or not front.isalpha():
        return 2

    # Rule 3: Middle between 1 to 9999
    if not middle.isdigit() or not (1 <= int(middle) <= 9999):
        return 3

    # Rule 4: Last part must be a single uppercase letter
    if len(end) != 1 or not end.isalpha():
        return 4

    return 0
```

```python
# Task 1.2

# Test case 0
veh_num_0 = "SKZ 1234 Z"
print(veh_num_check(veh_num_0))  # Expected output: 0
# Satisfies all rules

# Test case 1
veh_num_1 = "XYZ 1234 Z"
print(veh_num_check(veh_num_1))  # Expected output: 1
# Fails to satisfy rule 1, as vehicle number does not begin with 'S' or 'E'

# Test case 2
veh_num_2 = "SBCD 1234 Z"
print(veh_num_check(veh_num_2))  # Expected output: 2
# Fails to satisfy rule 2, as front characters are not within length of 1 to 3

# Test case 3
veh_num_3 = "SKZ 10000 Z"
print(veh_num_check(veh_num_3))  # Expected output: 3
# Fails to satisfy rule 3, as middle number is not between 1 and 9999 (inclusive)

# Test case 4
veh_num_4 = "SKZ 1234 9"
print(veh_num_check(veh_num_4))  # Expected output: 4
# Fails to satisfy rule 4, as last character is not an alphabet
```
