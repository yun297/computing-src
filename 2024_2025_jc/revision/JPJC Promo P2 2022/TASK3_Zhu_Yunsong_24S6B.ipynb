```python
# Task 3.1

class COEBid:
    def __init__(self, id: str, price: int):
            self.cust_id = id
            self.reserve_price = price
            self.status = "Bidding in progress"

    def get_cust_id(self):
        return self.cust_id

    def get_reserve_price(self):
        return self.reserve_price

    def get_status(self):
        return self.status

    def set_status(self, s: str):
        self.status = s

    def display(self):
        print(f"{self.cust_id}, {self.reserve_price}, {self.status}")
```

```python
# Task 3.2

class Node:
    def __init__(self, data: COEBid):
        self.data = data
        self.next = None

class COELinkedList:
    def __init__(self):
        self.head = None

    def insert_bid(self, data: COEBid):
        new_node = Node(data)
        
        if self.head is None or data.get_reserve_price() >= self.head.data.get_reserve_price():
            new_node.next = self.head
            self.head = new_node

        else:
            ptr = self.head

            while data.get_reserve_price() >= ptr.data.get_reserve_price() and ptr.next is not None:
                ptr = ptr.next

            new_node.next = ptr.next
            ptr.next = new_node

    def display_list(self):
        if self.head is None:
            print("Empty")
        else:
            ptr = self.head
            print(f"{ptr.data.get_cust_id():<5}{ptr.data.get_reserve_price():<4}{ptr.data.get_status():<20}")    
            
            while ptr.next is not None:
                ptr = ptr.next
                print(f"{ptr.data.get_cust_id():<5}{ptr.data.get_reserve_price():<4}{ptr.data.get_status():<20}")

    def calculate_COEPrice(self, quota: int):
        count = 0

        ptr = self.head

        while ptr is not None and count < quota:
            count += 1
            ptr = ptr.next

        return ptr.data.get_reserve_price() + 1

    def update_status(self, quota: int):
        ptr = self.head
        
        while ptr is not None:
            if ptr.data.get_reserve_price() > self.calculate_COEPrice(quota):
                ptr.data.set_status("Sucessful")
            else:
                ptr.data.set_status("Unsucessful")

            ptr = ptr.next
```

```python
COE_LL = COELinkedList()

BIDs = [
    ["T001", 41],
    ["T002", 88],
    ["T003", 67],
    ["T004", 70],
    ["T005", 100]
]

for BID in BIDs:
    COE_LL.insert_bid(COEBid(BID[0], BID[1]))

COE_LL.display_list()
```

```python
# Task 3.3

quota = 3
COE_price = COE_LL.calculate_COEPrice(quota)
print("The COE price is", COE_price)
COE_LL.update_status(quota)
COE_LL.display_list()
```
