```python
import socket

client = socket.socket()
client.connect(("127.0.0.1", 10000))

while True:
    article_no = input("Enter a article no. or 'exit' to quit")

    if article_no == "exit":
        client.send(b"bye\n")
        break
    else:
        client.send(article_no.encode())
    
    message = client.recv(1024).decode().strip()
    print(message)
```

```python
client.close()
```
