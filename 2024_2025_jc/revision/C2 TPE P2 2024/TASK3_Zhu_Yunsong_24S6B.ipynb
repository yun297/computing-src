```python
class TreeNode:
    def __init__(self, name: str, DOB: str):
        self.name = name
        self.DOB = DOB  # format: YYYYMMDD as a string
        self.left = None
        self.right = None

class Tree:
    def __init__(self):
        self.root = None

    def insert(self, name, DOB):
        new = TreeNode(name, DOB)
        if self.root is None:
            self.root = new
            return

        ptr = self.root
        while True:
            if DOB < ptr.DOB:
                if ptr.left:
                    ptr = ptr.left
                else:
                    ptr.left = new
                    break
            else:
                if ptr.right:
                    ptr = ptr.right
                else:
                    ptr.right = new
                    break

    def in_order(self):
        def in_order_helper(node):
            if node:
                in_order_helper(node.left)
                print(f"Name: {node.name}, DOB: {node.DOB}")
                in_order_helper(node.right)
        in_order_helper(self.root)

    def count_month(self, month):
        def count_helper(node):
            if not node:
                return 0
            count = count_helper(node.left)
            if node.DOB[4:6] == f"{int(month):02d}":
                count += 1
            count += count_helper(node.right)
            return count
        return count_helper(self.root)
```

```python
tree = Tree()

# Read from CLASS.txt (assumes format: Name,DOB in YYYYMMDD)
with open("CLASS.txt", "r") as file: # open file and automatically close it after with block ends
    for line in file:
        parts = line.strip().split(",")
        if len(parts) == 2:
            name, dob = parts[0].strip(), parts[1].strip()
            tree.insert(name, dob)

# In-order traversal output
print("In-Order Traversal:")
tree.in_order()

# Monthly count output
print("\nMonth  Total")
for month in range(1, 13):
    total = tree.count_month(month)
    print(f"{month:02d}     {total}")
```
