```python
import socket

listener = socket.socket()
listener.bind(("127.0.0.1", 10000))
listener.listen()
new, addr = listener.accept()

print("Connected on port", addr)

ITEMS_txt = open("ITEMS.txt", "r")
item_arr = []
items = {}

for line in ITEMS_txt:
    item_arr.append(line)

for entry in item_arr[1:]:
    item = entry.split(",")
    article_no = item[0].strip()
    item_name = item[1].strip()

    items[article_no] = item_name
    
ITEMS_txt.close()

while True:
    message = new.recv(1024).decode().strip()
    
    if message == "bye":
        break

    if message not in items.keys():
        new.send(b"Not Found\n")
    else:
        data = items[message] + "\n"
        new.send(data.encode())
```

```python
listener.close()
new.close()
```
