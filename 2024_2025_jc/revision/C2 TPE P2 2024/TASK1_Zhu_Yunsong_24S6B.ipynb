```python
# Task 1.1

from pymongo import MongoClient

Client = MongoClient("localhost", 27017)
Entertainment = client["Entertainment"]
Program = Entertainment["Program"]
```

```python
import json

PROGRAM_json = open("PROGRAM.json", "r")

program_arr = []

for line in json.load(PROGRAM_json):
    program_arr.append(line)

# print(program_arr)

results = Program.insert_many(program_arr)
```

```python
query = {"_id": {"$in":results.inserted_ids}}
proj = {"_id": 0}

for item in Program.find({}, proj):
    print(item)
```

```python
# Task 1.2

query = {"Top Cast": {"$exists": False}}
proj = {"_id": 0}

header = f"{'Title':<35} {'Type':<10} {'Year':<5} {'Genre':<35} {'Season':<6} {'Episodes':<8} {'Rating':<6} {'Top Cast'}"
print(header)
print('-' * len(header))

for item in Program.find(query, proj):
    title = item['Title']
    type_ = item['Type']
    year = item['Year']
    
    genre = item['Genre']
    genre_str = ', '.join(genre) if isinstance(genre, list) else genre
    
    season = item.get('Season', '')
    episodes = item.get('Episodes', '')
    rating = item['Rating']
    top_cast = item.get('Top Cast', '')
    
    print(f"{title:<35} {type_:<10} {year:<5} {genre_str:<35} {season!s:<6} {episodes!s:<8} {rating:<6} {top_cast}")
```

```python
query = {"Top Cast": "Taylor Swift"}

print("Number of programs with top cast Taylor Swift:", Program.count_documents(query))
```

```python
query = {"Title": "Black Mirror"}

total_ep = 0

for ep in Program.find(query):
    total_ep += ep["Episodes"]

print("Total number of episodes in Black Mirror:", total_ep)
```
