```python
# Task 4.1

class Book:
    def __init__(self, title: str, location: str):
        self.title = title
        self.location = location
```

```python
# Task 4.2

def task4_2(array, left, right):
    def quick(arr):   
        if len(arr) <= 1:
            return arr
    
        pivot = arr[0]
        left_part = []
        right_part = []

        for item in arr[1:]:
            if item.title <= pivot.title:
                left_part.append(item)
            else:
                right_part.append(item)

        return quick(left_part) + [pivot] + quick(right_part)

    sorted_slice = quick(array[left:right])
    
    for i in range(len(sorted_slice)):
        array[left + i] = sorted_slice[i]
```

```python
# Task 4.3

def task4_3(array, left, right, title):
    while left <= right:
        mid = (left + right)//2
        mid_title = array[mid].title

        if mid_title == title:
            return array[mid]
        elif mid_title < title:
            left = mid + 1
        else:
            right = mid - 1

    return
```

```python
# Task 4.4

def task4_4():
    bookArr = [None] * 50
    
    book_file = open("BOOKS.txt", "r")
    
    count = 0
    
    for line in book_file:
        book = line.split(",")
        title = book[0].strip()
        location = book[1].strip()
    
        bookArr[count] = Book(title, location)
        count += 1
    
    book_file.close()
    
    # Sort only the filled part of the array
    task4_2(bookArr, 0, count)
    
    search_title = input("Enter the title to search for: ").strip()
    
    found_book = task4_3(bookArr, 0, count - 1, search_title)
    
    if found_book:
        print(f"Book found: {found_book.title} at {found_book.location}")
    else:
        print("Not Found")
```

```python
task4_4()
```

```python
# Task 4.5

empty_i = -1
for i in range(len(bookArr)):
    if bookArr[i] is None:
        empty_i = i
        break # exit the loop once the first None item is found in bookArr

def add_book(index):
    book = input("Enter details of new book: ")
    book = book.split(",")
    title = book[0].strip()
    location = book[1].strip()
    bookArr[index] = Book(title, location)

# adding 2 books
add_book(empty_i)
add_book(empty_i + 1)
```
