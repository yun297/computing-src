```python
# Task 2.1

class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def print(self):
        print(self.name, self.age)
```

```python
# Task 2.2

def task2_2(filename):
    file = open(filename, "r")
    arr = []

    for line in file:
        person = line.split(",")
        name = person[0].strip()
        age = person[1].strip()

        arr.append(Person(name, age))

    return arr

list_of_person = task2_2('PERSON.txt')
for person in list_of_person:
    person.print()
```

```python
# Task 2.3

def task2_3(list_of_person, key, order):
    def get_value(person: Person):
        if key == "name":
            return person.name
        elif key == "age":
            return person.age
        
    for i in range(1, len(list_of_person)): # start at 1 to skip the first 
        compare = list_of_person[i]
        space = i
        
        if order == "asc":
            while space > 0 and get_value(list_of_person[space - 1]) > get_value(compare):
                list_of_person[space] = list_of_person[space - 1]
                space -= 1
        elif order == "desc":
            while space > 0 and get_value(list_of_person[space - 1]) < get_value(compare):
                list_of_person[space] = list_of_person[space - 1]
                space -= 1

        list_of_person[space] = compare

list_of_person = task2_2('PERSON.txt' )
task2_3(list_of_person, 'name', 'asc')
for person in list_of_person:
    person.print()        
```

```python
# Task 2.4

def task2_4(list_of_person, key, order):
    def get_value(person: Person):
        if key == "name":
            return person.name
        elif key == "age":
            return person.age
    
    def quick(arr):
        if len(arr) <= 0:
            return arr
        
        pivot = arr[0]
        left = []
        right = []
        
        for item in arr[1:]:
            if (order == "asc" and get_value(item) <= get_value(pivot)) or (order == "desc" and get_value(item) > get_value(pivot)):
                left.append(item)
            else:
                right.append(item)
                    
        left = quick(left)
        right = quick(right)

        return left + [pivot] + right

    list_of_person[:] = quick(list_of_person)

list_of_person = task2_2('PERSON.txt')
task2_4(list_of_person, 'age', 'desc')
for person in list_of_person:
    person.print()
```

```python
# Task 2.5

def task2_5(list_of_person, method, key, order):
    if method == "insertion sort":
        task2_3(list_of_person, key, order)
    elif method == "quick sort":
        task2_4(list_of_person, key, order)

list_of_person = task2_2('PERSON.txt')
task2_5(list_of_person, 'quick sort', 'name', 'desc')
for person in list_of_person:
    person.print()
```
