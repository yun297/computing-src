```python
# Task 3.1

class Player:
    def __init__(self, name: str, elo: int, ptr: int = -1):
        self.name = name
        self.elo = elo
        self.ptr = ptr

class PlayerList: 
    def __init__(self, n: int):
        self.head = -1
        self.free = 0
        self.data = []
        for i in range(n):
            self.data.append(Player("-", -1, i + 1))

        self.data[-1].ptr = -1

    def size(self):
        size = 0
        curr = self.head

        while curr != -1:
            size += 1
            curr = self.data[curr].ptr

        return size

    def register(self, name: str, elo: int):
        if self.free == -1:
            print("Error: Player list is full")
            return

        new_index = self.free
        self.free = self.data[self.free].ptr

        self.data[new_index].name = name
        self.data[new_index].elo = elo

        # Insert at the start if list is empty or new node has highest elo
        if self.head == -1 or elo > self.data[self.head].elo:
            self.data[new_index].ptr = self.head
            self.head = new_index
            return

        # Otherwise, find the correct position to insert
        prev = self.head
        curr = self.data[prev].ptr

        while curr != -1 and elo <= self.data[curr].elo:
            prev = curr
            curr = self.data[curr].ptr

        self.data[new_index].ptr = curr
        self.data[prev].ptr = new_index

    def withdraw(self, name: str):
        if self.head == -1:
            print("Player list is empty!")
            return

        prev = -1
        curr = self.head

        while curr != -1:
            if self.data[curr].name == name:
                if curr == self.head:
                    self.head = self.data[curr].ptr
                else:
                    self.data[prev].ptr = self.data[curr].ptr

                self.data[curr].name = "-"
                self.data[curr].elo = -1
                self.data[curr].ptr = self.free
                self.free = curr
                return
                
            prev = curr
            curr = self.data[curr].ptr

        print("Player not found!")

    def display(self):
        print(f"Head Pointer: {self.head}")
        print(f"Free Pointer: {self.free}")
        print("Index | Name       | ELO   | Ptr")
        print("----------------------------------")
        
        for i in range(len(self.data)):
            name = self.data[i].name
            elo = self.data[i].elo
            ptr = self.data[i].ptr
            print(f"{i:5} | {name:10} | {elo:5} | {ptr}")
```

```python
# Task 3.2

from csv import reader

chess_file = open("CHESS.csv", "r")

chess_csv = reader(chess_file)

chess_array = []

for line in chess_csv:
    chess_array.append(line)

chess_file.close()

cteam = PlayerList(7)

for player in chess_array:
    cteam.register(player[0], int(player[1]))

cteam.display()

cteam.withdraw("Taylor")

print(cteam.size())

cteam.display()
```
