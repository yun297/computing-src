```python
# Task 1.1

import random

maze_txt = open("maze.txt", "r")
maze_arr = []
for line in maze_txt:
    row = []
    for col in line.strip():
        row.append(col)
    maze_arr.append(row)

exit = [random.randint(1, 4), 5]
print(exit)

maze_arr[exit[1]][exit[0]] = "."

def display(maze):
    for row in maze:
        print("".join(row))
    print()

display(maze_arr)
```

```python
# Task 1.2

display(maze_arr)
pos = [4, 0]
next_pos = pos
moves = 0

exited = False
while not exited:
    valid_dir = False
    
    while not valid_dir:
        next_pos = pos.copy()
        
        direction = input("Enter direction ('U','D','L','R',''): ")
        
        if direction == 'U':
            next_pos[1] -= 1
        elif direction == 'D':
            next_pos[1] += 1
        elif direction == 'L':
            next_pos[0] -= 1
        elif direction == 'R':
            next_pos[0] += 1

        if 0 <= next_pos[0] < len(maze_arr) and 0 <= next_pos[1] < len(maze_arr[0]):
            if next_pos == exit:
                exited = True
                valid_dir = True
            elif maze_arr[next_pos[1]][next_pos[0]] == '.':
                valid_dir = True
            else:
                print("Can't go there!")

        else:
            print("Can't go there!")  

    moves += 1
    
    maze_arr[pos[1]][pos[0]] = 'X'
    maze_arr[next_pos[1]][next_pos[0]] = 'T'
    
    display(maze_arr)
    
    pos = next_pos.copy()

print(f"The robot takes {moves} moves to exit the maze.")
```
