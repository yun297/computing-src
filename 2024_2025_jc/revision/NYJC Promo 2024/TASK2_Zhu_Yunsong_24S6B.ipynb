```python
# Task 2.1

from random import randint

def task2_1(number_limit):
    result = []

    for i in range(80):
        result.append(randint(1, number_limit))

    return result
```

```python
# Task 2.2

def task2_2(unsorted_list):
    arr = unsorted_list.copy()

    for i in range(1, len(arr)):
        compare = arr[i]
        space = i

        while space > 0 and arr[space - 1] > compare:
            arr[space] = arr[space - 1]
            space -= 1

        arr[space] = compare

    return arr
```

```python
print(task2_2(task2_1(100)))
```

```python
# Task 2.3

def task2_3(sorted_list, value):
    low = 0
    high = len(sorted_list)

    while low <= high:
        mid = (low + high) // 2

        if sorted_list[mid] == value:
            return mid
        elif sorted_list[mid] >= value:
            high = mid - 1
        else:
            low = mid + 1
            
    return -1
```

```python
arr = task2_2(task2_1(100))
print(arr)
```

```python
for i in range(2):
    value = int(input("Enter a number to search for: "))

    output = task2_3(arr, value)
    
    if output != -1:
        print(output)
    else:
        print("Not found")
```
