```python
    # Task 3.1
    
    class Transaction:
        def __init__(self, amount: int, recipient: str):
            self.amount = amount
            self.recipient = recipient
    
    class ChequeDeposit(Transaction):
        def __init__(self, amount: int, recipient: str, chequeNumber: int):
            super().__init__(amount, recipient)
            self.chequeNumber = chequeNumber
    
        def toString(self):
            return f"Deposit cheque {self.chequeNumber} ${self.amount/100:.2f} into {self.recipient}"
    
    class BankTransfer(Transaction):
        def __init__(self, amount: int, recipient: str, sender: str):
            super().__init__(amount, recipient)
            self.sender = sender
    
        def toString(self):
            return f"Transfer ${self.amount/100:.2f} from {self.sender} to {self.recipient}"
```

```python
# Task 3.2

class Queue:    
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.data = [None] * capacity
        self.head = 0
        self.tail = 0

    def enqueue(self, transaction: Transaction):
        if self.data[self.tail]:
            return False  # Queue is full

        self.data[self.tail] = transaction
        self.tail = (self.tail + 1) % self.capacity
        return True

    def dequeue(self):
        if not self.data[self.head]:
            return None  # Queue is empty

        item = self.data[self.head]
        self.data[self.head] = None
        self.head = (self.head + 1) % self.capacity
        return item
```

```python
# Task 3.3

q = Queue(5)

transactions_file = open("transactions.txt", 'r')
transactions_arr = []

for line in transactions_file:
    transactions_arr.append(line.strip().split(','))

transactions_file.close()

transactions = []

for t in transactions_arr:
    if t[0][0] == 'T':
        transactions.append(BankTransfer(int(t[1]), t[2], t[3]))
    else:
        transactions.append(ChequeDeposit(int(t[1]), t[2], int(t[3])))

not_full = True
not_empty = True
for t in transactions:
    if not q.enqueue(t): # Queue is full, so flush it
        removed = q.dequeue()
        while removed:
            print(removed.toString())
            removed = q.dequeue()

        # Try enqueue again after clearing
        q.enqueue(t)

# Final flush of remaining items

removed = q.dequeue()
while removed:
    print(removed.toString())
    removed = q.dequeue()

        
    
```
