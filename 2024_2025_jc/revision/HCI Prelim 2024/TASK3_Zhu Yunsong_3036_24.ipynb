```python
# Task 3.1

class TaskNode:
    def __init__(self, taskName: str, priority: int):
        self.taskName = taskName
        self.priority = priority
        self.next = None

    def __str__(self):
        return f"{self.taskName},{self.priority}"
```

```python
# Task 3.2

class PriorityQueue:
    def __init__(self):
        self.head = None

    def viewPendingTasks(self):
        if self.head:
            ptr = self.head

            while ptr.next:
                print(ptr)
                ptr = ptr.next

            print(ptr)

    def enqueue(self, taskNode):
        if not self.head:
            self.head = taskNode
            return

        ptr = self.head

        while ptr.next and taskNode.priority < ptr.next.priority:
            ptr = ptr.next

        taskNode.next = ptr.next
        ptr.next = taskNode

    def dequeue(self):
        popped_task = self.head

        self.head = popped_task.next
        
        return popped_task

    def isEmpty(self):
        if self.head:
            return False
        return True

    def delete(self, taskName):
        if not self.head:
            return False

        if self.head.taskName == taskName:
            self.head = None
            return True
        
        ptr = self.head

        while ptr.next and ptr.next.taskName != taskName:
            ptr = ptr.next

        if not ptr.next:
            return False

        ptr.next = ptr.next.next
        return True

    def search(self, taskName):
        if not self.head:
            return

        ptr = self.head

        while ptr and ptr.taskName != taskName:
            ptr = ptr.next

        if ptr:
            return ptr
        return

    def updatePriority(self, taskName, newPriority):
        task = self.search(taskName)
        
        if not task:
            return
        
        self.delete(task.taskName)

        updated_task = TaskNode(taskName, newPriority)

        self.enqueue(updated_task)
```

```python
file = open("TASKS.txt", 'r')
queue = PriorityQueue()

for line in file:
    task = line.strip().split(',')
    queue.enqueue(TaskNode(task[0], int(task[1])))

queue.viewPendingTasks()

queue.dequeue()
queue.dequeue()

queue.viewPendingTasks()

file.close()
```
