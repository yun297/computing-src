```python
# Task 1.1

def task1_1(key, message):
    key_len = len(key)
    msg_len = len(message)
    arr = []

    for char in message:
        if char == ' ':
            arr.append('z')
        else:
            arr.append(char)

    x_no = key_len - msg_len % key_len
    for i in range(x_no):
        arr.append('x')

    grid = []
    row_no = int(len(arr) / key_len)

    for i in range(row_no):
        grid.append(arr[key_len * i:key_len * (i + 1)])

    key_arr = []
    for char in key:
        key_arr.append(char)
    key_arr_sorted = sorted(key_arr)

    order = []
    for char in key_arr_sorted:
        order.append(key_arr.index(char))

    cipher = ""
    
    for i in order:
        for row in grid:
            cipher += row[i]

    return cipher
```

```python
# Task 1.2

print(task1_1("ALGORITHM", "computing is an interesting subject"))
```
