```python
# Task 2.1

import csv

def read_csv(filename):
    file = open(filename, 'r')
    csv_file = csv.reader(file)
    csv_array = []

    for line in csv_file:
        csv_array.append(line)
    file.close()

    return csv_array

books_array = read_csv('booklist.csv')
print(len(books_array))
print(books_array)
```

```python
# Task 2.2

def bubble(array, sort_key):
    sort_keys = ['title', 'author', 'year']
    if sort_key in sort_keys:
        sort_i = sort_keys.index(sort_key)
    else:
        return -1

    for i in range(len(array) - 1):
        for j in range(len(array) - i - 1):
            if array[j][sort_i] > array[j + 1][sort_i]:
                array[j], array[j + 1] = array[j + 1], array[j]

    return array
                
print(bubble(books_array, "title"))
print(bubble(books_array, "ISBN")) 
```

```python
# Task 2.3

def merge(array, sort_key):
    sort_keys = ['title', 'author', 'year']
    if sort_key in sort_keys:
        sort_i = sort_keys.index(sort_key)
    else:
        return -1
    
    def merge_helper(arr):
        if len(arr) <= 1:
            return arr

        length = len(arr)

        mid = length // 2
        left = arr[:mid]
        right = arr[mid:]

        merge_helper(left)
        merge_helper(right)

        l = r = i = 0

        while l < len(left) and r < len(right):
            if left[l][sort_i] < right[r][sort_i]:
                arr[i] = left[l]
                l += 1
            else:
                arr[i] = right[r]
                r += 1
            i += 1

        while l < len(left):
            arr[i] = left[l]
            l += 1
            i += 1

        while r < len(right):
            arr[i] = right[r]
            r += 1
            i += 1

        return arr

    return merge_helper(array)

print(merge(books_array, "author"))
print(merge(books_array, "year"))
```

```python
# Task 2.4

def reverse(array):
    return array[::-1]

print(reverse([1,3,5,2,4]))
print(reverse([1,9,6,4]))
```
