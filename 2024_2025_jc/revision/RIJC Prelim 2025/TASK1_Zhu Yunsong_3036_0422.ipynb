```python
# Task 1.1

def validate_student_code(code_list):
    valid_codes = []
    
    for code in code_list:
        if code[:2] not in ["BW", "BB", "HH", "MR", "MT"]:
            print(f"{code} is invalid: code does not begin with house code!")
        elif not code[2:].isnumeric() or len(code) <= 2:
            print(f"{code} is invalid: code must end with a number!")
        elif not (1 <= int(code[2:]) <= 99):
            print(f"{code} is invalid: code must end with a number from 1 to 99!")
        else:
            if code not in valid_codes:
                valid_codes.append(code)

    return valid_codes

code_list = ['BW12', 'MR5', 'HH88', 'MT3', 'BB8', 'HH3', 'MR99', 'BW100', 'ZZ12', 'MR-5', 'mr22', 'MT0', 'HH', 'B100', 'M09', 'HH3A', 'MT01', 'BW1', 'MR99']

print(validate_student_code(code_list))
```

```python
# Task 1.2

def extract_house_and_number(code):
    return code[:2], int(code[2:])

def insertion_sort_by_number(code_list):
    arr = code_list.copy()

    for i in range(1, len(arr)):
        space = i
        temp = arr[space]
        temp_house, temp_num = extract_house_and_number(temp)

        stop = False

        while space > 0 and not stop:
            prev_house, prev_num = extract_house_and_number(arr[space - 1])
            
            if (prev_num > temp_num) or (prev_num == temp_num and prev_house > temp_house):
                arr[space] = arr[space - 1]
                space -= 1
            else:
                stop = True

        arr[space] = temp

    return arr
```

```python
# Task 1.3

file = open("student_codes.txt", 'r')

code_list = []

for line in file:
    code_list.append(line.strip())
    
file.close()

code_list_sorted = insertion_sort_by_number(validate_student_code(code_list))

print(code_list_sorted)

team_blue = []
team_red = []

for code in code_list_sorted:
    if int(code[2:]) % 2 == 0:
        team_blue.append(code)
    else:
        team_red.append(code)

print("Team blue:", team_blue)
print("Team red:", team_red)
```

```python
# Task 1.4

def generate_reverse_matchups(team_blue, team_red):
    number_of_matchups = len(team_blue) if len(team_blue) > len(team_red) else len(team_red)

    for i in range(number_of_matchups):
        print(f"Round {i + 1}: ", end='')

        if i < len(team_blue) and i < len(team_red):
            print(f"{team_blue[i]} vs {team_red[len(team_red) - i - 1]}")
        elif i >= len(team_red):
            print(f"{team_blue[i]} vs BYE")
        elif i >= len(team_blue):
            print(f"BYE vs {team_red[len(team_red) - i - 1]}")

generate_reverse_matchups(team_blue, team_red)
```
