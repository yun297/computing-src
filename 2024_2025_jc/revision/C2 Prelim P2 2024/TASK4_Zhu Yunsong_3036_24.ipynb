```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("TRIP.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS Flight (
    FlightNo TEXT PRIMARY KEY,
    DepartCity TEXT,
    ArrivalCity TEXT,
    DepartTime TEXT,
    ArrivalTime TEXT
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Customer (
    CustomerNo INTEGER PRIMARY KEY,
    Name TEXT,
    Gender TEXT,
    DOB TEXT
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Ticket (
    TicketNo INTEGER PRIMARY KEY,
    CustomerNo INTEGER,
    FlightNo TEXT,
    Date TEXT,
    Seat TEXT,
    FOREIGN KEY (CustomerNo) REFERENCES Customer(CustomerNo),
    FOREIGN KEY (FlightNo) REFERENCES Flight(FlightNo)
)
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

conn = sqlite3.connect("TRIP.db")
cur = conn.cursor()

# Insert flights
flight_file = open("FLIGHT.txt", 'r')
flights = []

for line in flight_file:
    flights.append(line.strip().split(','))

flight_file.close()

cur.executemany("""
INSERT INTO Flight VALUES (?,?,?,?,?)
""", flights)

# Insert customers
customer_file = open("CUSTOMER.txt", 'r')
customers = []

for line in customer_file:
    customers.append(line.strip().split(','))

customer_file.close()

cur.executemany("""
INSERT INTO Customer VALUES (?,?,?,?)
""", customers)

# Insert tickets
ticket_file = open("TICKET.txt", 'r')
tickets = []

for line in ticket_file:
    tickets.append(line.strip().split(','))

ticket_file.close()

cur.executemany("""
INSERT INTO Ticket VALUES (?,?,?,?,?)
""", tickets)

# Commit SQL statements
conn.commit()
conn.close()
```

```python
# Display all the flights departing from Singapore in ascending order of departure time

conn = sqlite3.connect("TRIP.db")
cur = conn.cursor()

cur.execute("""
SELECT * FROM Flight
WHERE DepartCity = ?
ORDER BY DepartTime ASC
""", ("Singapore", ))

rows = cur.fetchall()

for flight in rows:
    print(flight)
```

```python
# Cancel Shao Xingjuan's tickets numbered 1023 and 1024

conn = sqlite3.connect("TRIP.db")
cur = conn.cursor()

cur.execute("""
SELECT CustomerNo FROM Customer
WHERE Name = ?
""", ("Shao Xingjuan", ))

rows = cur.fetchall()

if rows:
    customer_no = rows[0][0]

    delete_query = [
        (1023, customer_no),
        (1024, customer_no)
    ]

    cur.executemany("""
    DELETE FROM Ticket
    WHERE TicketNo = ? and CustomerNo = ?
    """, delete_query)

conn.commit()
conn.close()
```
