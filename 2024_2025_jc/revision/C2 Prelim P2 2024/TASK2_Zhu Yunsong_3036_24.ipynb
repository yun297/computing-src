```python
# Task 2.1

import csv

def read_csv(filename):
    file = open(filename, 'r')
    result = []

    for row in csv.reader(file):
        result.append(row)

    return result

colours = read_csv("paint.csv")
print(len(colours))
print(colours)
```

```python
# Task 2.2

def find(array, r, g, b):
    for col in array:
        if col[:3] == [str(r), str(g), str(b)]:
            return col[3]

    return "nil"

print(find(colours, 253, 253, 5))
print(find(colours, 5, 253, 253))
```

```python
# Task 2.3

def rgb(array, colour):
    colour_arr = ["red", "green", "blue"]

    if colour not in colour_arr:
        print("Invalid colour! Please enter 'red', 'green' or 'blue': ")
        return []

    key = colour_arr.index(colour)
    
    arr = array.copy()

    for i in range(len(arr) - 1):
        for j in range(len(arr) - 1 - i):
            if int(arr[j][key]) < int(arr[j + 1][key]):
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr

print(rgb(colours, 'yellow'))
print(rgb(colours, 'blue'))
```

```python
# Task 2.4

def bright(array):
    if len(array) <= 1:
        return array

    pivot = array[0]
    left = []
    right = []

    for item in array[1:]:
        if int(item[0]) + int(item[1]) + int(item[2]) > int(pivot[0]) + int(pivot[1]) + int(pivot[2]):
            left.append(item)
        else:
            right.append(item)

    return bright(left) + [pivot] + bright(right)

print(bright(colours))
```

```python
# Task 2.5

def menu():
    colours = read_csv("paint.csv")

    option = input("Choose your preference: enter 'red', 'green', 'blue', 'dark' or 'bright': ")
    while option not in ['red', 'green', 'blue', 'dark', 'bright']:
        option = input("Choose your preference: enter 'red', 'green', 'blue', 'dark' or 'bright': ")

    if option == 'red':
        result = rgb(colours, 'red')[:3]
    elif option == 'green':
        result = rgb(colours, 'green')[:3]
    elif option == 'blue':
        result = rgb(colours, 'blue')[:3]
    elif option == 'dark':
        result = bright(colours)[-3:]
    elif option == 'bright':
        result = bright(colours)[:3]

    print("Top 3 matching  colours:")
    for colour in result:
        print(colour[3])

menu()
menu()
menu()
```
