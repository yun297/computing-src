```python
# Task 1.1

def task1_1(input_value):
    if not type(input_value) == int and not str(input_value).isnumeric():
        print("Input value should be an integer or a string container an integer value!")
        return -1

    if len(str(input_value)) != 14:
        print("Input value should be 14 digits long!")
        return -1

    digits = []
    input_value = str(input_value)

    for i in range(len(input_value)):
        if (i + 1) % 2 == 0:
            digits.append(str(int(input_value[i]) * 2))
        else:
            digits.append(input_value[i])

    total = 0

    for num in digits:
        if int(num) >= 10:
            total += int(num[0]) + int(num[1])
        else:
            total += int(num)

    check_digit = 10 - (total % 10)

    return check_digit
```

```python
# Task 1.2

print(task1_1("12735671252317") == 8)
print(task1_1(12345678901234) == 7)
print(task1_1("abcdefghijklmn") == -1)
print(task1_1(123) == -1)
```

```python
# Task 1.3

def task1_3(input_value):
    if not type(input_value) == int and not str(input_value).isnumeric():
        print("Input value should be an integer or a string container an integer value!")
        return False

    if len(str(input_value)) != 15:
        print("Input value should be 15 digits long!")
        return False

    no_check_digit = str(input_value)[:-1]
    check_digit = task1_1(no_check_digit)
    return check_digit == int(str(input_value)[-1:])

print(task1_3("127356712523178"))
print(task1_3(123456789012347))
print(not task1_3("abcdefghijklmno"))
print(not task1_3(123))
```
