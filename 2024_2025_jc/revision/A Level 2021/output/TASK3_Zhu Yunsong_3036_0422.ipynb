```python
# Task 3.1

class Node:
    def __init__(self, integer_value):
        self.integer_value = integer_value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, integer_value):
        new_node = Node(integer_value)

        if not self.head:
            self.head = new_node
            return

        new_node.next = self.head
        self.head = new_node

    def delete(self, integer_value):
        if not self.head:
            return
            
        if self.head.integer_value == integer_value:
            self.head = self.head.next
            return

        ptr = self.head

        while ptr.next and ptr.next.integer_value != integer_value:
            ptr = ptr.next

        if ptr.next.integer_value == integer_value:
            ptr.next = ptr.next.next
            return

    def search(self, integer_value):
        if not self.head:
            return False

        ptr = self.head

        while ptr:
            if ptr.integer_value == integer_value:
                return True
            ptr = ptr.next

        return False

    def count(self):
        if not self.head:
            return 0

        count = 0
        ptr = self.head

        while ptr:
            count += 1
            ptr = ptr.next

        return count

    def to_string(self):
        if not self.head:
            return "[]"

        integers = []
        ptr = self.head

        while ptr:
            integers.append(str(ptr.integer_value))
            ptr = ptr.next

        result = '[' + ", ".join(integers) + ']'
        
        return result

file = open("Task3data.txt", 'r')
data = []

for line in file:
    data.append(int(line.strip()))

file.close()
# print(data)

ll = LinkedList()

for i in data:
    ll.insert(i)

print(ll.count())
# print(ll.search(11))
print(ll.to_string())
```

```python
# Task 3.2

class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()

    def insert(self, integer_value):
        new_node = Node(integer_value)

        if not self.head:
            self.head = new_node
            return

        if integer_value < self.head.integer_value:
            new_node.next = self.head
            self.head = new_node
            return

        ptr = self.head

        while ptr.next and ptr.next.integer_value < integer_value:
            ptr = ptr.next

        new_node.next = ptr.next
        ptr.next = new_node

file = open("Task3data.txt", 'r')
data = []

for line in file:
    data.append(int(line.strip()))

file.close()
# print(data)

sll = SortedLinkedList()

for i in data:
    sll.insert(i)

print(sll.count())
print(sll.to_string())
```

```python
# Task 3.2

class Queue(LinkedList):
    def __init__(self):
        super().__init__()

    def enqueue(self, integer_value):
        new_node = Node(integer_value)
        
        if not self.head:
            self.head = new_node
            return

        ptr = self.head

        while ptr.next:
            ptr = ptr.next

        ptr.next = new_node

    def dequeue(self):
        if not self.head:
            return

        return_val = self.head.integer_value

        self.head = self.head.next

        return return_val

file = open("Task3data.txt", 'r')
data = []

for line in file:
    data.append(int(line.strip()))

file.close()
# print(data)

q = Queue()

for i in data:
    q.enqueue(i)

print(q.count())
print(q.to_string())

for i in range(5):
    print(q.dequeue())

print(q.to_string())
```
