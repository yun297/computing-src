```python
# Task 2.1

def task2_1(filename):
    file = open(filename, 'r')

    result = []

    for line in file:
        result.append(int(line.strip()))

    return result

result = task2_1("TEN.txt")
print(result)
print(len(result))
```

```python
# Task 2.2

def task2_2(list_of_integers):
    arr = list_of_integers.copy()

    for i in range(1, len(arr)):
        space = i
        temp = arr[space]

        while space > 0 and arr[space - 1] > temp:
            arr[space] = arr[space - 1]
            space -= 1

        arr[space] = temp

    return arr

print(task2_2(task2_1("TEN.txt"))) 
```

```python
# Task 2.3

def task2_3(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers

    piv = list_of_integers[0]
    left = []
    right = []

    for num in list_of_integers[1:]:
        if num < piv:
            left.append(num)
        else:
            right.append(num)

    return task2_3(left) + [piv] + task2_3(right)

print(task2_3(task2_1("TEN.txt")))
```

```python
# Task 2.4

import timeit

print("Insertion sort")
print("")

t10 = task2_1('TEN.txt')
time10 = timeit.timeit(lambda: task2_2(t10), number=1)
print(time10)

t100 = task2_1('HUNDRED.txt')
time100 = timeit.timeit(lambda: task2_2(t100), number=1)
print(time100)

t1000 = task2_1('THOUSAND.txt')
time1000 = timeit.timeit(lambda: task2_2(t1000), number=1)
print(time1000)

print("\nMerge sort")

t10 = task2_1('TEN.txt')
time10 = timeit.timeit(lambda: task2_3(t10), number=1)
print(time10)

t100 = task2_1('HUNDRED.txt')
time100 = timeit.timeit(lambda: task2_3(t100), number=1)
print(time100)

t1000 = task2_1('THOUSAND.txt')
time1000 = timeit.timeit(lambda: task2_3(t1000), number=1)
print(time1000)

# Insertion sort has a time complexity of O(n²) in the average and worst cases, while merge sort has a time complexity of O(n log n). 
# For very small inputs, insertion sort can run faster because it has little overhead, whereas merge sort requires recursive calls and
# additional memory for merging. However, as input size increases, the quadratic growth of insertion sort causes its runtime to increase
# much more rapidly than merge sort, whose growth is slower due to O(n log n). This means insertion sort is only efficient for small
# datasets, while merge sort is much more scalable and efficient for larger inputs.
```
