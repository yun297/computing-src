```python
# Task 3.1

class Person:
    def __init__(self, name: str, personID: str):
        self.name = name
        self.personID = personID

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}"


class Student(Person):
    def __init__(self, name: str, personID: str, course: str, year: int):
        super().__init__(name, personID)
        self.course = course
        self.year = year

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}, Course: {self.course}, Year: {self.year}"


class Staff(Person):
    def __init__(self, name: str, personID: str, department: str, position: str, salary: int):
        super().__init__(name, personID)
        self.department = department
        self.position = position
        self.salary = salary

    def getDetails(self):
        return f"Name: {self.name}, ID: {self.personID}, Department: {self.department}, Position: {self.position}, Salary: {self.salary}"
```

```python
# Task 3.2

class LinkedList:
    def __init__(self, data):
        self.data = data
        self.pointer = None

    def insert_rec(self, data):
        if self.data == None:
            self.data = data
            
        elif self.pointer:
            self.pointer.insert_rec(data)
            
        else:
            self.pointer = LinkedList(data)

    def delete_rec(self, data):
        if self.data.personID == data:
            self.data = self.pointer.data
            self.pointer = self.pointer.pointer
            
        elif self.pointer:
            if self.pointer.data.personID == data:
                self.pointer = self.pointer.pointer
                
            else:
                self.pointer.delete_rec(data)

        self.pointer.delete_rec(data)

    def display_rec(self):
        print(self.data.getDetails())
        
        if self.pointer:
            self.pointer.display_rec()
    
```

```python
# Task 3.3

class StudentLinkedList(LinkedList):
    def __init__(self, data = None, count = 1, freq = None):
        super().__init__(data)
        self.studentCount = count
        self.courseFreq = freq

    def updateCourseFreq(self):
        courseFreq = {}
        courseFreq[self.data.course] = 1
        
        ptr = self.pointer
        
        while ptr:
            if ptr.data.course not in courseFreq.keys():
                courseFreq[ptr.data.course] = 1
                
            else:
                courseFreq[ptr.data.course] += 1
                
            ptr = ptr.pointer
            
        self.courseFreq = courseFreq
    
    def displayCourseFreq(self):
        for set in self.courseFreq.items():
            print(f"Course: {set[0]}, Students: {set[1]}")

    def insert_rec(self, data):
        super().insert_rec(data)
        self.studentCount += 1
        self.updateCourseFreq()
    
    def delete_rec(self, data):
        super().delete_rec(data)
        self.studentCount -= 1
        self.updateCourseFreq()

class StaffLinkedList(LinkedList):
    def __init__(self, data=None, count=1):
        super().__init__(data)
        self.staffCount = count

    def insert_rec(self, data):
        super().insert_rec(data)
        self.staffCount += 1

    def delete_rec(self, data):
        super().delete_rec(data)
        self.staffCount -= 1
```
