```python
# Task 1.1

from pymongo import MongoClient

client = MongoClient("127.0.0.1", 27017)

online_shop = client["online_shop"]

products = online_shop["products"]
customers = online_shop["customers"]
orders = online_shop["orders"]
```

```python
import json

products_json = open("PRODUCTS.json", 'r')
customers_json = open("CUSTOMERS.json", 'r')
orders_json = open("ORDERS.json", 'r')

products_data = json.load(products_json)
customers_data = json.load(customers_json)
orders_data = json.load(orders_json)

products_json.close()
customers_json.close()
orders_json.close()

products.insert_many(products_data)
customers.insert_many(customers_data)
orders.insert_many(orders_data)
```

```python
# Task 1.2

from pprint import pprint

query = {'category': 'Books'}

for product in products.find(query):
    pprint(product)
```

```python
# Task 1.3

charles_id = customers.find_one({'name': 'Charles'}).get("customer_id")
# print(charles_id)

for order in orders.find({'customer_id': charles_id}):
    pprint(order)
```

```python
# Task 1.4

prices = {}

for product in products.find():
    prices[product.get("product_id")] = product.get("price")

print(prices)

print("OrderID | Revenue")

for order in orders.find():
    products_list = order.get("products")
    revenue = 0

    for p in products_list:
        revenue += prices[p.get("product_id")] * p.get("quantity")

    print(f"{order.get("order_id"):<7} | {revenue}")
```
