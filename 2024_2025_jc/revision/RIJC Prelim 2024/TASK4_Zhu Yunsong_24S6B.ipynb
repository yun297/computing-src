```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("CLINIC.db")
cur = conn.cursor()

patient_file = open("PATIENT.txt", 'r')
staff_file = open("STAFF.txt", 'r')
appointment_file = open("APPOINTMENT.txt", 'r')

patients = []
staffs = []
appointments = []

for line in patient_file:
    patients.append(line.strip().split(','))

for line in staff_file:
    staffs.append(line.strip().split(','))

for line in appointment_file:
    appointments.append(line.strip().split(','))

patient_file.close()
staff_file.close()
appointment_file.close()

cur.executemany("""
INSERT INTO Patient (PatientID, Name, DOB, ContactNumber) VALUES (?, ?, ?, ?)
""", patients)

cur.executemany("""
INSERT INTO Staff (StaffID, Name, Role, Specialization) VALUES (?, ?, ?, ?)
""", staffs)

cur.executemany("""
INSERT INTO Appointment (AppointmentID, PatientID, StaffID, AppointmentDate, Diagnosis) VALUES (?, ?, ?, ?, ?)
""", appointments)

conn.commit()

conn.close()
```

```python
# Task 4.2

conn = sqlite3.connect("CLINIC.db")
cur = conn.cursor()

cur.execute("""
SELECT a.AppointmentID, p.PatientID, p.Name, s.StaffID, s.Name, a.AppointmentDate, a.Diagnosis
FROM Appointment AS a
INNER JOIN Patient AS p 
ON p.PatientID = a.PatientID
INNER JOIN Staff AS s
ON a.StaffID = s.StaffID
""")

fetched = cur.fetchall()

for line in fetched:
    print(line)

conn.close()
```

```python
# Task 4.3

conn = sqlite3.connect("CLINIC.db")
cur = conn.cursor()

cur.execute("""
SELECT s.StaffID, s.Name, COUNT(*)
FROM Appointment AS a
INNER JOIN Staff AS s
ON a.StaffID = s.StaffID
GROUP BY s.StaffID
""")

fetched = cur.fetchall()

for line in fetched:
    print(line)
```
