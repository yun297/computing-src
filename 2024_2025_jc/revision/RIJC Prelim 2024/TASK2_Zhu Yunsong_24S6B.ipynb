```python
# Task 2.1

def insertion_sort(data_list: list, index: int, ascending: bool):
    arr = data_list.copy()

    for i in range(1, len(data_list)):
        space = i
        compare = arr[space]

        if ascending == True:
            while space > 0 and arr[space - 1][index] > compare[index]:
                arr[space] = arr[space - 1]
                space -= 1
        else:
            while space > 0 and arr[space - 1][index] < compare[index]:
                arr[space] = arr[space - 1]
                space -= 1

        arr[space] = compare

    return arr
```

```python
# Task 2.2

def task2_2(filename: str):
    f = open(filename, 'r')
    result = []

    for l in f:
        array = l.strip().split(',')
        
        result.append(tuple(array[:3] + [int(array[3])]))

    f.close()

    return result

lst = task2_2("CODING_COMPETITION.txt")
```

```python
# Task 2.3

def task2_3():
    sorted_by_country = insertion_sort(lst, 2, True)
    
    countries = []
    
    for participant in sorted_by_country:
        if participant[2] not in countries:
            countries.append(participant[2])
    
    result = []
    
    for country in countries:
        country_grouped = []
        
        for p in sorted_by_country:
            if p[2] == country:
                country_grouped.append(p)
    
        country_sorted = insertion_sort(country_grouped, 3, False)
    
        result += country_sorted
    
    write = ""
    
    for p in result:
        write += f"{p[2]},{p[0]},{p[1]},{p[3]}\n"

    f = open("SCORE_BY_COUNTRY.txt", 'w')
    f.write(write)
    f.close()

task2_3()

        
```
