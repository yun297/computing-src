### Task 4.2 [9]
Complete the backend app.py in the following code cell. You do not need to run the code.

````python
# Task 4.2 app.py

from flask import Flask, render_template, request

app = Flask(__name__)

@app.route("/", methods = ["GET", "POST"])
def index():
    if request.method == "GET":
        return render_template("index.html")

    elif request.method == "POST":
        m = request.form.get("m").lower().strip()
        k = request.form.get("k")

        k_strings = ["one", "two", "three", "four", "five", "six", "seven", "eight", "nine", "ten"]
        k_string = k_strings[int(k) - 1]

        ret_str = ""

        for letter in k_string:
            if letter in m:
                ret_str += letter

        if ret_str == "":
            ret_str = "Nil"

        return render_template("index.html", ret_str = ret_str)


if __name__ == "__main__":
    app.run()
````

### Task 4.3 [7]
Write Python and SQL code to create and connect to a database `magik.db` with the table `mk`. The table should have the following fields:

- a system generated integer primary key `mid`
- a text field m
- an integer field k

Use a suitable data structure to store and insert the following records into your table:

- "Good morning!", 5
- "Computing", 2
- "Bye", 4

You should run and test your code to show that the records are successfully inserted.  

````python
# Task 4.3

import sqlite3

conn = sqlite3.connect("magik.db")
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS mk") # for reruns

cursor.execute("""
    CREATE TABLE mk (
        mid INTEGER PRIMARY KEY AUTOINCREMENT,
        m TEXT NOT NULL,
        k INT NOT NULL
    )
""")

records = [
    ("Good morning!", 5),
    ("Computing", 2),
    ("Bye", 4)
]

for record in records:
    cursor.execute("""
        INSERT INTO mk (m, k)
        VALUES (?, ?)
    """, record)

conn.commit()

cursor.execute("""
    SELECT * FROM mk
""")

cursor.execute("SELECT * FROM mk")
rows = cursor.fetchall()

print("Records in mk table:")
for row in rows:
    print(row)

conn.close()
````
