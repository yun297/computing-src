### Task 3.1 [4]
Generate a dictionary of 500 student IDs from S001 to S500 with randomly generated marks in the range 0 to 100.

```python
# Task 3.1
from random import randint

students_marks = {}

for i in range(1, 501):
    students_marks[f"S{i:03}"] = randint(0, 100)

for student_id, mark in students_marks.items():
    print(f"{student_id}: {mark}")
```

### Task 3.2 [6]
Preprocess the marks such that they can be inserted to the BST to achieve optimal search performance. Explain in comments your preprocessing logic.

Note: If you are unable to do this in time, you may still proceed to do the later parts. :)

```python
# Task 3.2

# Merge sort to sort dictionary by students' marks
def merge_sort(dictionary):
    arr = list(dictionary.items())

    def merge(arr):
        if len(arr) <= 1:
            return arr
        
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        merge(left)
        merge(right)

        l = r = i = 0

        while l < len(left) and r < len(right):
            if left[l][1] < right[r][1]:
                arr[i] = left[l]
                l += 1
            else:
                arr[i] = right[r]
                r += 1
            i += 1

        while l < len(left):
            arr[i] = left[l]
            i += 1
            l += 1

        while r < len(right):
            arr[i] = right[r]
            i += 1
            r += 1

        return arr

    return merge(arr)

students_marks_arr = merge_sort(students_marks)

# Balancer function

def balancer(arr):
    if not arr:
        return []
        
    mid = len(arr) // 2
    root = arr[mid]

    left = balancer(arr[:mid]) # already skips mid element
    right = balancer(arr[mid + 1:]) # +1 to skip mid element

    return [root] + left + right

students_marks_arr = balancer(students_marks_arr)

for student in students_marks_arr: # convert items back into dict
    students_marks[student[0]] = student[1]
```

### Task 3.3 [13]
Implement the BST of queues data structure with insert, inorder traversal and search capabilities.

```python
# Task 3.3

class TreeNode:
    def __init__(self, student_id: str, mark: int):
        self.student_id = student_id
        self.mark = mark
        self.left = None
        self.right = None

class Tree:
    def __init__(self):
        self.root = None

    def insert(self, student_id: str, mark: int):
        new_student = TreeNode(student_id, mark)
        
        if not self.root:
            self.root = new_student
            return

        ptr = self.root
        inserted = False

        while not inserted:
            if new_student.mark > ptr.mark:
                if not ptr.right:
                    ptr.right = new_student
                    inserted = True
                else:
                    ptr = ptr.right
            else:
                if not ptr.left:
                    ptr.left = new_student
                    inserted = True
                else:
                    ptr = ptr.left

    def inorder(self):
        def inorder_helper(root: TreeNode):
            if not root:
                return
            inorder_helper(root.left)
            print(f"{root.student_id}: {root.mark}")
            inorder_helper(root.right)

        inorder_helper(self.root)

    def search(self, student_id: str):
        def search_helper(root: TreeNode):
            if not root:
                return
            result = search_helper(root.left) # result will be defined if found
            if result:
                return result
            if root.student_id == student_id:
                return root
            return search_helper(root.right)

        result = search_helper(self.root)

        if result:
            print(f"{result.student_id}: {result.mark}")
        else:
            print("Student not found.")

    def find_min(self):
        if not self.root:
            print("Tree is empty.")
            return

        ptr = self.root
        while ptr.left:
            ptr = ptr.left

        print(f"Lowest Mark: {ptr.student_id}: {ptr.mark}")

    def find_max(self):
        if not self.root:
            print("Tree is empty.")
            return

        ptr = self.root
        while ptr.right:
            ptr = ptr.right

        print(f"Highest Mark: {ptr.student_id}: {ptr.mark}")


student_tree = Tree()
for student_id, mark in students_marks.items():
    student_tree.insert(student_id, mark)

student_tree.inorder()
student_tree.search("S124")
```
