The text file [`LIFE.csv`](https://drive.google.com/file/d/103-EWut9ed2-LVf3Va2VO1Ak7l9_Bv1j/view?usp=sharing) contains collated data by World Bank on global life expectancy with socio-economic variables from about 2000-2019.

Life expectancy at birth indicates the number of years a newborn infant would live if prevailing
patterns of mortality at the time of its birth were to stay the same throughout its life. It is a key
metric for assessing population health.

Life expectancy has burgeoned since the advent of industrialization in the early 1900s and the world average has now more than doubled to 70 years. Yet, we still see inequality in life expectancy across and within countries. Further analysis is necessary to determine how the allocation of a country’s wealth through investments in healthcare, education, environmental management, and other socioeconomic factors have an overall effect in determining average life expectancy.

## Task 1.1 [5]
Extract the columns `CountryName`, `CountryCode`, `IncomeGroup` and `LifeExpectancy` into a Python list named `lifes`. Skip the header row and any data rows containing missing values. Output the contents of `lifes` using a loop.

PS: If you do not know how to do this by code, you may preprocess the dataset manually to solve the remaining subtasks. 🙂

````python
# Task 1.1

import csv

LIFE_csv = open("LIFE.csv", "r")
lifes = []

for line in csv.reader(LIFE_csv):
    if line[0] and line[1] and line[3] and line[5]:  # Check for missing values
            entry = [line[0], line[1], line[3], line[5]]
            lifes.append(entry)

lifes = lifes[1:]

for entry in lifes:
    entry[3] = float(entry[3])

for entry in lifes:
    print(entry)
````

## Task 1.2 [7]
Determine the countries with the highest and lowest life expectancies. Your output should look like the following:
```
Highest life expectancy: Country A (99.9)
Lowest life expectancy: Country S (45.0)
```
You may not use the built-in `max()`, `min()` and `sort()` functions.

````python
# Task 1.2

highest_country = ""
highest_value = 0
lowest_country = ""
lowest_value = 0

for entry in lifes:
    if highest_value == 0:
        highest_value = entry[3]
    if lowest_value == 0:
        lowest_value = entry[3]
        
    if entry[3] > highest_value:
        highest_value = entry[3]
        highest_country = entry[0]
    
    if entry[3] < lowest_value:
        lowest_value = entry[3]
        lowest_country = entry[0]

print(f"Highest life expectancyy: {highest_country} ({highest_value})")

print(f"Lowest life expectancyy: {lowest_country} ({lowest_value})")
````

## Task 1.3 [6]
It is hypothesised that countries with higher life expectancies tend to be from the higher income groups and countries with lower life expectancies tend to be from the lower income groups.

Provide in program comments how you would verify this hypothesis empirically. Implement your experiment to prove or disprove this claim.

Hint: One approach you may consider is to make use of a dictionary with frequency counts.

````python
# Task 1.3
# Experiment design:

# Step 1: Create a dictionary to group life expectancies by income group.
# The keys will be income groups (e.g., "High income", "Low income"),
# and the values will be a list of life expectancy values for each country in that group.

# Step 2: Loop through the 'lifes' list and populate this dictionary.
# Convert life expectancy to float while doing this.

# Step 3: For each income group, calculate the average life expectancy.

# Step 4: Print out the average life expectancy for each income group.
# This will allow us to observe whether higher income groups do indeed have higher life expectancy.

income_groups = {}

for entry in lifes:
    income_group = entry[2]
    life_expectancy = entry[3]

    if income_group not in income_groups:
        income_groups[income_group] = []

    income_groups[income_group].append(life_expectancy)

print("Average life expectancy by income group:")

for group, life_expectancy in income_groups.items():
    if life_expectancy:
        avg = sum(life_expectancy) / len(life_expectancy)
        print(f"{group}: {avg:.1f}")

# The data supports the hypothesis.
````
