```python
# Task 2.1

from pymongo import MongoClient

client = MongoClient("localhost", 27017)
# client.server_info()

Store = client["Store"]
Order = Store["Order"]
```

```python
import json

drink_file = open("DRINK.json")
drink = json.load(drink_file)
drink_file.close()

Order.insert_many(drink)
```

```python
proj = {"_id": 0}

# Header
header = f"{'Date':<10}{'Time':<10}{'Tea':<25}{'Sugar':<7}{'Price':<7}{'Upsize':<8}{'Ice':<8}{'Toppings':<40}{'Member':<15}"
print(header)
print("-" * len(header))

# Rows
for o in Order.find({}, proj):
    date = o.get("Date", "")
    time = o.get("Time", "")
    tea = o.get("Tea", "")
    sugar = o.get("Sugar", "")
    price = o.get("Price", "")
    upsize = o.get("Upsize", "")
    ice = o.get("Ice", "")
    toppings = o.get("Toppings", [])
    member = o.get("Member", "")

    # If toppings is a list, join it nicely
    if isinstance(toppings, list):
        toppings = ", ".join(toppings)
    
    message = f"{date:<10}{time:<10}{tea:<25}{sugar:<7}{price:<7}{str(upsize):<8}{ice:<8}{toppings:<40}{member:<15}"
    print(message)
```

```python
query = {"Member": "Jason"}

number_of_orders_jason = Order.count_documents(query)

print("Number of orders by Jason:", number_of_orders_jason)
```

```python
price = 0
orders = 0

for o in Order.find():
    price += o["Price"]
    orders += 1

print(f"Average price per order: ${price/orders:.2f}")
```
