```python
# Task 4.1

import sqlite3

conn = sqlite3.connect("Airport.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE Flight (
    flightNum TEXT PRIMARY KEY,
    departure TEXT,
    destination TEXT,
    departureTime TEXT,
    arrivalTime TEXT
)
""")

conn.commit()
conn.close()
```

```python
# Task 4.2

f = open("TASK4.txt", 'r')

data = []

for line in f:
    data.append(line.strip().split(','))

data = data[1:]

f.close()
```

```python
conn = sqlite3.connect("Airport.db")
cur = conn.cursor()

cur.executemany("""
INSERT INTO Flight (flightNum, departure, destination, departureTime, arrivalTime)
VALUES (?, ?, ?, ?, ?)
""", data)

conn.commit()
conn.close()
```

```python
conn = sqlite3.connect("Airport.db")
cur = conn.cursor()

cur.execute("SELECT * FROM Flight")

fetched_rows = cur.fetchall()
for r in fetched_rows:
    print(r)

conn.close()
```
