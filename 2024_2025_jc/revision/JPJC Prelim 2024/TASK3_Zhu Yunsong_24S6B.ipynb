```python
# Task 3.1

class Node:
    def __init__(self, value: int):
        self.value = value
        self.left = None
        self.right = None
```

```python
# Task 3.2

class Tree:
    def __init__(self, root_node: Node):
        self.root = root_node

    def insert(self, new_node):
        ptr = self.root
        inserted = False

        while not inserted:
            if new_node.value < ptr.value:
                if ptr.left:
                    ptr = ptr.left
                else:
                    ptr.left = new_node
                    inserted = True
            else:
                if ptr.right:
                    ptr = ptr.right
                else:
                    ptr.right = new_node
                    inserted = True

    def in_order_traversal(self):
        def in_order_helper(root):
            if root:
                in_order_helper(root.left)
                print(root.value)
                in_order_helper(root.right)

        in_order_helper(self.root)

    def flatten(self):
        def flatten_helper(root, prev = [None]): # Use array so it is mutable
            if not root:
                return
    
            # If there's a previous node, link its right to this node
            if prev[0]:
                prev[0].right = root
                prev[0].left = None  # remove left link
    
            # Update prev to this node
            prev[0] = root
    
            # Save right child before overwriting
            right = root.right
    
            # Recurse left first, then right
            flatten_helper(root.left, prev)
            flatten_helper(right, prev)
    
        flatten_helper(self.root)
```

```python
# Create the BST
BST = Tree(Node(5))
BST.insert(Node(3))
BST.insert(Node(2))
BST.insert(Node(4))
BST.insert(Node(7))
BST.insert(Node(8))

# Print BST in-order (sorted order)
print("In-order traversal of BST:")
BST.in_order_traversal()

# Flatten the BST into a linked list
BST.flatten()

# Traverse the flattened linked list and print values
print("\nTraversal of flattened linked list:")
ptr = BST.root
while ptr:
    print(ptr.value, end=" -> ")
    ptr = ptr.right
```
