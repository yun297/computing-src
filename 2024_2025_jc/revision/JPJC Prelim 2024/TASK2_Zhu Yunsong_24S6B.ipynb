```python
# Task 2.1

from random import randint

def createIsland(rowsize, colsize):
    grid = []

    for i in range(rowsize):
        grid.append(['.'] * colsize)

    grid[0][0] = '*'

    return grid

def plantCoconuts(grid):
    r = len(grid) - 1
    c = len(grid[0]) - 1
    
    count = 0
    coconut_count = 3
    
    while count != coconut_count:
        i = randint(0, r)
        j = randint(0, c)
        if grid[i][j] == '.':
            grid[i][j] = 'C'
            count += 1
    return grid

def display(grid):
    for r in grid:
        print(' '.join(r))

def move(grid, current, direction, distance, coconut):
    offset = {
        'L': (0, -1),
        'R': (0, 1),
        'U': (-1, 0),
        'D': (1, 0)
    }

    y_dir = offset[direction][0]
    x_dir = offset[direction][1]

    if current[0] + (distance * y_dir) > len(grid) - 1 or current[1] + (distance * x_dir) > len(grid[0]) - 1:
        print("Out of range!")
        return False

    for i in range(distance - 1):
        new_y = current[0] + ((i + 1) * y_dir)
        new_x = current[1] + ((i + 1) * x_dir)
        
        if grid[new_y][new_x] != 'C':
            grid[new_y][new_x] = '*'

    new_pos = (current[0] + (distance * y_dir), current[1] + (distance * x_dir))
    
    if grid[new_pos[0]][new_pos[1]] == 'C':
        grid[new_pos[0]][new_pos[1]] = 'E'
        coconut += 1
    else:
        grid[new_pos[0]][new_pos[1]] = '*'

    current = new_pos
    return(grid, current, coconut)
```

```python
# Task 2.2

def main():
    grid = createIsland(10, 15)
    plantCoconuts(grid)
    
    current = (0, 0)
    coconut = 0
    moves = 0
    
    display(grid)

    while (current != (9, 14) or coconut != 3) and moves != 8:
        print(f"Your are at position: {current} | Coconuts eaten: {coconut}")

        moves += 1
        user_input = input("Enter your move (L/R/U/D distance): ")
        direction, distance = user_input.split(' ')
        
        grid, current, coconut = move(grid, current, direction, int(distance), coconut)
        display(grid)

    pos_c = current == (9, 14)
    coconut_c = coconut == 3

    if pos_c and coconut_c:
        print("You have won!\nYou have reached finish point and eaten 3 coconuts.")
    else:
        print("Sorry you did not win.")
        if coconut_c and not pos_c:
            print("You did not reach the finish point.")
        elif pos_c and not coconut_c:
            print("You have not eaten enough coconut.")
        else:
            print("You did not reach the finish point.\nYou have not eaten enough coconut.")

main()
```
