```python
# Task 1.1

def CharToNum(char: str):
    return ord(char) - 65

print(CharToNum("Z"))
```

```python
# Task 1.2

def encrypt(text: str, key: str):
    def NumToChar(num: str):
        return chr(num + 65)
    
    text_num = []
    for c in text:
        text_num.append(CharToNum(c))
        
    key_repeated = ""
    for i in range(len(text)):
        key_repeated += key[i % len(key)]

    key_num = []
    for c in key_repeated:
        key_num.append(CharToNum(c))

    encrypt_num = ""
    for i in range(len(text)):
        if text_num[i] == -33:
            encrypt_num += '!'
        else:
            encrypt_num += NumToChar((text_num[i] + key_num[i]) % 26)

    return encrypt_num
```

```python
f = open("TASK1.txt", 'r')
write_f = open("ENCRYPTED.txt", 'w')

key = "JPJC"
messages = []

for line in f:
    messages.append(line.strip())

write = ""

for m in messages:
    write += f"{encrypt(m, key)}\n"

write_f.write(write)

f.close()
write_f.close()
```
