```python
#Task 1.1

from pymongo import MongoClient
client = MongoClient('localhost', 27017)

Travel = client["Travel"]
Flight = Travel["Flight"]
```

```python
import json

ticket_json_file = open("TICKET.json", "r")
ticket_json = json.load(ticket_json_file)
ticket_json_file.close()

# print(ticket_json)

Flight.insert_many(ticket_json)
```

```python
proj = {'_id': 0}

print(f"{'country':<15}{'city':<15}{'price':<10}{'company':<15}{'durationHour':<15}{'stop':<2}")
for item in Flight.find({}, proj):
    try:
        print(f"{item['country']:<15}{item['city']:<15}{item['price']:<10}{item['company']:<15}{item['durationHour']:<15}{item['stop']:<2}")
    except:
        print(f"{item['country']:<15}{item['city']:<15}{item['price']:<10}{item['company']:<15}{item['durationHour']:<15}")
```

```python
# Task 1.2

query = {'stop': {'$exists': False}}
set = {'$set': {'stop': 0}}

# for item in Flight.find(query):
#     print(item)

result = Flight.update_many(query, set)

# print(result.modified_count)
```

```python
proj = {'_id': 0}
query = {'company': 'S Airlines'}

print(f"{'country':<15}{'city':<15}{'price':<10}{'company':<15}{'durationHour':<15}{'stop':<2}")
for item in Flight.find(query, proj):
    print(f"{item['country']:<15}{item['city']:<15}{item['price']:<10}{item['company']:<15}{item['durationHour']:<15}{item['stop']:<2}")
```

```python
proj = {'_id': 0, 'city': 1, 'price': 1}
query = {'durationHour': {'$lt': 10}, 'price': {'$lt': 1500}}

print(f"{'city':<15}{'price':<10}")
for item in Flight.find(query, proj):
    print(f"{item['city']:<15}{item['price']:<10}")
```

```python
proj = {'_id': 0}
sort = [('price', 1)]

print(f"{'country':<15}{'city':<15}{'price':<10}{'company':<15}{'durationHour':<15}{'stop':<2}")
for item in Flight.find({}, proj).sort(sort).limit(1):
    print(f"{item['country']:<15}{item['city']:<15}{item['price']:<10}{item['company']:<15}{item['durationHour']:<15}{item['stop']:<2}")
```
