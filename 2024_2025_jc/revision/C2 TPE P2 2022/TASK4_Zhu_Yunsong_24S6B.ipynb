```python
import socket
from random import randint

server = socket.socket()
server.bind(('127.0.0.1', 12345))
server.listen()

new_socket, addr = server.accept()
print("Connected to:", addr)

answer = str(randint(1000, 9999))
print(repr(answer))
tries = 0

while True:
    tries += 1
    if tries <= 5:
        new_socket.sendall(b'GUESS\n')
        guess = new_socket.recv(1024).decode().strip()

        if guess == answer:
            new_socket.sendall(b'WIN\n')
            break
            
        else:
            message = ''
            
            for i in range(4):
                if guess[i] == answer[i]:
                    message += 'G'
                elif guess[i] in answer:
                    message += 'Y'
                else:
                    message += 'R'
                    
            message += '\n'
            new_socket.sendall(message.encode())
    else:
        new_socket.sendall(b'LOSE\n')
        message = answer + '\n'
        new_socket.sendall(message.encode())
        break

server.close()
new_socket.close()
```
