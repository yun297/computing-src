```python
# Task 2.1

def task2_1(filename):
    file = open(filename, 'r')
    word_list = []

    file_str = file.read()
    file_str = file_str.lower()
    file_str = file_str.replace('\n', ' ').replace(',', ' ').replace('-', ' ')

    for word in file_str.split():
        word_list.append(word.strip())
        
    return word_list

    file.close()

print(len(task2_1('LYRICS.txt')))
```

```python
# Task 2.2

def task2_2(list_of_words):
    ret = []
    for i in range(len(list_of_words)):
        if i != len(list_of_words) -1 and list_of_words[i + 1] == "it" and list_of_words[i] not in ret:
            ret.append(list_of_words[i])

    return ret

result = task2_2(task2_1('LYRICS.txt'))
print(result)
print(len(result))
```

```python
# Task 2.3
from random import choices

def task2_3(list_of_words, number):
    word_list = choices(list_of_words, k=number)
    ret = ""

    for i in range(len(word_list)):
        if i == len(word_list) - 1:
            ret += (word_list[i] + " it.")
        else:
            ret += (word_list[i] + " it, ")

    return ret

print(task2_3(result, 8))
```
