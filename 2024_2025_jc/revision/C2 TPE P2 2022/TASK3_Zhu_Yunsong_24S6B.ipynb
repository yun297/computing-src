```python
    # Task 3.1
    
    class Node:
        def __init__(self, data: str = ""):
            self.leftPtr = -1
            self.data = data
            self.rightPtr = -1
            
    class BSTree:
        def __init__(self):
            self.thisTree = []
            for i in range(10):
                self.thisTree.append(Node())
    
            self.root = -1 # empty tree
            self.nextFree = 0
    
            # linking all the free nodes up
    
            for i in range(9):
                self.thisTree[i].leftPtr = i + 1
            self.thisTree[9].leftPtr = -1

        def add(self, newItem):
            if self.nextFree == -1:
                print("Tree is full!")
                return
            
            # store newItem at next free node
            newIndex = self.nextFree
            self.thisTree[newIndex].data = newItem

            # update the nextFree
            self.nextFree = self.thisTree[newIndex].leftPtr

            # set new node pointers
            self.thisTree[newIndex].leftPtr = -1
            self.thisTree[newIndex].rightPtr = -1

            if self.root == -1:
                self.root = newIndex
                return

            # if the tree is not empty,
            current = self.root
            inserted = False

            while not inserted:
                if newItem < self.thisTree[current].data:
                    if self.thisTree[current].leftPtr == -1:
                        self.thisTree[current].leftPtr = newIndex
                        inserted = True
                    else:
                        current = self.thisTree[current].leftPtr
                        
                elif newItem > self.thisTree[current].data:
                    if self.thisTree[current].rightPtr == -1:
                        self.thisTree[current].rightPtr = newIndex
                        inserted = True
                    else:
                        current = self.thisTree[current].rightPtr

        def displayInOrder(self):
            def inOrder(index):
                if index != -1:
                    inOrder(self.thisTree[index].leftPtr)
                    print(self.thisTree[index].data)
                    inOrder(self.thisTree[index].rightPtr)
                    
            inOrder(self.root)

        def displayPostOrder(self):
            def postOrder(index):
                if index != -1:
                    postOrder(self.thisTree[index].leftPtr)
                    postOrder(self.thisTree[index].rightPtr)
                    print(self.thisTree[index].data)
                    
            postOrder(self.root)

        def display(self):
            print(f"Root index: {self.root}")
            print(f"Next Free index: {self.nextFree}")

            print("Contents of the tree (in index order):")
            for i in range(10):
                print(f"Index {i}: {self.thisTree[i].data}")
```

```python
# Task 3.2

bst = BSTree()

names_txt = open("NAMES.txt", "r")
names = []

for line in names_txt:
    names.append(line.strip())

names_txt.close()

for name in names:
    bst.add(name)

bst.display()
bst.displayInOrder()
bst.displayPostOrder()

```
