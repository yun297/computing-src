### 2. Write an iterative algorithm for inserting an item in a tree.

Binary search property is maintained

````python
# 1., 2.

class TreeNode: 
  def __init__(self, item, left = None, right = None):
    self.item = item
    self.left = left
    self.right = right
    
class Tree:
  def __init__(self, root = None):
    self.root = root
    
  def InsertItem (self, tree: TreeNode, item):
    if self.root is None:
      self.root = TreeNode(item)
    
    elif item < tree.item:
      
      if tree.left:
        self.InsertItem(tree, item)
      
      else:
        tree.left = TreeNode(item)
        
    else:
      
      if tree.right:
        self.InsertItem(tree, item)
        
      else:
        tree.right = TreeNode(item)
    
  def SearchItem(self, tree: TreeNode, item: TreeNode):
    if tree is None:
      return None, None
    
    elif tree.item == item:
      return tree, None

    elif tree < item:
      return self.SearchItem(tree.left, item)
    
    else:
      return self.SearchItem(tree.left, item)
      
      
````
