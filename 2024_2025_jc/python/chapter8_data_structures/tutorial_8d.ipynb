### 3. For each sequence of characters, draw the binary search tree and traverse the tree using
inorder, preorder, and postorder.  
(a) M, T, V, F, U, N  
(b) F, L, O, R, I, D, A

````python
# 1., 2., 3.

class TreeNode: 
  def __init__(self, item, left = None, right = None):
    self.item = item
    self.left = left
    self.right = right
    
class Tree:
  def __init__(self, root = None):
    self.root = root
  
  # 2.
  def InsertItem (self, tree: TreeNode, item):
    if self.root is None:
      self.root = TreeNode(item)
    
    elif item < tree.item:
      
      if tree.left:
        self.InsertItem(tree, item)
      
      else:
        tree.left = TreeNode(item)
        
    else:
      
      if tree.right:
        self.InsertItem(tree, item)
        
      else:
        tree.right = TreeNode(item)
  
  # 1.
  def SearchItem(self, tree: TreeNode, item: TreeNode):
    if tree is None:
      return None, None
    
    elif tree.item == item:
      return tree, None

    elif tree < item:
      return self.SearchItem(tree.left, item)
    
    else:
      return self.SearchItem(tree.left, item)
      
  def InOrder(self, tree):
    if tree:
      self.InOrder(tree.left)
      print(tree.data)
      self.InOrder(tree.right)
      
  def PreOrrder(self, tree):
    if tree:
      print(tree.data)
      self.PreOrrder(tree.left)
      self.PreOrrder(tree.right)
      
  def PostOrder(self, tree):
    if tree:
      self.PostOrder(tree.left)
      self.PostOrder(tree.right)
      print(tree.data)
      
````
