### 9.1.1 Linear Search $O(n)$

A **linear search** is the most basic search method that works on any list. However, this is also the least efficient.

The analogy for this is having a line of boxes with a number in each of them. To check if a certain number is in (at least) one of the boxes, you'll need to check through all the boxes. (or until a the number is found)

Write a Linear Search Algoithm `LinearSearch(list_to_search, key)` such that the index is return if `key` found and `-1` is returned if the `key` is not in the list.

```python
# Linear Search Algorithm

def LinearSearch(list_to_search, key):
  for i in range(len(list_to_search)):
    if list_to_search[i] == key:
      return i
  return -1
```

```python
# example 1: Given [6,4,6,8,3,9,5,5] and trying to find 0

print(LinearSearch([6,4,6,8,3,9,5,5], 0))
```

### 9.1.2 Binary Search $O(\lg n)$

A binary search is the usual way we search for anything if we have a `sorted list`.

Examples: looking through a calendar/schedule, finding a particular page in a book, number guessing...

Write a Linear Search Algoithm `BinarySearch(list_to_search, key)` such that the index is return if `key` found and `-1` is returned if the `key` is not in the list.

```python
# Iterative Binary Search Algorithm

def BinarySearch(list_to_search, key):
  min = 0
  max = len(list_to_search) - 1
  
  while min <= max:
    mid = (min + max) // 2
    
    if list_to_search[mid] == key:
      return mid
    elif list_to_search[mid] > key:
      max = mid - 1
    else:
      min = mid + 1
      
  return -1
#
```

#### Binary Search (recursive)

Binary Search can also be done recursively. However, we will need to keep track of more parameters if we want a recursive method.

```python
# Binary Search Algorithm (Recursive)

def BinarySearchR(list_to_search, key, min = 0, max = None):
  if max is None:
    max = len(list_to_search) - 1

  if min > max:
    return -1

  mid = (min + max) // 2

  if list_to_search[mid] == key:
    return mid
  elif list_to_search[mid] > key:
    return BinarySearchR(list_to_search, key, min, mid - 1)
  else:
    return BinarySearchR(list_to_search, key, mid + 1, max)
  
```

Linear probing refers to the method of finding the next available slot in the hash table.


```python
arr = []

# Initialize the array with 100 empty slots
for i in range(100):
    arr.append([0,'-'])
# print(arr)

def hash(num):
    return num % 100 # return hash key, moding by 100 returns the last 2 digits of student ID (index)

############################################

for [data, num] in s6b: # Storing data into hash table based on hash key
  key = hash(num)
  
  if arr[key][0] == 0: # If the hash key is empty, store the data
    arr[key] = [num, data]
    
  else: # If the hash key is not empty, find the next empty hash key
    while arr[key][0] != 0: # Keep searching for the next empty hash key until empty spot is found
      key = (key + 1) % 100 # Increment the hash key by 1 and mod by 100 to ensure it is within the range of 100
    arr[key] = [num, data]
    
counter = 1
for i in range(100):
  if arr[i][0] != 0:
    print(counter, arr[i][1])
    counter += 1
```

#### 9.1.3.2 Searching by Linear Probing (Collision Resolution)

When we have a collision, we will need to search through the hash table to find the next available slot.

For example we are looking for a person sitting at seat 13. If there is more than one person with hash key 13 in the class, searching for the person at 13 might not yield the correct person, as the person might be sitting at the next available seat (and so on).

So in order to find the right person, we will have to search through the hash table from 13 until the next empty seat.

```python
def hash_search(num, arr):
  key = hash(num)
  while arr[key][0] != 0:
    if arr[key][0] == num: # If the hash key matches the student ID, return the student name (i.e. they might not be sitting at the seat that matches their key)
      return arr[key]
    key += 1
  
  else: return None
  
print(hash_search(203013, arr))
print(hash_search(202913, arr))
print(hash_search(100913, arr))
```

### 9.2.1 Bubble Sort $O(n^2)$

Bubble sort is a straightforward comparison-based sorting algorithm that repeatedly traverses the list, comparing adjacent elements and swapping them if they are in the wrong order. This process continues until the entire list is sorted. The algorithm works by "bubbling" the largest (or smallest) element to its correct position in each pass.

```python
from random import choices

rand_nums = choices(list(range(100)), k = 10)
print(rand_nums)
```

```python
rand_nums = choices(list(range(100)), k = 10)

def bubble(og_arr):
  arr = og_arr.copy()
  
  for k in range(len(rand_nums)):
    for l in range(len(rand_nums) - 1):
      if arr[l] > arr[l + 1]:
        arr[l], arr[l + 1] = arr[l + 1], arr[l]
  
  return arr
        
# or

def bubble2(og_arr):
  arr = og_arr.copy()
  
  for k in range(len(rand_nums)):
    for l in range(1, len(rand_nums)):
      if arr[l] < arr[l - 1]:
        arr[l], arr[l - 1] = arr[l - 1], arr[l]
        
  return arr
        
print(rand_nums)
print(bubble(rand_nums))
print(bubble2(rand_nums))
```

### 9.2.2 Insertion Sort $O(n^2)$

Insertion sort is a simple and efficient sorting algorithm that works similarly to how we sort playing cards in our hands. It processes one element at a time, inserting it into its correct position relative to the already sorted section of the list. This continues until the entire list is sorted.

```python
rand_nums = choices(list(range(100)), k = 10)

def insertion(og_arr):
  arr = og_arr.copy()

  for i in range(len(arr)):
    curr_i = i
    temp = arr[i]
    while arr[curr_i - 1] > temp and curr_i != 0:
      arr[curr_i], arr[curr_i - 1] = arr[curr_i - 1], arr[curr_i]
      curr_i -= 1
  
  return arr
      
def insertion2(og_arr): #! broken
  arr = og_arr.copy()

  for i in range(len(arr)):
    temp = arr[i]
    space = i
    
    while (space > 0) and (arr[space - 1] > temp):
      arr[space] = arr[space - 1]
      space = -1
    arr[space] = temp
    
  return arr
    
print(rand_nums)
print(insertion(rand_nums))
print(insertion2(rand_nums))
```

### 9.2.3 Quick Sort $O(n^2)$

Quick sort is an efficient sorting algorithm that uses a divide-and-conquer strategy to sort the list. It works by selecting a pivot element from the list and partitioning the other elements into two sublists based on whether they are less than or greater than the pivot. The sublists are then recursively sorted.

```python
def quick(arr):
  if len(arr) <= 1:
    return arr
  
  pivot = arr[0]
  left = []
  right = []
  
  for item in arr[1:]: # everything other than pivot
    if item < pivot:
      left.append(item)
    else:
      right.append(item)
      
  left = quick(left)
  right = quick(right)
  
  return left + [pivot] + right

rand_nums = choices(list(range(100)), k = 10)

print(rand_nums)
print(quick(rand_nums))
```

This version of quick of sort is very inefficient as it takes up a lot of space. A more efficient version of quick sort is the in-place quick sort. The code below is the suggested code for quick sort (better than the one in the notes).

```python
def quick_sort(arr, left, right):
    if left < right: # at least 2 elements
        pivot = arr[left]
        space = left
        for i in range(left + 1, right + 1): # from left+1 until right
            if arr[i] < pivot:
                space += 1 # space and i increment together until i finds a element larger than pivot   , then space lag behind
                # when we reach the next pass where i is less than the pivot, we swap the elements at i and space, as space is the last element that is less than the pivot
                arr[i], arr[space] = arr[space], arr[i]
        arr[left], arr[space] = arr[space], arr[left]

        quick_sort(arr, left, space-1)
        quick_sort(arr, space+1, right)

def quick(arr):
    quick_sort(arr,0,len(arr)-1)
    
rand_nums = choices(list(range(100)), k = 10)

print(rand_nums)
quick(rand_nums)
print(rand_nums)
```

### 9.2.4 Merge Sort $O(n\log n)$

Merge sort is another efficient sorting algorithm that uses a divide-and-conquer strategy to sort the list. It works by dividing the list into two halves, sorting each half, and then merging the sorted halves back together. This process continues recursively until the entire list is sorted.

```python
def merge(arr):
  # edge case
  if len(arr) <= 1:
    return arr
  
  length = len(arr)
  
  mid = length // 2
  left = arr[:mid] # 0 to mid-1 (mid not included, mid elements)
  right = arr[mid:] # mid to end (mid included, mid or mid+1 elements)
  
  merge(left)
  merge(right)
  
  l = r = i = 0
  
  while l < len(left) and r < len(right):
    if left[l] < right[r]:
      # if left is smaller than right, put left into arr
      arr[i] = left[l]
      l += 1
    else:
      # if right is smaller than left, put right into arr
      arr[i] = right[r]
      r += 1
    i += 1 # increment i to move to the next element in left and right
    
  # if all the elements in left are already in arr, put the remaining elements in right into arr and vice versa
    
  while l < len(left):
    arr[i] = left[l]
    l += 1
    i += 1
    
  while r < len(right):
    arr[i] = right[r]
    r += 1
    i += 1
    
  return arr
    
rand_nums = choices(list(range(100)), k = 10)

print(rand_nums)
merge(rand_nums)
print(rand_nums)
```
