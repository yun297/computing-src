### Question 1

For the following array A, show A after each pass using **bubble sort** and **insertion sort** to arrange the elements in ascending order


| i   | 0  | 1  | 2  | 3  | 4  | 5  |
|-----|----|----|----|----|----|----|
| A[i]| 30 | 50 | 70 | 10 | 40 | 60 |

---

#### Bubble Sort (Ascending Order)
##### Pass 1:
- Compare `30 & 50` → No swap  
- Compare `50 & 70` → No swap  
- Compare `70 & 10` → Swap → `[30, 50, 10, 70, 40, 60]`  
- Compare `70 & 40` → Swap → `[30, 50, 10, 40, 70, 60]`  
- Compare `70 & 60` → Swap → `[30, 50, 10, 40, 60, 70]`  

##### Pass 2:
- Compare `30 & 50` → No swap  
- Compare `50 & 10` → Swap → `[30, 10, 50, 40, 60, 70]`  
- Compare `50 & 40` → Swap → `[30, 10, 40, 50, 60, 70]`  
- Compare `50 & 60` → No swap  

##### Pass 3:
- Compare `30 & 10` → Swap → `[10, 30, 40, 50, 60, 70]`  
- Compare `30 & 40` → No swap  
- Compare `40 & 50` → No swap  

##### Pass 4:
- Compare `10 & 30` → No swap  
- Compare `30 & 40` → No swap  

##### Pass 5:
- No swaps needed → Sorting complete  

**Final Sorted Array (Bubble Sort):** `[10, 30, 40, 50, 60, 70]`

---

#### Insertion Sort (Ascending Order)
##### Step 1: (Key = 50)
- 50 > 30 → No shift  
  **Array:** `[30, 50, 70, 10, 40, 60]`

##### Step 2: (Key = 70)
- 70 > 50 → No shift  
  **Array:** `[30, 50, 70, 10, 40, 60]`

##### Step 3: (Key = 10)
- 10 < 70 → Shift 70  
- 10 < 50 → Shift 50  
- 10 < 30 → Shift 30  
- Insert 10 at index 0  
  **Array:** `[10, 30, 50, 70, 40, 60]`

##### Step 4: (Key = 40)
- 40 < 70 → Shift 70  
- 40 < 50 → Shift 50  
- Insert 40 at index 2  
  **Array:** `[10, 30, 40, 50, 70, 60]`

##### Step 5: (Key = 60)
- 60 < 70 → Shift 70  
- Insert 60 at index 4  
  **Array:** `[10, 30, 40, 50, 60, 70]`

**Final Sorted Array (Insertion Sort):** `[10, 30, 40, 50, 60, 70]`

(a) Using a trace table with columns labelled `i`, `j`, `X[1]`, `X[2]`, `X[3]`, `X[4]`, and `X[5]`, show how the contents of array `X` are sorted into order when the following Bubble Sort algorithm is followed:  

```python
set i = 5  
repeat  
    for j = 1 to i - 1 do  
        if X[j] > X[j+1] then  
            swap X[j] and X[j+1]  
        endif  
    endfor  
    set i = i - 1  
until i = 1
```

(b) Rewrite the code so that the sort process terminates as soon as possible when the array is fully sorted.

```python
set i = 5  
set last_swap = i  
repeat  
    set check = last_swap
    last_swap = 0
    for j = 1 to check - 1 do  
        if X[j] > X[j+1] then  
            swap X[j] and X[j+1]
            last_swap = last_swap + 1
        endif  
    endfor  
    set i = i - 1  
until last_swap = 0
```

### Question 3

Data is stored in the array `NameList[1:10]`. This data is to be sorted in ascending order.  
Complete the pseudocode algorithm for an insertion sort.

```python
FOR ThisPointer <- 2 TO ____________  
    // use a temporary variable to store item which is to  
    // be inserted into its correct location  
    Temp <- NameList[ThisPointer]  
    Pointer <- ThisPointer – 1  

    WHILE (______________) AND (NameList[Pointer] > Temp)  
        // move list item to next location  
        NameList[_________] <- NameList[__________]  
        Pointer <- _________________  
    ENDWHILE  

    // insert value of Temp in correct location  
    NameList[___________] <- __________  
ENDFOR  
```

```python
FOR ThisPointer <- 2 TO 10  
    // use a temporary variable to store item which is to  
    // be inserted into its correct location  
    Temp <- NameList[ThisPointer]  
    Pointer <- ThisPointer – 1  

    WHILE (Pointer > 1) AND (NameList[Pointer] > Temp)  
        // move list item to next location  
        NameList[Pointer - 1] <- NameList[Pointer]  
        Pointer <- Pointer - 1  
    ENDWHILE  

    // insert value of Temp in correct location  
    NameList[Pointer] <- Temp  
ENDFOR  
```
