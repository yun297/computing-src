# Question 1

We want to use linear search to count the number of occurrences of a key in a list. When the key does not exist in the list, it is 0 occurrence. Write a recursive function `RecSearch(A,key,start)` which requires three parameters, the list `A`, the key to be found, and the start index to search in the list. This function returns the number of occurrences of this key in the list.

```python
def RecSearch(A, key, start):
  if start >= len(A):
    return 0
  elif A[start] == key:
    return 1 + RecSearch(A, key, start + 1)
  else:
    return RecSearch(A, key, start + 1)
    
# Test cases
list_1 = [2, 5, 6, 1, 2, 9, 2]
print(RecSearch(list_1, 2, 0))  # Output: 3
print(RecSearch(list_1, 5, 0))  # Output: 1
print(RecSearch(list_1, 7, 0))  # Output: 0
print(RecSearch(list_1, 9, 0))  # Output: 1
print(RecSearch(list_1, 1, 0))  # Output: 1
```

# Question 2

Write a recursive function `BinSearch(A, low, high, key)` which searches a sorted list **A, A[low] . . . . . A[high]**, for a match with `key` using binary search. Return the index of the matching item or `−1` if the key is not found.

```python
def BinSearch(A, key, low = 0, high = None):
  if high is None:
    high = len(A) - 1
    
  if low > high:
    return -1
  
  mid = (low + high) // 2
  
  if A[mid] == key:
    return mid
  
  elif A[mid] < key:
    return BinSearch(A, key, mid + 1)
  
  elif A[mid] > key:
    return BinSearch(A, key, low, mid - 1)
  
# Test cases
list = [1, 2, 5, 7, 11, 15, 20, 24, 35, 46]

print(BinSearch(list, 2))
print(BinSearch(list, 5)) 
print(BinSearch(list, 7)) 
print(BinSearch(list, 9)) 
print(BinSearch(list, 8)) 
print(BinSearch(list, 50))
  
```

# Question 3

Customers are identified by ID numbers. These ID numbers are to be stored in a hash table. The hashing function to be used is

`Address <-- IDnumber MOD Max`

The hash table is implemented as a one-dimensional array with elements indexed `0` to `(Max-1)`.

**Task 1**

Write program code to:

- Read ID numbers from a text file and store them in a hash table. For the purpose of testing the program, `Max` is to be set to the value 20. Assume different IDs will hash to different addresses (no collisions).
- Print out the contents of the hash table in the order in which the elements are stored in the array.

Use `KEYS1.TXT` to test your program code.

```python
keys1 = open("KEYS1.TXT", "r")
keys1_list = []
hash_list = []

for id in keys1:
  keys1_list.append(int(id.strip()))
  
for i in range(len(keys1_list)):
  hash_list.append(0)
  
def hash(id):
  return id % len(keys1_list)
  
for id in keys1_list:
  hash_list[hash(id)] = id
  # print(hash_list)
  
print(hash_list)
```

**Task 2**

Amend your program code so that collisions can be managed using open hashing. This means a collision is resolved by searching sequentially from the hashed address for an empty location and storing the ID at this empty location.

Use `KEYS2.TXT` to test your program code.

```python
keys2 = open("KEYS2.TXT", "r")
keys2_list = []
hash_list = []

for id in keys2:
  keys2_list.append(int(id.strip()))
  
for i in range(len(keys2_list)):
  hash_list.append(0)
  
def hash(id):
  return id % len(keys2_list)
  
for id in keys2_list:
  hash_key = hash(id)
  if hash_list[hash_key] == 0:
    hash_list[hash_key] = id
  else:
    while hash_list[hash_key] != 0:
      hash_key = (hash_key + 1) % len(keys2_list)
    hash_list[hash_key] = id
  # print(hash_list)
  
print(hash_list)
```
