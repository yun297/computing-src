# Question 2

Below is a recursive algorithm for finding a value, SearchItem, in an ordered array, X.

```python
Search(Low, High)
Mid = (Low+High) div 2
If X(Mid) = SearchItem then output “Found” : exit
If X(Mid) > SearchItem then Search(Low, Mid-1)
Else Search(Mid+1, High)
End Search
```

Note: the div operation returns an integer value after division e.g. 7 div 2 = 3

---

Using the above algorithm:

#### (a) Explain what is meant by a recursive algorithm.

A **recursive algorithm** is an algorithm that solves a problem by breaking it down into smaller instances of the same problem and calling itself repeatedly until it reaches a base/terminal case. In each recursive call, the problem size is reduced, ensuring that the recursion eventually terminates.  

In the given algorithm, `Search(Low, High)` follows a recursive approach to find `SearchItem` in the ordered array `X`. The function calls itself with a reduced range, either `Search(Low, Mid-1)` or `Search(Mid+1, High)`, depending on whether `SearchItem` is smaller or larger than `X(Mid)`. The recursion continues until either `SearchItem` is found (base case) or the search range becomes invalid (termination condition).

#### (b) Describe what might occur during execution with an incorrectly written recursive routine.

If a recursive algorithm is incorrectly written and does not reach its base case, it will continue calling itself indefinitely, leading to **infinite recursion**. In Python, this results in a **RecursionError**, a type of **RuntimeError**, when the maximum recursion depth is exceeded.  

In the given algorithm, if `SearchItem` is not in the array `X`, and there is no condition to stop the recursion when `Low > High`, the function will keep calling itself with `Low = Mid+1` or `High = Mid-1` until it exceeds the recursion limit.

#### (e) For this method of searching, state the maximum number of comparisons for array X, justify your answers.

##### **Worst-Case Time Complexity**
The worst-case time complexity of the given recursive **binary search** algorithm is:  
$$ O(\log N) $$

##### **Justification:**
1. **Reduction by Half:** Each recursive call divides the search space by half, reducing the problem size from $N$ to $\frac{N}{2}$, then $\frac{N}{4}$, and so on.  
2. **Number of Recursive Calls:** The recursion continues until the search space is reduced to a single element, which takes at most $\log_2 N$ recursive calls.  
3. **Comparison Per Call:** In each recursive call, one comparison is made to check whether `X(Mid) = \text{SearchItem}`.  

Since the number of comparisons grows logarithmically with the size of the array, the worst-case time complexity is $O(\log N)$.

##### **Worst-Case Scenario:**
- The worst case occurs when `SearchItem` is not in the array or is at one of the extreme ends, requiring the maximum number of recursive calls.  
- The algorithm performs $O(\log N)$ comparisons before concluding that `SearchItem` is absent.  

# Question 3

Using a hash table with eleven locations and the hashing function `h(i) = i % 11`, show the hash table that results when the following integers are inserted in the order given: `26, 42, 5, 44, 92, 59, 40, 36, 12, 60, 80`. Assume that collisions are resolved using

#### (a) linear probing

````python
hash_arr = []

arr = [26, 42, 5, 44, 92, 59, 40, 36, 12, 60, 80]

def h(i):
  return i % 11

for i in range(11):
  hash_arr.append([0, 0])
  
for num in arr:
  hash_key = h(num)
  while hash_arr[hash_key][0] != 0:
    hash_key = (hash_key + 1) % 11
  hash_arr[hash_key] = [hash_key, num]

print(hash_arr)

for i in range(len(hash_arr)):
  print(hash_arr[i])
````

#### (b) chaining

````python
hash_arr_c = []

arr = [26, 42, 5, 44, 92, 59, 40, 36, 12, 60, 80]

def h(i):
  return i % 11

for i in range(11):
  hash_arr_c.append([0, 0])
  
for num in arr:
  hash_key = h(num)
  if hash_arr_c[hash_key][0] == 0:
    hash_arr_c[hash_key] = [[hash_key, num]]
  else:
    hash_arr_c[hash_key].append([hash_key, num])
    
print(hash_arr_c)

for i in range(len(hash_arr_c)):
  print(hash_arr_c[i])
````
