### 1. Binary Search Tree

```python
class TreeNode:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left = left
        self.right = right
    

# BST with search, insert and display with different orders
class Tree:
    def __init__(self, root=None):
        self.root = root

    def search_itr(self, target):
        if not self.root: #empty tree
            return False
        
        curr = self.root
        
        while curr is not None:
            if curr.data == target:
                return True

            elif target < curr.data:
                curr = curr.left

            else:
                curr = curr.right

        return False

    def search_rec(self, root, target):
        if not root:
            return False

        if root.data == target: 
            return True

        elif target < root.data:
            return self.search_rec(root.left, target) 
        
        else:
            return self.search_rec(root.right, target)

    def insert_rec(self, root, newNode):
        if not root:
            self.root = newNode #initially, tree is empty
            return

        if newNode.data < root.data: #newNode smaller, to traverse to left of tree
            if not root.left:
                root.left = newNode
            
            else:
                self.insert_rec(root.left, newNode)
            
        else:
            if not root.right:
                root.right = newNode

            else:
                self.insert_rec(root.right, newNode)

        
    def insert_itr(self, newNode):
        if not self.root:
            self.root = newNode
            return

        curr = self.root

        while curr is not None:
            if newNode.data < curr.data:
                if not curr.left:
                    curr.left = newNode
                    return

                else:
                    curr = curr.left

            else:
                if not curr.right:
                    curr.right = newNode
                    return

                else:
                    curr = curr.right

    def inorder(self, root):
        if root:
            self.inorder(root.left)
            print(root.data)
            self.inorder(root.right)

    def preorder(self, root):
        if root:
            print(root.data)
            self.preorder(root.left)
            self.preorder(root.right)

    def postorder(self, root):
        if root:
            self.postorder(root.left)
            self.postorder(root.right)
            print(root.data)
```

```python
bst2 = Tree()
bst2.insert_itr(TreeNode(2))
bst2.insert_itr(TreeNode(3))
bst2.insert_itr(TreeNode(5))
bst2.insert_itr(TreeNode(6))
bst2.insert_itr(TreeNode(1))

bst2.inorder(bst2.root)
print("")
bst2.preorder(bst2.root)
print("")
bst2.postorder(bst2.root)

print(bst2.search_itr(3))
print(bst2.search_rec(bst2.root, 3))
print(bst2.search_itr(7))
print(bst2.search_rec(bst2.root, 7))
```

### 2. Linear Search

```python
def lin_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return True
        
    return False
```

### 3. Binary Search

```python
def bin_search_itr(arr, target):
    left = 0
    right = len(arr) - 1
    mid = (left + right) // 2
    
    while right >= left:
        if arr[mid] == target:
            return True
        
        elif target < arr[mid]:
            right = mid - 1

        else:
            left = mid + 1

    return False 

def bin_search_rec(arr, target, right, left=0):
    mid = (right + left) // 2

    if arr[mid] == target:
        return True
    
    elif target < arr[mid]:
        return bin_search_rec(arr, target, mid - 1, left)
    
    else:
        return bin_search_rec(arr, target, right, mid + 1)
```

### 5. Bubble Sort

```python
def bubble_sort(arr):
    length = len(arr)
    last_swap = length
    
    while last_swap >0:
        check = last_swap
        last_swap = 0
    
        for i in range(check-1):
            if arr[i] > arr[i+1]: # compare
                arr[i], arr[i+1] = arr[i+1], arr[i]
                last_swap = i+1
```

```python
lst = [1, 5, 3, 7, 2, 4]
bubble_sort(lst)
lst
```

### 6. Insertion Sort

```python
def insertion_sort(arr):
    length = len(arr)

    for i in range(length):
        temp = arr[i]
        space = i

        while (space > 0) and arr[space-1] > temp:
            arr[space] = arr[space-1]
            space -= 1

        arr[space] = temp
```

```python
lst = [1, 5, 3, 7, 2, 4]
insertion_sort(lst)
lst
```

### 7. Quick Sort

```python
def quick(arr, left, right):
    if left < right:
        
        pivot = arr[left]
        space = left

        for i in range(left+1, right+1):
            if arr[i] < pivot:
                space += 1
                arr[i], arr[space] = arr[space], arr[i]

        arr[space], arr[left] = pivot, arr[space]

        quick(arr, left, space -1)
        quick(arr, space+1, right)


    

def quick_sort(arr):
    left = 0
    right = len(arr) - 1
    quick(arr, left, right)
```

```python
lst = [1, 5, 3, 7, 2, 4]
quick_sort(lst)
lst
```

### 8. Merge Sort

```python
def merge_sort(arr):
    length = len(arr)
    
    if length <= 1:
        return
    
    mid = length // 2
    left = arr[:mid]
    right = arr[mid:]

    merge_sort(left)
    merge_sort(right)

    l = r = i = 0

    while len(left) > l and len(right) > r:
        if left[l] > right[r]:
            arr[i] = right[r]
            r += 1
        else:
            arr[i] = left[l]
            l += 1
        
        i += 1

    while len(left) > l:
        arr[i] = left[l]
        i += 1
        l += 1

    while len(right) > r:
        arr[i] = right[r]
        i += 1
        r += 1
```

```python
l1 = [16, 23, 44, 37, 2, 1]
merge_sort(l1)
l1
```
