```python
# binary search

from random import choices
randarr = choices(list(range(20)), k = 10)
randarr.sort()

def binary(arr, target):
  low = 0
  high = len(arr) - 1
  
  for i in range(len(arr)):
    mid = (low + high) // 2
    
    if arr[mid] == target:
      return mid
    elif arr[mid] > target:
      high = mid - 1
    else: 
      low = mid + 1
      
print(randarr, binary(randarr, 2))
```

```python
# binary search recur

from random import choices
randarr = choices(list(range(20)), k = 10)
randarr.sort()

def binary(arr, target, low = 0, high = None):
  if high is None:
    high = len(arr) - 1
  
  if low >= high:
    return
  
  mid = (high + low) // 2
  
  if arr[mid] == target:
    return mid
  elif arr[mid] > target:
    return binary(arr, target, low, mid - 1)
  else:
    return binary(arr, target, mid + 1, high)
      
print(randarr, binary(randarr, 2))
```

```python
# bubble sort

from random import choices
randarr = choices(list(range(20)), k = 10)

print(randarr)

def bubble_helper(arr):
  for i in range(len(arr)):
    for j in range(1, len(arr)):
      if arr[j] < arr[j - 1]:
        arr[j], arr[j - 1] = arr[j - 1], arr[j]
        
def bubble(arr):
  bubble_helper(arr)
  
  return arr

print(bubble(randarr))
```

```python
# bubble sort

from random import choices
randarr = choices(list(range(20)), k = 10)

print(randarr)

def bubble_helper(arr):
  length = len(arr)
  last = length
  
  while last > 0:
    check = last
    last = 0
    
    for j in range(1, check):
      if arr[j] < arr[j - 1]:
        arr[j], arr[j - 1] = arr[j - 1], arr[j]
        last = j + 1
        
def bubble(arr):
  bubble_helper(arr)
  
  return arr

print(bubble(randarr))
```

```python
# insertion sort

from random import choices
randarr = choices(list(range(20)), k = 10)

def insertion_helper(arr):
  for i in range(len(arr)):
    space = i
    temp = arr[i]
    
    while space < 0 and arr[space - 1] > temp:
      arr[space] = arr[space - 1]
      space -= 1
    
    arr[space] = temp
    
```

```python
# quick sort

from random import choices
randarr = choices(list(range(20)), k = 10)

print(randarr)

def quick_helper(arr, left, right):
  if left < right:
    pivot = arr[left]
    space = left
    
    for i in range(left + 1, right + 1):
      if arr[i] < pivot:
        space += 1
        arr[space], arr[i] = arr[i], arr[space]
      
    arr[space], arr[left] = arr[left], arr[space]
    
    quick_helper(arr, left, space - 1)
    quick_helper(arr, space + 1, right)
    
def quick(arr):
    quick_helper(arr, 0, len(arr) - 1)

    return arr
 
print(quick(randarr))
```

```python
from random import choices
randarr = choices(list(range(20)), k = 10)

print(randarr)

def quick(arr):
  if len(arr) <= 1:
    return arr
  
  pivot = arr[0]
  left = []
  right = []
  
  for item in arr[1:]:
    if item <= pivot:
      left.append(item)
    else:
      right.append(item)
      
  left = quick(left)
  right = quick(right)
  
  return left + [pivot] + right

print(quick(randarr))
```
