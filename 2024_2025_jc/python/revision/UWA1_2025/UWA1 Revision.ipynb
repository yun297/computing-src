```python
# linear search

from random import choices
randarr = choices(list(range(20)), k = 10)

def linear(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return None

print(randarr, linear(randarr, 5))
```

```python
# binary search (iterative)

from random import choices

randarr = choices(list(range(20)), k = 10)
randarr.sort()

def binary(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            high = mid - 1
        elif arr[mid] < target:
            low = mid + 1 

    return None

print(randarr, binary(randarr, 5))
```

```python
# binary search (recursive)

from random import choices

randarr = choices(list(range(20)), k = 10)
randarr.sort()

def binaryR(arr, target, low = 0, high = None):
    if high is None:
        high = len(arr) - 1

    if low >= high:
        return None

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binaryR(arr, target, low, mid - 1)
    elif arr[mid] < target:
        return binaryR(arr, target, mid + 1, high)

print(randarr, binaryR(randarr, 5))
```

```python
# bubble sort

from random import choices
randarr = choices(list(range(20)), k = 10)

def bubble(og_arr):
    arr = og_arr.copy()

    for i in range(len(arr)):
        for j in range(1, len(arr)):
            if arr[j] < arr[j - 1]:
                arr[j], arr[j - 1] = arr[j - 1], arr[j]

    return arr

print(randarr, bubble(randarr))
```

```python
# bubble sort 3

from random import choices
randarr = choices(list(range(20)), k = 10)

def bubble(og_arr):
    arr = og_arr.copy()

    last_swap = len(arr)

    while last_swap > 0:
        check = last_swap
        last_swap = 0
        for i in range(len(arr)):
            for j in range(1, check):
                if arr[j] < arr[j - 1]:
                    arr[j], arr[j - 1] = arr[j - 1], arr[j]
                    last_swap = j + 1

    return arr

print(randarr, bubble(randarr))
```

```python
# insertion sort

from random import choices
randarr = choices(list(range(20)), k = 10)

def insertion(og_arr):
    arr = og_arr.copy()

    for i in range(1, len(arr)):
        curr_i = i
        space = arr[i]

        while arr[curr_i - 1] > space and curr_i != 0:
            arr[curr_i], arr[curr_i - 1] = arr[curr_i - 1], arr[curr_i]
            curr_i -= 1

    return arr

print(randarr, insertion(randarr))
```

```python
# quick sort

from random import choices
randarr = choices(list(range(20)), k = 10)

def quick(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    right = []

    for item in arr[1:]:
        if item >= pivot:
            right.append(item)
        elif item < pivot:
            left.append(item)

    left = quick(left)
    right = quick(right)

    return left + [pivot] + right

print(randarr, quick(randarr))
```

```python
# quick sort 2

from random import choices
randarr = choices(list(range(20)), k = 10)

def quick_helper(arr, left, right):
    
    if left < right:
        pivot = arr[left]
        space = left

        for i in range(left + 1, right + 1):
            if arr[i] < pivot:
                space += 1
                arr[i], arr[space] = arr[space], arr[i]

        arr[left], arr[space] = arr[space], arr[left]

        quick_helper(arr, left, space - 1)
        quick_helper(arr, space + 1, right)

def quick(arr):
    quick_helper(arr, 0, len(arr) - 1)

    return arr

 
print(randarr, quick(randarr))
```
