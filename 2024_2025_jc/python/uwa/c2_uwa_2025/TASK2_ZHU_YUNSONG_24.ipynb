```python
class Array:
    def __init__(self, length: int):
        self.length = length
        self.array = [0] * length

    def size(self) -> int:
        return self.length

    def get(self, index: int):
        return self.array[index]

    def set(self, data, index: int):
        self.array[index] = data
```

```python
class BST(Array):
    def __init__(self, length: int):
        super().__init__(length)
        self.node_count = 0

    def count(self) -> int:
        return self.node_count  

    def insert(self, data):
        if self.node_count == 0:
            self.array[0] = data
            self.node_count += 1
            return
        
        i = 0 
        while i < self.length:
            if self.array[i] == 0:
                self.array[i] = data
                self.node_count += 1
                return
            
            if data < self.array[i]:  # Go left
                i = 2 * i + 1
            else:  # Go right
                i = 2 * i + 2
    
    def in_order(self, i = 0):
        if i >= self.length or self.array[i] == 0:
            return []

        result = []
        left_child = 2 * i + 1
        right_child = 2 * i + 2
        
        result += self.in_order(left_child)
        result.append(self.array[i])
        result += self.in_order(right_child)
        
        return result

    def double_copy(self):
        in_order_values = self.in_order()

        new_bst = BST(self.length)

        for value in in_order_values:
            new_bst.insert(value * 2)

        return new_bst
```
