```python
# chloe

import socket

client_socket = socket.socket()
client_socket.connect(("127.0.0.1", 12345))

print("Connected")

nugget_count = int(client_socket.recv(1024).decode().strip("\n"))
print(f"There are {nugget_count} nuggets left.")

while True:
    take_nuggets = -1
    while take_nuggets not in [1, 2, 3] or take_nuggets > nugget_count:  # CHANGED
        take_nuggets = int(input("How many to take? "))

    nugget_count -= take_nuggets
    print(f"{take_nuggets} nuggets taken, {nugget_count} left.")

    # if chloe takes the last nugget, she loses
    if nugget_count == 0:  # NEW
        client_socket.sendall("end\n".encode())
        print("You lose.")
        break

    message = f"There are {nugget_count} nuggets left.\n"
    client_socket.sendall(message.encode())

    data = client_socket.recv(1024)
    decoded = data.decode().strip("\n")

    # if sarah sends end, means sarah took the last nugget, so chloe wins
    if decoded == "end":  # NEW
        print("You win.")
        break

    nugget_count = int(decoded.split(" ")[2])
    print(f"There are {nugget_count} nuggets left.")  # NEW
```
