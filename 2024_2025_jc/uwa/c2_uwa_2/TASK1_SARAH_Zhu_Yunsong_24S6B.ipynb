```python
import socket

server_socket = socket.socket()
server_socket.bind(("127.0.0.1", 12345))
server_socket.listen()

new_socket, addr = server_socket.accept()
print("Connected to client on:", addr)

nugget_count = int(input("How many nuggets? "))
message = str(nugget_count) + "\n"
new_socket.sendall(message.encode())

while True:
    data = new_socket.recv(1024)
    decoded = data.decode().strip("\n")
    nugget_count = int(decoded.split(" ")[2])
    
    take_nuggets = -1
    while take_nuggets not in [1, 2, 3] and take_nuggets > nugget_count:
        print(decoded)
        take_nuggets = int(input("How many to take? "))
    nugget_count -= take_nuggets
    print(f"{take_nuggets} nuggets taken, {nugget_count} left.")

    message = f"There are {nugget_count} nuggets left.\n"  
    new_socket.sendall(message.encode())
```
