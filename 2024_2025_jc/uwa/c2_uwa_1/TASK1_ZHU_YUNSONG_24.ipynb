```python
def task1_1(filename):
    file = open(filename, "r")
    list_of_strings = []

    for line in file:
        list_of_strings.append(line.strip())

    return list_of_strings

names = task1_1('namelist.txt')
print(names[ :10])
print(len(names))
        
```

```python
def task1_2(list_of_names):
    list_of_tuples = []
    
    for i in range(len(list_of_names)):
        list_of_tuples.append((list_of_names[i], i // 10 + 1))

    return list_of_tuples

print(task1_2(names))
```

```python
def quick(arr, left, right):
    if left < right:
        pivot_index = left
        pivot = arr[pivot_index]
        space = left

        for i in range(left + 1, right + 1):
            if arr[i][0] < pivot[0]:
                space += 1
                arr[i], arr[space] = arr[space], arr[i]

        arr[left], arr[space] = arr[space], arr[left]

        quick(arr, left, space - 1)
        quick(arr, space + 1, right)

def task1_3(list_of_tuples):
    quick(list_of_tuples, 0, len(list_of_tuples) - 1)

    return list_of_tuples

list_of_tuples = task1_2(names)

print(task1_3(list_of_tuples))        
```

```python
def binary(arr, target, low=0, high=None):
    if high is None:
        high = len(arr) - 1

    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid][0] == target:
        return arr[mid][1]
    elif arr[mid][0] > target:
        return binary(arr, target, low, mid - 1)
    else:
        return binary(arr, target, mid + 1, high)

def task1_4(sorted_list, name):
    return binary(sorted_list, name)

# list_of_tuples = task1_2(names)
# sorted_list_of_tuples = task1_3(list_of_tuples)

# print(task1_4(sorted_list_of_tuples, "REGINA OWENS"))
```

```python
def task1_5(filename_in, filename_out):
    names = task1_1(filename_in)

    names_table_allocated = task1_2(names)

    names_sorted = task1_3(names_table_allocated)
    
    file_out = open(filename_out, mode='w')
    
    file_out.write('name,table\n')
    
    for name, table in names_sorted:
        file_out.write(f"{name},{table}\n")
    
    file_out.close()

print(task1_5("namelist.txt", "table_allocation.csv"))
```
