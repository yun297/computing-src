```python
import socket

server = socket.socket()
server.bind(('127.0.0.1', 12345))
server.listen()

new_socket, addr = server.accept()
print(f"Connection established with {addr}")

while True:
    data = b''
    while b'\n' not in data:
        data += new_socket.recv(1024)
    
    if data.decode().strip().lower() == "exit":
        print("Connection closed by client.")
        break
    
    print(f"Received message: {data.decode().strip()}")

    message = input("Enter message to send: ")
    
    while message != "send" and message != "exit":
        new_socket.send(f"{message}\t".encode())
        message = input("Enter message to send: ")
    
    if message.lower() == "send":
        new_socket.send(b'\n')

    if message.lower() == 'exit':
        new_socket.send(b'exit\n')
        print("Closing connection.")
        break
```
