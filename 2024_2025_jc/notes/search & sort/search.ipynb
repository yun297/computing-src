### Linear Search

```python
from random import choices

rand_list = choices(range(100), k = 20)

print(rand_list)
```

```python
def linear_search(arr: list, key: int):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
        
linear_search(rand_list, 85)
```

### Binary Search (Iterative)

```python
from random import choices

rand_list = choices(range(100), k = 20)

rand_list.sort()

print(rand_list)
```

```python
def binary_search_iterative(arr: list, key: int):
    low = 0
    high = len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == key:
            return mid
        elif arr[mid] > key:
            high = mid - 1
        else:
            low = mid + 1
    
    return -1

binary_search_iterative(rand_list, 3)
```

### Binary Search (Recursive)

```python
from random import choices

rand_list = choices(range(100), k = 20)

rand_list.sort()

print(rand_list)
```

```python
def binary_search_recursive(arr: list, key: int):
    def binary_search_recursive_helper(arr: list, key: int, low: int, high: int):
        if low > high:
            return -1
        
        mid = (low + high) // 2
        
        if arr[mid] == key:
            return mid
        elif arr[mid] > key:
            return binary_search_recursive_helper(arr, key, low, mid - 1)
        else:
            return binary_search_recursive_helper(arr, key, mid + 1, high)
    
    return binary_search_recursive_helper(arr, key, 0, len(arr) - 1)
        
binary_search_recursive(rand_list, 1)
```
