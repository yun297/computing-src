### Bubble Sort

```python
from random import choices

rand_list = choices(range(100), k = 20)

print(rand_list)
```

```python
def bubble1(arr: list):
    new_arr = arr.copy()
    
    for i in range(len(new_arr) - 1):
        for j in range(len(new_arr) - 1):
            if new_arr[j] > new_arr[j + 1]:
                new_arr[j], new_arr[j + 1] = new_arr[j + 1], new_arr[j]
    
    return new_arr

print(bubble1(rand_list))
```

```python
def bubble2(arr: list):
    new_arr = arr.copy()
    
    for i in range(len(new_arr) - 1):
        for j in range(len(new_arr) - 1 - i):
            if new_arr[j] > new_arr[j + 1]:
                new_arr[j], new_arr[j + 1] = new_arr[j + 1], new_arr[j]
    
    return new_arr

print(bubble2(rand_list))
```

```python
def bubble3(arr: list):
    new_arr = arr.copy()
    
    last_swap = len(arr)
    
    while last_swap > 0:
        temp = last_swap
        last_swap = 0
        
        for i in range(temp - 1):
            if new_arr[i] > new_arr[i + 1]:
                new_arr[i], new_arr[i + 1] = new_arr[i + 1], new_arr[i]
                last_swap = i + 1
    
    return new_arr

print(bubble3(rand_list))
```

### Insertion Sort

```python
from random import choices

rand_list = choices(range(100), k = 20)

print(rand_list)
```

```python
def insertion(arr: list): 
    new_arr = arr.copy()
    
    for i in range(1, len(new_arr)):
        compare = new_arr[i]
        space = i
        
        while space > 0 and arr[space - 1] > compare:
            arr[space] = arr[space - 1]
            space -= 1
        
        arr[space] = compare
    
    return arr

print(insertion(rand_list))
```

### Quick Sort

```python
from random import choices

rand_list = choices(range(100), k = 20)

print(rand_list)
```

```python
def quick(arr: list): # not in-place version
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    right = []

    for item in arr[1:]:
        if item > pivot:
            right.append(item)
        else:
            left.append(item)

    return quick(left) + [pivot] + quick(right)

print(quick(rand_list))
```

```python
def quick2(arr: list):  # "functional" style, but updates in place
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    right = []

    for item in arr[1:]:
        if item > pivot:
            right.append(item)
        else:
            left.append(item)

    # get sorted result
    sorted_arr = quick2(left) + [pivot] + quick2(right)

    # overwrite arr with sorted_arr (important part)
    arr[:] = sorted_arr
    return arr

print(rand_list)
quick2(rand_list)
print(rand_list)
```

```python
def quick3(arr): # in-place
    def quick_helper(arr, left, right):
        if left < right: # at least 2 elements
            pivot = arr[left]
            space = left
            for i in range(left + 1, right + 1): # from left + 1 until right
                if arr[i] < pivot:
                    space += 1
                    # space and i increment together until i finds a element larger than pivot, then space lag behind
                    # when we reach the next pass where i is less than the pivot, we swap the elements at i and space, as space is the last element that is less than the pivot
                    
                    arr[i], arr[space] = arr[space], arr[i]
            
            arr[left], arr[space] = arr[space], arr[left]

            quick_helper(arr, left, space - 1)
            quick_helper(arr, space + 1, right)
        
    quick_helper(arr, 0, len(arr) - 1)
    
print(rand_list)
quick3(rand_list)
print(rand_list)
```

### Merge Sort

```python
from random import choices

rand_list = choices(range(100), k = 20)

print(rand_list)
```

```python
def merge1(arr: list): # in place version
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    
    left = merge1(arr[:mid])
    right = merge1(arr[mid:])
    
    l = r = i = 0
    
    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            arr[i] = left[l]
            l += 1
        else:
            arr[i] = right[r]
            r += 1
        i += 1
    
    while l < len(left):
        arr[i] = left[l]
        i += 1
        l += 1
        
    while r < len(right):
        arr[i] = right[r]
        i += 1
        r += 1
    
    return arr

print(merge1(rand_list))
```

```python
def merge2(arr: list): # not in place version
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    
    left = merge2(arr[:mid])
    right = merge2(arr[mid:])

    result = []
    l = r = 0

    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            result.append(left[l])
            l += 1
        else:
            result.append(right[r])
            r += 1

    while l < len(left):
        result.append(left[l])
        l += 1

    while r < len(right):
        result.append(right[r])
        r += 1

    return result

print(merge2(rand_list))
```
