### Node Class

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

### Stack Class

```python
class Stack:
    def __init__(self):
        self.root = None
    
    def push(self, data):
        new_node = Node(data)
        
        if not self.root:
            self.root = new_node
            return

        new_node.next = self.root
        self.root = new_node
    
    def pop(self):
        if self.root:
            removed = self.root.data
            
            self.root = self.root.next
            
            return removed
```
