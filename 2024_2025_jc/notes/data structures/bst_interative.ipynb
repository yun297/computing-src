### Node Class

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

### Tree Class

```python
class Tree:
    def __init__(self):
        self.root = None
    
    def insert(self, data):
        new_node = Node(data)
        
        if not self.root:
            self.root = new_node
            return

        ptr = self.root
        inserted = False
        
        while not inserted:
            if data > ptr.data:
                if not ptr.right:
                    ptr.right = new_node
                    inserted = True
                else:
                    ptr = ptr.right
            else:
                if not ptr.left:
                    ptr.left = new_node
                    inserted = True
                else:
                    ptr = ptr.left
    
    def search(self, target):
        ptr = self.root
        
        while ptr:
            if target == ptr.data:
                return True
            
            if target > ptr.data:
                ptr = ptr.right
            
            else:
                ptr = ptr.left
                
        return False              
```
