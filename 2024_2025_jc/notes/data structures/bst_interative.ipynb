### Node Class

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

### Tree Class

```python
class Tree:
    def __init__(self):
        self.root = None
    
    def insert(self, data):
        new_node = Node(data)
        
        if not self.root:
            self.root = new_node
            return

        ptr = self.root
        inserted = False
        
        while not inserted:
            if data > ptr.data: # insert on the right
                if not ptr.right: # space avaliable on right
                    ptr.right = new_node
                    inserted = True
                else:
                    ptr = ptr.right
            else: # insert on the left
                if not ptr.left: # space avaliable on the left
                    ptr.left = new_node
                    inserted = True
                else:
                    ptr = ptr.left
    
    def search(self, target):
        ptr = self.root
        
        while ptr:
            if target == ptr.data:
                return True
            
            elif target > ptr.data:
                ptr = ptr.right
            
            elif target < ptr.data:
                ptr = ptr.left
                
        return False

    def inorder(self):
        stack = []
        ptr = self.root
        result = []

        while ptr or stack:
            # go as far left as possible
            while ptr:
                stack.append(ptr)
                ptr = ptr.left

            # visit the node
            ptr = stack.pop()
            result.append(ptr.data)

            # go right
            ptr = ptr.right

        return result

    def preorder(self):
        if not self.root:
            return []

        stack = [self.root]
        result = []

        while stack:
            node = stack.pop()
            result.append(node.data)

            # push right first, then left (so left is processed first)
            if node.right:
                stack.append(node.right)
            if node.left:
                stack.append(node.left)

        return result
    
    def postorder(self):
        if not self.root:
            return []

        stack = [(self.root, False)]
        result = []

        while stack:
            node, visited = stack.pop()

            if visited:
                result.append(node.data)
            else:
                # Postorder: push node back onto stack with visited=True
                stack.append((node, True))
                # Push right and left children onto stack
                if node.right:
                    stack.append((node.right, False))
                if node.left:
                    stack.append((node.left, False))

        return result
```
