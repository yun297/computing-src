### Node Class

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

### Tree Class

```python
class Tree:
    def __init__(self):
        self.root = None
        
    def insert(self, data):
        def insert_helper(root, data):
            if not root:
                return Node(data)
            
            if data < root.data:
                root.left = insert_helper(root.left, data)
            else:
                root.right = insert_helper(root.right, data)
            
            return root
    
        self.root = insert_helper(self.root, data)

    def insert2(self, data):
        if not self.root:
            self.root = Node(data)
            return

        def insert_helper(data, root):
            if data > root.data:
                if root.right:
                    insert_helper(data, root.right)  # just recurse
                else:
                    root.right = Node(data)          # attach in place
            else:
                if root.left:
                    insert_helper(data, root.left)
                else:
                    root.left = Node(data)

        insert_helper(data, self.root)

    def search(self, target):
        def search_helper(root, target):
            if not root:
                return False
            
            if target == root.data:
                return True
            elif target < root.data:
                return search_helper(root.left, target)
            else:
                return search_helper(root.right, target)
            
        return search_helper(self.root, target)
    
    def preorder(self, root):
        if root:
            print(root.data)
            self.preorder(root.left)
            self.preorder(root.right)
    
    def inorder(self, root):
        if root:
            self.inorder(root.left)
            print(root.data)
            self.inorder(root.right)
        
    def postorder(self, root):
        if root:
            self.postorder(root.left)
            self.postorder(root.right)
            print(root.data)
            
    def count(self, root):
        if not root:
            return 0
        
        return self.count(root.left) + 1 + self.count(root.right) # + 1 to count the root of the subtree
    
    # preorder, inorder, postorder and count functions can be wrapped in helper functions to remove the need to call them with root each time
    
    def preorder2(self):
        def preorder_helper(root):
            if root:
                print(root.data)
                preorder_helper(root.left)
                preorder_helper(root.right)
                
        preorder_helper(self.root)

    def inorder2(self):
        def inorder_helper(root):
            if root:
                inorder_helper(root.left)
                print(root.data)
                inorder_helper(root.right)
                
        inorder_helper(self.root)

    def postorder2(self):
        def postorder_helper(root):
            if root:
                postorder_helper(root.left)
                postorder_helper(root.right)
                print(root.data)
    
    def count2(self):
        def count_helper(root):
            if not root:
                return 0
            
            return count_helper(root.left) + count_helper(root.right) + 1

        count_helper(self.root)
```
