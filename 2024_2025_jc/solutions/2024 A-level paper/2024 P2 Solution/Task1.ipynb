```python
# task 1.1
from random import randint

# generate cards for one player
def generate():
    colours = ['red','green','blue']
    player = [] # list of cards 
    for i in range(10): 
        # each card is a list of color and number
        card = [colours[randint(0,2)], randint(1,4)]
        player.append(card)
    return player

# output contents
print('Player 1 cards')
player1 = generate()
print(player1)

print('Player 2 cards')
player2 = generate()
print(player2)
```

```python
# task 1.2
def compare_colour(colour1, colour2):
    if colour1 == colour2: # draw
        return '3'
    elif colour1 == 'red' and colour2 == 'green': # red wins green
        return '1'
    elif colour1 == 'green' and colour2 == 'blue': # green wins blue
        return '1'
    elif colour1 == 'blue' and colour2 == 'red': # blue wins red
        return '1'
    else:
        return '2'
```

```python
# task 1.3
def sort_card(player):
    '''when two cards have the same number, arrange colours 
    in ascending order of index in colours list'''
    colours = ['red', 'green', 'blue'] 
    
    # bubble sort
    for i in range(len(player)):
        for j in range(len(player) - i - 1):
            colour_j = colours.index(player[j][0]) # index of colour at card j
            colour_j_1 = colours.index(player[j + 1][0]) # index of colour at card j + 1
            
            if player[j][1] > player[j + 1][1]: # sort by number
                player[j], player[j + 1] = player[j + 1], player[j]
            elif player[j][1] == player[j + 1][1] and colour_j > colour_j_1:
                # sort by color if numbers are the same
                player[j], player[j + 1] = player[j + 1], player[j]
    return player
    

# output contents before and after sorting
print('Player 1 cards')
print(player1)
print('Player 1 cards after sorting')
sort_card(player1)
print(player1)
print()

print('Player 2 cards')
print(player2)
print('Player 2 cards after sorting')
sort_card(player2)
print(player2)
print()
```

```python
# task 1.4
def scoring(player1,player2):
    # scores for both players
    score1 = 0
    score2 = 0 
    
    for i in range(10):
        card1 = player1[i]
        card2 = player2[i]
        
        # check who wins
        win = compare_colour(card1[0],card2[0])
        
        # win: add number x 2; lose: add number
        # draw: add number for both
        if win == '1': # player 1 wins
            score1 += card1[1] * 2 
            score2 += card2[1]
        elif win == '2': # player 2 wins
            score1 += card1[1]
            score2 += card2[1] * 2
        
        else: # draw 
            score1 += card1[1]
            score2 += card2[1]
        
    return score1, score2 

# testing
score1, score2 = scoring(player1,player2)

print(f'Player 1 sorted hand: {player1}, score: {score1}')
print(f'Player 2 sorted hand: {player2}, score: {score2}')
```

```python
# task 1.5
def result(score1, score2):
    if score1 == score2: # draw
        print('Draw!')
    elif score1 > score2: # player1 wins
        print('Player 1 wins and Player 2 comes second!')
    else: # player2 wins
        print('Player 2 wins and Player 1 comes second!')
    
result(score1, score2)
```
