```python
# task 2.1

class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1 
    
    def set_data(self,data):
        self.data_value =  data
    
    def set_left(self,left):
        self.left_node = left
    
    def set_right(self,right):
        self.right_node = right 
    
    def get_data(self):
        return self.data_value

    def get_right(self):
        return self.right_node
    
    def get_left(self):
        return self.left_node 
    
    # output data and pointers    
    def __str__(self):
        return f'left: {self.get_left()} data: {self.get_data()} right: {self.get_right()}'
```

```python
# task 2.2
def calculate_nodes(levels):
    total_nodes = 0
    for i in range(levels):
        total_nodes += 2 ** i # at each level there are 2^i nodes
    print(f"A {levels} level number tree has {total_nodes} nodes")
    return total_nodes
```

```python
# task 2.3
# test 1, 3, 5, 10
calculate_nodes(1)
calculate_nodes(3)
calculate_nodes(5)
calculate_nodes(10)
```

```python
# task 2.4
def create_empty_tree(list_nodes,levels): 
    # find the number of nodes through the number of levels 
    length = calculate_nodes(levels)
    
    # creating the list of node objects with all attributes set to -1
    for i in range(length):
        list_nodes.append(Node())
    
    return list_nodes

# # testing
# list_nodes = create_empty_tree([], 3)
# for node in list_nodes:
#     print(node)
```

```python
# task 2.5
def assign_leaf_nodes(list_nodes,levels):
    # find the total number of leaf nodes
    total_leaf = 2 ** (levels - 1)
    
    for i in range(total_leaf):
        # index starts from 0, data value starts from 1
        list_nodes[i].set_data(i + 1)  

    return list_nodes 

# # testing
# list_nodes = create_empty_tree([], 3)
# list_nodes = assign_leaf_nodes(list_nodes, 3)
# for i in list_nodes:
#     print(i)
```

```python
# task 2.6

def create_number_tree(list_nodes, levels):
    # parent pointer to start with
    parent = 2 ** (levels - 1)
    
    for left in range(0, len(list_nodes) - 1, 2): # left pointer to trace each pair to add up to parent
        list_nodes[parent].set_left(left) # udpate left pointer
        list_nodes[parent].set_right(left + 1) # update right pointer
        
        # update parent data 
        parent_data = list_nodes[left].get_data() + list_nodes[left + 1].get_data()
        list_nodes[parent].set_data(parent_data)
        
        # move on to next parent node
        parent += 1
        
    return list_nodes

# # testing
# list_nodes = create_empty_tree([],3)
# list_nodes = assign_leaf_nodes(list_nodes,3)
# list_nodes = create_number_tree(list_nodes,3)
# for node in list_nodes:
#     print(node)
```

```python
# task 2.7

def tree(levels): # function to call everything and output content
    calculate_nodes(levels)
    list_nodes = create_empty_tree([], levels)
    list_nodes = assign_leaf_nodes(list_nodes, levels)
    list_nodes = create_number_tree(list_nodes, levels)

    print('Number tree for tree with {} levels:'.format(levels))
    for node in list_nodes:
        print(node)
        
# testing 1, 3, 5  
tree(1)
print()
tree(3)
print()
tree(5)
```
