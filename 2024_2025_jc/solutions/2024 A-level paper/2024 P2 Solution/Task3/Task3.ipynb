```python
# task 3.1

# write a function for each element type
def heading(text):
    return '<h1>{}</h1>'.format(text)

def subheading(text):
    return '<h2>{}</h2>'.format(text)

def paragraph(text):
    return '<p>{}</p>'.format(text)

# testing
text = 'My element'
print('heading:', end = ' ')
print(heading(text))

print('subheading:', end = ' ')
print(subheading(text))

print('paragraph:', end = ' ')
print(paragraph(text))
```

```python
#task 3.2

def hyperlink(text):
    # split text into URL and string
    URL, string = text.split(',')
    return f'<a href = "{URL}" >{string}</a>'

# testing
print('hyperlink:')
print(hyperlink('https://www.cambridgeinternational.org,Click here'))
```

```python
#task 3.3

def body(ls):
    string = '<html><body>'# insert HTML and body opening tags
    for i in ls:
        # extract element and text from each item in the list
        element, text = i
        if element == 'heading':
            string += heading(text) # add to string
        elif element == 'subheading':
            string += subheading(text)
        elif element == 'paragraph':
            string += paragraph(text)
        elif element == 'hyperlink': 
            string += hyperlink(text)
        else:
            return False # if there are unrecognised elements
    string += '</body></html>'# insert HTML and body closing tags
    return string
```

```python
# task 3.4

def read_data():
    filename = input('Enter textfile name (including extension):')
    ls = [] # list for body() function
    with open(filename,'r') as file: # file closes automatically
        # split each line of 'element: text' into a list
        for line in file:
            line = line.strip().split(': ') 
            ls.append(line)
    
    # return the formated string or output a error msg
    final = body(ls)
    if final == False: # if data cannot be converted
        print('Data cannot be converted to html')
    else:
        return final
    

def write_data():
    filename = input('Enter filename (excluding extension):')
    text = input('Enter formatted string:')
    
    # manually add file extension, file closes automatically 
    with open(filename + '.html','w') as file: 
        file.write('<!doctype html>\n') # inital string at the start of file
        file.write(text)
    print(f'Data written to {filename}')

# Test 1
print(read_data()) # print out the formated string for write_data to enter input
write_data()
print()

# Test 2
print(read_data())
write_data()
```
