```python
# Task 2.1
class QueueItem:
    def __init__(self, ID, quantity):
        self.ID = ID
        self.quantity = quantity
        
    def getID(self):
        return self.ID
    
    def getQuantity(self):
        return self.quantity
```

```python
# Task 2.2
class Queue:
    def __init__(self):
        self.queueList = [None]*4
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0
        
    def enqueue(self,QueueItem):
        if self.numberItems == 4: #queue is full
            return False
        else:
            self.queueList[self.tailPointer] = QueueItem #assign location at tail pointer to new item
            
            self.numberItems += 1 #increment number of items
            self.tailPointer = (self.tailPointer + 1) % 4
        
        # headPointer need not be adjusted, it just works
        return True #added successfully
    
    
    def dequeue(self):
        if self.numberItems == 0: #empty queue
            return False
        else:
            temp = self.queueList[self.headPointer] #item to be returned
            
            self.numberItems -= 1 #decrement number of items
            self.headPointer = (self.headPointer + 1) % 4
            
            return temp

        
    def printQueue(self):
        if self.numberItems == 0: # empty queue
            print('\nQueue is empty.')
        else:
            pointer = self.headPointer
            print('\nStart of Queue')
            print('ID\tQuantity')
            for i in range(self.numberItems):
                print(f'{self.queueList[pointer].getID()}\t{self.queueList[pointer].getQuantity()}')
                pointer = (pointer + 1) % 4
            print('End of Queue\n')
```

```python
# Task 2.3
q = Queue()
for i in range(5):
    ID, quantity = input("Enter id and quantity, comma seperated. (Eg. (3, 33)): ").split(",")
    qitem = QueueItem(ID, quantity)
    result = q.enqueue(qitem)
    if result:
        print(f"Item with id: {ID} and quantity: {quantity} has been successfully enqueued")
    else:
        print(f"Item with id: {ID} and quantity: {quantity} failed to be enqueued")
```

```python
# Task 2.4
for i in range(5):
    ret = q.dequeue()
    if not ret:
        print("Empty queue!")
    else:
        print(f"Item with id: {ret.getID()} and quantity: {ret.getQuantity()} has been successfully dequeued")
```
