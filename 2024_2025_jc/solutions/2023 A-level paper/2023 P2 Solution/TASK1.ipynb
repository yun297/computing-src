```python
# Task 1.1
import sqlite3

# create database
conn = sqlite3.connect("TESTDATABASE.db")

# ensure we are working with a clean database
conn.execute("""DROP TABLE IF EXISTS `Student`""")
conn.execute("""DROP TABLE IF EXISTS `Test`""")
conn.execute("""DROP TABLE IF EXISTS `Class`""")
conn.execute("""DROP TABLE IF EXISTS `Student_Test`""")
conn.commit()

# create tables
conn.execute("""
CREATE TABLE `Student` (
	`StudentID`	INTEGER NOT NULL UNIQUE,
	`GivenName`	TEXT NOT NULL,
	`FamilyName`	TEXT NOT NULL,
	`ClassName`	TEXT NOT NULL,
	PRIMARY KEY(`StudentID`),
	FOREIGN KEY(`ClassName`) REFERENCES `Class`(`ClassName`)
);
""")
conn.commit()
conn.execute("""
CREATE TABLE `Test` (
	`TestID`	TEXT NOT NULL UNIQUE,
	`MaxMarks`	INTEGER NOT NULL,
	PRIMARY KEY(`TestID`)
);
""")
conn.commit()
conn.execute("""
CREATE TABLE `Class` (
	`ClassName`	TEXT NOT NULL UNIQUE,
	`YearGroup`	INTEGER NOT NULL,
	PRIMARY KEY(`ClassName`)
);
""")
conn.commit()
conn.execute("""
CREATE TABLE `Student_Test` (
    `StudentTestID` INTEGER NOT NULL UNIQUE,
    `StudentID` INTEGER NOT NULL,
    `TestID` TEXT NOT NULL,
    `Mark` INTEGER NOT NULL,
    `DateTest` TEXT NOT NULL,
    FOREIGN KEY(`StudentID`) REFERENCES Student(`StudentID`),
    FOREIGN KEY(`TestID`) REFERENCES Test(`TestID`),
    PRIMARY KEY(`StudentTestID`)
)
""")
conn.commit()

# close the connection
conn.close()
```

```python
# Task 1.2
# insert info into TESTDATABASE.db
conn = sqlite3.connect("TESTDATABASE.db")

# because of dependency, we need to create Class before Student, and end with Student_Test (after Student & Test)

conn.execute("""INSERT INTO Class('ClassName', 'YearGroup')
                VALUES('CS3', 12)""")
conn.commit()

conn.execute("""INSERT INTO Student('StudentID', 'GivenName', 'FamilyName', 'ClassName')
                VALUES (102, 'Mary', 'Lim', 'CS3')""")
conn.commit()

conn.execute("""INSERT INTO Test(TestID, MaxMarks) VALUES('PG1', 100)""")
conn.commit()

conn.execute("""INSERT INTO Student_Test(StudentTestID, StudentID, TestID, Mark, DateTest)
                VALUES(123, 102, 'PG1', 85, '01-02-2023')""")
conn.commit()

conn.close()
```
