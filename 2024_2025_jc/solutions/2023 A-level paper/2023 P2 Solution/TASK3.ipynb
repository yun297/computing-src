```python
# Task 3.1
def validate_row(row):
    if len(row) != 17:
        return False
    rowarr = row.split(" ")
    if len(rowarr) != 9:
        return False
    for digit in rowarr:
        if digit not in list("123456789"):
            return False
    return True

def inputPuzzle():
    i = 0
    grid = []
    while i <= 8:
        row = input("Enter a row of 9 digits with a space between each digit: ")
        if not validate_row(row):
            print("invalid input, try again.")
        else:
            split = row.split(" ")
            new_row = []
            for x in split:
                new_row.append(int(x))
            grid.append(new_row)
            i += 1
    return grid

print("Please enter puzzle 1")
example1 = inputPuzzle()
print("Please enter puzzle 2")
example2 = inputPuzzle()
```

```python
# Task 3.2
def check_rows(puzzle):
    for row_num in range(len(puzzle)):
        tmp=[]
        for digit in puzzle[row_num]:
            if digit not in tmp:
                tmp.append(digit)
            else:
                return row_num + 1 # convert from 0-index to 1-indexing
    return -1

res = check_row(example1)
if res == -1:
    print("no duplicates in rows for the puzzle `example1`")
else:
    print(f"duplicate in row {res} for the puzzle `example1`")
res = check_row(example2)
if res == -1:
    print("no duplicates in rows for the puzzle `example2`")
else:
    print(f"duplicate in row {res} for the puzzle `example2`")
```

```python
# Task 3.3
def check_columns(puzzle):
    for col_num in range(len(puzzle)):
        tmp = []
        for row_num in range(len(puzzle)):
            digit = puzzle[row_num][col_num]
            if digit not in tmp:
                tmp.append(digit)
            else:
                return col_num + 1
    return -1
            
res = check_columns(example1)
if res == -1:
    print("no duplicates in columns for the puzzle `example1`")
else:
    print(f"duplicate in column {res} for the puzzle `example1`")
res = check_columns(example2)
if res == -1:
    print("no duplicates in columns for the puzzle `example2`")
else:
    print(f"duplicate in column {res} for the puzzle `example2`")
```

```python
# Task 3.4
def check_mini_squares(puzzle):
    for i in range(0, 9, 3):
        rows = puzzle[i:i+3]
        for j in range(0, 9, 3):
            # iterating through each mini square
            tmp = []
            for row in rows:
                for elem in row[j:j+3]:
                    if elem in tmp:
                        return False
                    else:
                        tmp.append(elem)
    return True

res = check_mini_squares(example1)
if res:
    print("no duplicates in mini squares for the puzzle `example1`")
else:
    print("duplicates found in mini squares for the puzzle `example1`")
res = check_mini_squares(example2)
if res:
    print("no duplicates in mini squares for the puzzle `example2`")
else:
    print("duplicates found in mini squares for the puzzle `example2`")
```
