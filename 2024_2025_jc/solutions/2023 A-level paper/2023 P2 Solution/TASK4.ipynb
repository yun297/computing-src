```python
# Task 4.1
def readfile(filename):
    f = open(filename, "r")
    content = f.readlines() # raw data
    # process the data
    cleaned_data = []
    for line in content:
        cleaned_data.append(line.strip())
    f.close()
    return cleaned_data

def compress(list_of_colours):
    prev = None
    ret = ""
    count = 0
    for colour in list_of_colours:
        if prev != colour: # different colour
            # append count for previous colour
            if count != 0:
                ret += str(count)
            # start counting for new colour
            count = 1
            prev = colour # update prev
            ret += colour # append colour code
        else: # same color encountered, apply compression
            count += 1
#         print(ret)
    # add the count of the last colour
    ret += str(count)
    return ret

def writefile(compressed_colours):
    f = open("compressedimage.txt", "w")
    f.write(compressed_colours)
    f.close()

colours = readfile("imagefile.txt")
compressed_colours = compress(colours)
writefile(compressed_colours)
```

```python
# Task 4.2
def readcompfile(filename):
    f = open(filename, "r")
    content = f.read()
    f.close()
    return content

def decompress(compimg):
    retstr = ""
    for i in range(0, len(compimg), 4):
        code, count = compimg[i:i+3], compimg[i+3]
        for _ in range(int(count)):
            retstr += code + "\n"
    retstr = retstr[:-1] # remove last "\n"
    return retstr

def writedecompfile(decompressed_img):
    f = open("decompressedimage.txt", "w")
    f.write(decompressed_img)
    f.close()

def menu():
    print("Hi user, this program will help compress and decompress image files")
    filename = input("Please enter filename: ")
    mode = input("Please enter mode of operation. (Enter '0' for compression, '1' for decompression): ")
    if mode == "0": # compression
        colours = readfile(filename)
        compressed_colours = compress(colours)
        writefile(compressed_colours)
        print("Your compressed file has been written to `compressedimage.txt`")
    elif mode == "1": # decompression
        compimg = readcompfile(filename)
        decompressed_img = decompress(compimg)
        writedecompfile(decompressed_img)
        print("Your decompressed file has been written to `decompressedimage.txt`")
    else:
        print("invalid mode")
        
menu()
```
