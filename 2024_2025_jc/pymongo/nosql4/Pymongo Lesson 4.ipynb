### Task 1.1

```python
from pymongo import MongoClient

client = MongoClient("localhost", 27017)
```

```python
library = client["library"]
books = library["books"]
```

```python
book_txt = [
  {"book_id": 234, "title": "Father Night", "author": "Kurt", "publisher": "APress", "page_count": 433, "year": "2018"},
  {"book_id": 134, "title": "Mother Night", "author": ["Kurt", "Dan"], "publisher": "APress", "year": "2015"},
  {"book_id": 334, "title": "Programming C## 6.0", "author": ["Andrew", "Dan"], "page_count": 300, "year": "2000"},
  {"book_id": 534, "title": "Introduction to Python", "publisher": "MPH", "year": "1999"},
  {"book_id": 434, "title": "Travel with Dogs", "author": "Andy", "publisher": "APress", "page_count": 100, "year": "2017"},
]

books.insert_many(book_txt)
```

```python
for book in books.find():
  print(book)
```

### Task 1.2

```python
query = {"year": "2015"}

for book in books.find(query):
  print(book)
```

```python
query = {"page_count": {"$lt": 400, "$gt": 100}}

for book in books.find(query):
  print(book)
```

```python
filter = {"page_count": {"$exists": False }}
update = {"$set": {"page_count": "Less Than 100 Pages"}}

books.update_many(
  filter,
  update
)

for book in books.find():
  print(book)
```

```python
projection = {"_id": 0, "title": 1, "page_count": 1, "year": 1}
sort = [("year", -1)]

for book in books.find({}, projection).sort(sort):
  print(book)
```
