```python
import socket
```

```python
passive_socket = socket.socket()
passive_socket.bind(("127.0.0.1", 12345))
passive_socket.listen()
```

```python
new_socket, addr = passive_socket.accept()
print("Connected to", str(addr))
```

```python
data = b""

while True:
  # sending
  message = str(input("Enter a message to send to the client: "))
  
  new_socket.sendall(message.encode())
  
  if message == "quit":
    print("Quitting program...")
    break
  
  # receiving
  data = new_socket.recv(1024)
  
  if data == b"quit":
    print("Client sent \"quit\", quitting program..." )
    break
  
  print(data) 
  
passive_socket.close()
new_socket.close()
```
