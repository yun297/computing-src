```python
import socket
```

```python
listen_socket = socket.socket()
listen_socket.bind(('127.0.0.1', 12345))

listen_socket.listen()
```

```python
new_socket, addr = listen_socket.accept()
print(f"Connection established at: {addr}")
```

```python
# REMEMBER TO RUNT HE CLIENT SIDE AS WELL BEFORE INPUTING ANYTHING INTO TERMINAL

while True:
    opt = int(input('Press 1 to text or 2 to quit'))

    if opt == 2:
        new_socket.sendall(b'QUIT\n') # Sends data to client
        print('Stopping system...')
        break 

    elif opt == 1:
        msg = input('Enter what you want to tell the client: ')
        data = b'TEXT' + msg.encode() + b'\n'
        new_socket.sendall(data) # Sends data to client

    received = b''
    while b'\n' not in received: # This is good practice
        received += new_socket.recv(1024) # Receiving from client

    if received[:4] == b'TEXT':
        print(received[4:-1].decode())

    elif received[:4] == b'QUIT':
        print('Client has disconnected. Shutting down...')
        break

# Always close any opened things (files, sockets etc) 
new_socket.close() 
listen_socket.close()
```
