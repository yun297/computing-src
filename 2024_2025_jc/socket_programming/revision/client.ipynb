```python
import socket
```

```python
# ALWAYS RUN THIS AFTER THE SERVER!
# This is trying to connect to the server, so it will return error if there is no server yet
client = socket.socket()
client.connect(('127.0.0.1', 12345))
```

```python
while True:
    received = b''
    while b'\n' not in received: # This is good practice
        received += client.recv(1024) # Receving from server


    if received[:4] == b'QUIT':
        print('Server closed. Bye!')
        break

    elif received[:4] == b'TEXT':
        print(received[4:-1].decode())

    opt = int(input('Press 1 to text or 2 to quit'))

    if opt == 2:
        client.sendall(b'QUIT\n') # Sends data to server
        print('Signing off...')
        break

    elif opt == 1:
        msg = input('Enter what you want to tell the server: ')
        data = b'TEXT' + msg.encode() + b'\n'
        client.sendall(data) # Sends data to server

# Always close any opened things (files, sockets etc) 
client.close()
```
