```python
import socket
from random import choice

### read words from file ###

words_txt = open("words.txt", "r")
words = []

for line in words_txt:
    words.append(line.strip())
    
words_txt.close()

# pick a random word
word = choice(words)
print("Correct word:", word)

count = 0

### set up server ###

listener = socket.socket()
listener.bind(("127.0.0.1", 12345))
listener.listen()

new_socket, addr = listener.accept()
print("Connected to", addr)

### game loop ###

new_socket.sendall(b'Guess a word of 5 distinct letters!\n')

while True:
  if count < 5:
    new_socket.sendall(b'Guess a letter in the word: \n')
    guess = new_socket.recv(1024).decode().strip()
    
    if guess in word:
      new_socket.sendall(b'YES\n')
    else:
      new_socket.sendall(b'NO\n')
      
    count += 1
    
  else:
    new_socket.sendall(b'Guess the word: \n')
    word_guess = new_socket.recv(1024).decode().strip()
    
    if word_guess == word:
      new_socket.sendall(b'You WIN')
    else:
      new_socket.sendall(f"You LOSE. The correct word is {word}".encode())
    
    break
```
