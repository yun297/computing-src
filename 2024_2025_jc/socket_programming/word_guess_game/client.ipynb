```python
import socket 

client = socket.socket()
client.connect(("127.0.0.1", 12345))

### game loop ###

while True:
  message = client.recv(1024).decode()
  print(message)
  
  if message == "Guess a letter in the word: \n":
    guess = input()
    client.sendall(f"{guess}\n".encode())
  
  if message == "Guess the word: \n":
    guess = input()
    client.sendall(f"{guess}\n".encode())
    
    message = client.recv(1024).decode()
    print(message)
    
    break
```
