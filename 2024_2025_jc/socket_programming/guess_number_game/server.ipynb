```python
import socket
from random import randint

### set up server ###

listener = socket.socket()
listener.bind(("127.0.0.1", 12345))
listener.listen()

new_socket, addr = listener.accept()
print("Connected to:", addr)

### game logic ###

answer = randint(1, 10)
print("Correct answer:", answer)

attempts = 0

### game loop ###

while attempts < 5:  # Limit to 5 attempts
    # send to client
    new_socket.sendall(b"GUESS\n")

    # receive from client
    data = new_socket.recv(1024)
    if not data:
        break # Exit if connection is lost

    try:
        guess = int(data.decode().strip("\n"))
    except ValueError:
        continue  # Ignore invalid input

    attempts += 1

    if guess == answer:
        new_socket.sendall(b"WIN\n")
        break
    elif guess > answer:
        new_socket.sendall(b"HIGH\n")
    else:
        new_socket.sendall(b"LOW\n")

if attempts >= 5:
    new_socket.sendall(b"LOSE\n")

new_socket.close()
listener.close()
```
