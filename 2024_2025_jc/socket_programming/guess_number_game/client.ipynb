```python
import socket

client_socket = socket.socket()
client_socket.connect(("127.0.0.1", 12345))

while True:
    # Wait for "GUESS" prompt
    data = client_socket.recv(1024)
    if not data:
        break # Exit if connection is lost

    response = data.decode().strip("\n")
    print(response)

    if response == "GUESS":
        guess = input("Enter a number: ").strip()
        client_socket.sendall(f"{guess}\n".encode())

    # Receive response after sending a guess
    data = client_socket.recv(1024)
    if not data:
        break

    response = data.decode().strip("\n")
    print(response)

    if response in ("WIN", "LOSE"):
        break

client_socket.close()
```
