### Connects to the database `Northwind_small.sqlite`

```python
import sqlite3

conn = sqlite3.connect("Northwind_small.sqlite")
cursor = conn.cursor()
```

### Insert into the table called `Customer` with these 5 columns:

* `CustomerID`
* `CompanyName`
* `ContactName`
* `Country`
* `Phone`

### Asks the user for the 5 inputs. Eg: `What is the Customer ID?`

```python
CustomerID = input("What is the CustomerID? ")
CompanyName = input ("What is the CompanyName? ")
ContactName = input ("What is the ContactName? ")
Country = input ("What is the Country? ")
Phone = input ("What is the Phone? ")
```

### Inserts the Customer into the table. Avoid using f-strings in SQL, which (should not be?) allowed in A-levels. The notes from CPDD (in cohort g-classroom) explains why we need to use `(?,?,?)`

```python
cursor.execute(
    """
    INSERT INTO Customer (ID, CompanyName, ContactName, Country, Phone)
    VALUES (?, ?, ?, ?, ?)
    """, (CustomerID, CompanyName, ContactName, Country, Phone)
)
```

```python
conn.commit()
```
