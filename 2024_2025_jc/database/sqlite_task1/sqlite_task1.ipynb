```python
import sqlite3

conn = sqlite3.connect("Northwind_small.sqlite")
cursor = conn.cursor()
```

### 1. Find Order Details
 Retrieve a list of all orders, including the `Order ID`, `Customer Name`, `Product Name`, and `Quantity Ordered`. Use an `INNER JOIN` to connect the `Orders`, `OrderDetails`, `Customers`, and `Products` tables.

```python
cursor.execute(
    """
    SELECT [Order].Id, [Customer].ContactName, [Product].ProductName, [OrderDetail].Quantity
    FROM [Order]
    INNER JOIN [Customer]
    ON [Order].CustomerId = [Customer].Id
    INNER JOIN [OrderDetail]
    ON [OrderDetail].OrderId = [Order].Id
    INNER JOIN [Product]
    ON [Product].Id = [OrderDetail].ProductId
    """
)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

### 2. Find Expensive Products
List all products that cost more than **$50**. Show the `Product Name`, `Unit Price`, and `Supplier Name`.

```python
cursor.execute(
    """
    SELECT [Product].ProductName, [Product].UnitPrice, [Supplier].CompanyName
    FROM [Product]
    INNER JOIN [Supplier]
    ON [Product].SupplierId = Supplier.Id
    WHERE [Product].UnitPrice > 50
    """
)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

### 3. Modify Product Prices
Due to inflation, all products in the "`Beverages`" category need a **10% price increase**. Update the `Products` table accordingly.

```python
cursor.execute(
    """
    SELECT p.ProductName, p.UnitPrice, c.CategoryName
    FROM [Product] AS p
    INNER JOIN [Category] AS c
    ON p.CategoryId = c.Id
    WHERE c.CategoryName = "Beverages"
    """
)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute(
    """
    UPDATE [Product]
    SET UnitPrice = UnitPrice * 1.1
    WHERE CategoryId = (
        SELECT Id
        FROM [Category]
        WHERE CategoryName = "Beverages"
    )
    """
)

conn.commit()
```

### 6. Find Top Customers by Order Count
Find the **top 5 customers** who have placed the **most orders**. Display the `Customer Name` and the `Number of Orders`, sorted in **descending order**.

```python
cursor.execute(
    """
    SELECT c.ContactName, COUNT(o.CustomerId)
    FROM Customer AS c
    INNER JOIN [Order] as o
    ON c.Id = o.CustomerId
    GROUP BY c.ContactName
    ORDER BY COUNT(o.CustomerId) DESC
    LIMIT 5
    """
)

rows = cursor.fetchall()

for row in rows:
    print(row)
```
